````python
# ============================================================
# Step 1: Requirement Gathering Checklist
# ============================================================

print("""
📋 REQUIREMENT GATHERING — QUESTIONS TO ALWAYS ASK

Functional Requirements:
  ✅ "What are the core features?" (top 3-5 only)
  ✅ "Who are the users? (consumers, creators, admins?)"
  ✅ "What operations? (read-heavy? write-heavy? both?)"
  ✅ "Real-time needed? Or eventual consistency OK?"
  ✅ "Any specific features to focus on?"

Non-Functional Requirements:
  ✅ "Expected DAU / MAU?"
  ✅ "Read:Write ratio?"
  ✅ "Latency expectations? (P99 < 200ms?)"
  ✅ "Availability target? (99.9%? 99.99%?)"
  ✅ "Consistency model? (strong or eventual?)"
  ✅ "Data retention? (forever? 5 years? 30 days?)"

⚠️ NEVER SKIP THIS STEP!
  Interviewers WANT you to ask questions.
  It shows you think before coding.
  Designing without requirements = building without spec.

Template answer:
  "Let me confirm the scope. We need [feature 1, 2, 3].
   We're targeting [X M] DAU, [Y:1] read/write ratio,
   [Z ms] latency, [W%] availability. Sound right?"
""")
````

````python
# ============================================================
# Step 2: Back-of-Envelope Template
# ============================================================

print("""
🧮 ESTIMATION TEMPLATE

Fill in the blanks for any system:

1. QPS
   Write QPS = DAU × writes_per_user / 86400
   Read QPS  = DAU × reads_per_user / 86400
   Peak QPS  = Average × 2-3x

2. Storage (over retention period)
   = items_per_day × item_size × days
   Example: 10M posts/day × 1KB × 365 × 5 = 18 TB

3. Bandwidth
   Ingress = Write QPS × avg_write_size
   Egress  = Read QPS × avg_response_size

4. Cache
   80/20 rule: 20% of data serves 80% of reads
   Cache size = Daily reads × 0.2 × item_size

Numbers to Memorize:
┌────────────────────────────────────────────┐
│ 1 day              = 86,400 seconds        │
│ 1 day              ≈ 100,000 seconds (est) │
│ 1 million / day    ≈ 12 / sec              │
│ 100 million / day  ≈ 1,200 / sec           │
│ 1 billion / day    ≈ 12,000 / sec          │
│ 1 KB               = 1,000 bytes           │
│ 1 MB               = 1,000 KB              │
│ 1 GB               = 1,000 MB              │
│ 1 TB               = 1,000 GB              │
│ 1 PB               = 1,000 TB              │
│ char               ≈ 1 byte                │
│ int/long            ≈ 8 bytes               │
│ UUID               ≈ 16 bytes              │
│ timestamp           ≈ 8 bytes               │
│ image (compressed)  ≈ 200KB-1MB             │
│ short video (1min)  ≈ 50MB                  │
└────────────────────────────────────────────┘
""")
````

````python
# ============================================================
# Step 3: High-Level Design Toolkit
# ============================================================

print("""
🏗️ HIGH-LEVEL DESIGN — BUILDING BLOCKS CHEAT SHEET

Pick the right tool for each job:

┌──────────────────────────────────────────────────────────┐
│ Need                     │ Component                     │
├──────────────────────────────────────────────────────────┤
│ High read throughput     │ Cache (Redis) + Read replicas │
│ High write throughput    │ Message queue (Kafka)         │
│ File/media storage       │ Object store (S3) + CDN      │
│ Full-text search         │ Elasticsearch                │
│ Real-time communication  │ WebSocket                    │
│ Async processing         │ Message queue + workers      │
│ Rate limiting            │ Token bucket (Redis)         │
│ Unique ID generation     │ Snowflake / UUID             │
│ Nearby search            │ Geohash / Quadtree           │
│ Session/fast KV          │ Redis                        │
│ Relational data          │ PostgreSQL / MySQL           │
│ Time-series / write-heavy│ Cassandra / InfluxDB         │
│ Graph relationships      │ Neo4j / GraphQL              │
│ Load distribution        │ Load balancer (L7)           │
│ API management           │ API Gateway                  │
│ Service coordination     │ Zookeeper / etcd             │
└──────────────────────────────────────────────────────────┘

API Design Template:
  POST /api/v1/{resource}       → Create
  GET  /api/v1/{resource}/{id}  → Read
  PUT  /api/v1/{resource}/{id}  → Update
  DELETE /api/v1/{resource}/{id}→ Delete
  GET  /api/v1/{resource}?limit=20&cursor=xxx → List (paginated)

Data Model Template:
  • Start with entities (User, Post, Comment...)
  • Define relationships (1:1, 1:N, M:N)
  • Choose SQL or NoSQL per entity
  • Add indexes for query patterns
""")
````

````python
# ============================================================
# Step 4: Deep Dive Topics
# ============================================================

print("""
🔬 DEEP DIVE — TOPICS TO ALWAYS COVER

Pick 2-3 based on the system. These impress interviewers:

1. SCALING
   • How to handle 10× growth?
   • Horizontal scale: add more servers
   • DB sharding: how to partition? What's the shard key?
   • Caching: what layer? What eviction? Cache invalidation?

2. FAILURE HANDLING
   • What if a server dies? (stateless → any server can handle)
   • What if DB goes down? (replicas, failover)
   • What if cache goes down? (thundering herd → stale-while-revalidate)
   • What if message queue fills up? (backpressure, dead letter queue)

3. CONSISTENCY & TRADEOFFS
   • CAP: are you CP or AP? Why?
   • Eventual vs strong consistency — for THIS system
   • Where you sacrifice and why

4. MONITORING & OBSERVABILITY
   • Key metrics to track (QPS, latency P99, error rate)
   • Alerting thresholds
   • Distributed tracing (correlation IDs)

5. SECURITY
   • Authentication (JWT, OAuth)
   • Rate limiting
   • Data encryption (at rest, in transit)
   • Input validation
""")
````

````python
# ============================================================
# RESHADED Framework
# ============================================================

print("""
╔════════════════════════════════════════════════════════════════╗
║  RESHADED FRAMEWORK — Memory Aid                             ║
╠════════════════════════════════════════════════════════════════╣
║                                                                ║
║  R — Requirements (functional + non-functional)               ║
║  E — Estimation (QPS, storage, bandwidth, cache)              ║
║  S — Storage schema (data model, DB choice)                   ║
║  H — High-level design (architecture diagram)                 ║
║  A — API design (endpoints, request/response)                 ║
║  D — Detailed design (deep dive into components)              ║
║  E — Evaluate (tradeoffs, alternatives, improvements)         ║
║  D — Distinguish (what makes YOUR design unique/good)         ║
║                                                                ║
║  Alternatively, the simpler 4-step:                           ║
║  1. Requirements (5 min)                                      ║
║  2. Estimation (5 min)                                        ║
║  3. Design (15 min) — draw, APIs, data model                 ║
║  4. Deep Dive (15 min) — scale, failures, tradeoffs          ║
║                                                                ║
╚════════════════════════════════════════════════════════════════╝
""")
````

````python
# ============================================================
# Common Mistakes & Pro Tips
# ============================================================

print("""
❌ COMMON MISTAKES

  1. Jumping into design without asking requirements
     → ALWAYS ask 3-5 clarifying questions first
  
  2. Going too deep too early
     → First show the big picture, THEN drill down
  
  3. Not drawing diagrams
     → Draw boxes and arrows. Verbal-only = lost interviewer
  
  4. Over-engineering
     → Start simple. Scale when needed. Say "we can add X later"
  
  5. Ignoring tradeoffs
     → Every decision has pros/cons. STATE them explicitly.
     → "I chose X over Y because..."
  
  6. Forgetting non-functional requirements
     → Latency, availability, consistency are as important as features
  
  7. Not communicating
     → Think out loud. The interviewer wants to see your PROCESS

✅ PRO TIPS

  1. Drive the conversation — don't wait for prompts
     → "Next, let me discuss the data model..."
  
  2. Propose → Justify → Get feedback
     → "I'd use Cassandra here because write-heavy. Make sense?"
  
  3. Use real numbers
     → "With 100M DAU, that's ~1200 QPS which a single DB can handle"
  
  4. Mention real systems by name
     → "Similar to how Netflix uses Open Connect for CDN..."
  
  5. End with improvements
     → "Given more time, I'd add monitoring, circuit breakers, and..."
""")
````

````python
# ============================================================
# Universal Template — Apply to ANY design question
# ============================================================

print("""
📝 UNIVERSAL SD INTERVIEW TEMPLATE

Copy this mental model for any question:

"Design X" → Your 40-minute plan:

[0:00-5:00] REQUIREMENTS
  • "What are the core features we need?" → list 3-5
  • "What scale are we targeting?" → DAU, QPS
  • "Any specific constraints?" → latency, consistency
  • Write them on the board

[5:00-10:00] ESTIMATION
  • QPS: DAU × actions / 86400
  • Storage: items × size × retention
  • Bandwidth: QPS × payload_size
  • "This tells us we need [caching / sharding / CDN]"

[10:00-25:00] HIGH-LEVEL DESIGN
  • Draw: Client → LB → API → [Services] → [Data stores]
  • API: 3-5 core endpoints
  • Data model: 3-5 tables + relationships
  • Walk through a user flow on the diagram

[25:00-40:00] DEEP DIVE
  • "Let me deep dive into [hardest component]..."
  • Scaling: "At this QPS, we need..."
  • Failures: "If this fails, we..."
  • Tradeoffs: "I chose X over Y because..."
  • End: "Given more time, I'd add..."

═══════════════════════════════════════════════

Quick reference — Systems we covered:
┌──────────────────────────────────────────────────┐
│ System          │ Key Challenge                   │
├──────────────────────────────────────────────────┤
│ URL Shortener   │ Base62, cache, 301 vs 302      │
│ Twitter         │ Fan-out (write vs read vs hybrid)│
│ Instagram       │ Media storage, S3+CDN, pre-sign │
│ WhatsApp        │ WebSocket, delivery states, E2E │
│ YouTube/Netflix │ Transcoding, ABR, HLS/DASH      │
│ Uber            │ Geohash, 1.7M loc updates/sec   │
│ Parking Lot     │ Class hierarchy, thread-safe     │
│ LRU Cache       │ DLL + HashMap, thread-safe       │
│ Rate Limiter    │ Token bucket, sliding window     │
│ Elevator        │ SCAN algorithm, dispatch         │
│ Booking System  │ Optimistic locking, concurrency  │
│ Snake Game      │ Deque + set, O(1) per move       │
└──────────────────────────────────────────────────┘
""")
````

## 📝 Interview Questions

| # | Question | Key Points |
|---|----------|-----------|
| 1 | Walk me through your SD approach | Requirements→Estimation→Design→Deep Dive. 4 steps, 40 min |
| 2 | How do you handle scale? | Horizontal scaling, caching, sharding, CDN, async processing |
| 3 | What tradeoffs do you consider? | CAP, consistency vs availability, latency vs throughput |
| 4 | How to handle failures? | Replicas, failover, circuit breaker, retry, DLQ |
| 5 | SQL vs NoSQL? | SQL: ACID, joins. NoSQL: scale, flexible schema. Choose per use case |
