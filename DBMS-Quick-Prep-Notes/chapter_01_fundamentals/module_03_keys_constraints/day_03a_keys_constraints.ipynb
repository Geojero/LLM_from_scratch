```
╔══════════════════════════════════════════════════════════════╗
║                   KEY HIERARCHY                              ║
╠══════════════════════════════════════════════════════════════╣
║                                                              ║
║  Super Key                                                   ║
║  └── Any set of attributes that uniquely identifies a row    ║
║      {id}, {id,name}, {id,name,email}, {email}...            ║
║                                                              ║
║  Candidate Key                                               ║
║  └── MINIMAL super key (remove any attr → no longer unique)  ║
║      {id}, {email} ← both are candidate keys                ║
║                                                              ║
║  Primary Key                                                 ║
║  └── ONE chosen candidate key  →  {id}                      ║
║                                                              ║
║  Alternate Key                                               ║
║  └── Candidate keys NOT chosen as PK  →  {email}            ║
║                                                              ║
║  Foreign Key                                                 ║
║  └── References PK of another table (creates relationship)  ║
║                                                              ║
╚══════════════════════════════════════════════════════════════╝
```

````python
# ============================================
# 1. Key Identification Exercise
# ============================================

print("  Key Identification Exercise")
print("  " + "═" * 55)
print()
print("  Table: EMPLOYEE(emp_id, name, email, phone, dept_id, ssn)")
print()

keys = {
    "Super Keys": [
        "{emp_id}", "{ssn}", "{email}", "{emp_id, name}",
        "{ssn, email}", "{emp_id, name, email, phone}...",
        "(any superset of a candidate key)"
    ],
    "Candidate Keys": [
        "{emp_id}  — minimal, unique",
        "{ssn}     — minimal, unique",
        "{email}   — minimal, unique",
    ],
    "Primary Key": ["{emp_id}  — chosen as PK (auto-increment, stable)"],
    "Alternate Keys": ["{ssn}, {email}  — UNIQUE constraints"],
    "Foreign Key": ["{dept_id} → references department(dept_id)"],
}

for key_type, examples in keys.items():
    print(f"  {key_type}:")
    for ex in examples:
        print(f"    • {ex}")
    print()
````

````python
# ============================================
# 2. All SQL Constraints — DDL Examples
# ============================================

ddl_examples = """
-- 1. PRIMARY KEY
CREATE TABLE department (
    dept_id   INT PRIMARY KEY,
    name      VARCHAR(50) NOT NULL
);

-- 2. FOREIGN KEY + referential integrity
CREATE TABLE employee (
    emp_id    INT PRIMARY KEY AUTO_INCREMENT,
    name      VARCHAR(100) NOT NULL,
    email     VARCHAR(100) UNIQUE,          -- 3. UNIQUE (alternate key)
    salary    DECIMAL(10,2) CHECK (salary > 0),  -- 4. CHECK constraint
    status    VARCHAR(10) DEFAULT 'active',  -- 5. DEFAULT
    dept_id   INT NOT NULL,                  -- 6. NOT NULL
    FOREIGN KEY (dept_id) REFERENCES department(dept_id)
        ON DELETE RESTRICT       -- prevent deleting dept with employees
        ON UPDATE CASCADE        -- update FK if dept_id changes
);

-- 7. COMPOSITE PRIMARY KEY
CREATE TABLE enrollment (
    student_id INT,
    course_id  INT,
    semester   VARCHAR(10),
    grade      CHAR(1),
    PRIMARY KEY (student_id, course_id, semester),
    FOREIGN KEY (student_id) REFERENCES student(id),
    FOREIGN KEY (course_id) REFERENCES course(id)
);
"""

print("  SQL Constraint Examples:")
for line in ddl_examples.strip().split('\n'):
    print(f"  {line}")
````

````python
# ============================================
# 3. Referential Integrity Actions
# ============================================

actions = [
    ("RESTRICT / NO ACTION",  "Block the delete/update",             "Default. Error if child rows exist"),
    ("CASCADE",               "Delete/update children too",          "ON DELETE CASCADE — delete order → delete order_items"),
    ("SET NULL",              "Set FK to NULL",                       "ON DELETE SET NULL — delete manager → employees.mgr_id = NULL"),
    ("SET DEFAULT",           "Set FK to default value",             "ON DELETE SET DEFAULT — rare, need DEFAULT defined"),
]

print("  Referential Integrity Actions (ON DELETE / ON UPDATE)")
print(f"  {'Action':<22} {'Behavior':<32} {'Example'}")
print("  " + "─" * 85)
for action, behavior, example in actions:
    print(f"  {action:<22} {behavior:<32} {example}")

print("\n  💡 Interview tip: 'What happens if you delete a parent row?'")
print("  → Depends on ON DELETE action: RESTRICT (error), CASCADE (delete children), SET NULL")
````

## 📝 Interview Questions

| # | Question | Key Point |
|---|----------|------------|
| 1 | Super key vs candidate key? | Super: any uniquely identifying set. Candidate: MINIMAL super key |
| 2 | Can a table have multiple PKs? | No. One PK, but can have multiple candidate keys (alternates become UNIQUE) |
| 3 | Can FK be NULL? | Yes if column allows NULL (partial participation). PK can NEVER be NULL |
| 4 | ON DELETE CASCADE vs RESTRICT? | CASCADE: delete children. RESTRICT: block delete if children exist |
| 5 | Composite key vs composite PK? | Composite key = multi-column key. Can be PK or unique constraint |
