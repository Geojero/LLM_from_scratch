## The Story: Building Complex Pipelines From Simple Blocks 🧱

Think of LEGO bricks. Each brick is simple, but you can combine them into anything — a house, a spaceship, a castle.

**LCEL (LangChain Expression Language)** works the same way. Each component is a **Runnable** — a simple block with a standard interface. Connect them with `|` and you get arbitrarily complex pipelines.

This is the **core skill** for LangChain development. Master this and everything else clicks.

## 🔗 LCEL: The Pipe Operator

```
╔═══════════════════════════════════════════════════════════════════════╗
║            LCEL — LangChain Expression Language                       ║
╠═══════════════════════════════════════════════════════════════════════╣
║                                                                       ║
║  The | operator chains Runnables together:                           ║
║                                                                       ║
║    chain = step_1 | step_2 | step_3                                  ║
║                                                                       ║
║  When you call chain.invoke(input):                                  ║
║    input → step_1 → output_1 → step_2 → output_2 → step_3 → result ║
║                                                                       ║
║  Every LCEL component ("Runnable") has these methods:                ║
║    • invoke(input)          — process one input                      ║
║    • stream(input)          — yield output tokens one by one         ║
║    • batch([input1, ...])   — process multiple inputs in parallel    ║
║    • ainvoke(input)         — async version of invoke                ║
║                                                                       ║
║  This is the killer feature: streaming & batching come FREE!        ║
║  Build with invoke, get stream/batch automatically.                  ║
║                                                                       ║
╚═══════════════════════════════════════════════════════════════════════╝
```

## 🔧 The Runnable Family

```
╔═══════════════════════════════════════════════════════════════════════╗
║              THE RUNNABLE TOOLBOX                                     ║
╠═══════════════════════════════════════════════════════════════════════╣
║                                                                       ║
║  RunnablePassthrough                                                  ║
║    → Passes input through unchanged                                  ║
║    → Used to forward data past steps that don't need it             ║
║    → .assign() adds new keys while keeping existing ones            ║
║                                                                       ║
║  RunnableLambda                                                       ║
║    → Wraps ANY Python function as a Runnable                        ║
║    → Your custom transformation step                                 ║
║    → RunnableLambda(my_function) or just use @chain decorator       ║
║                                                                       ║
║  RunnableParallel                                                     ║
║    → Runs multiple Runnables AT THE SAME TIME                        ║
║    → Returns a dict with each result                                 ║
║    → {'key1': runnable1, 'key2': runnable2}                         ║
║                                                                       ║
║  itemgetter (from operator module)                                    ║
║    → Extracts specific keys from dict input                          ║
║    → itemgetter('name') extracts the 'name' key                     ║
║                                                                       ║
╚═══════════════════════════════════════════════════════════════════════╝
```

## 📊 Data Flow Visualization

```
  RunnablePassthrough:
    {"name": "AI"} ─────→ {"name": "AI"}  (unchanged)

  RunnableLambda(str.upper):
    "hello" ─────→ "HELLO"  (transformed)

  RunnableParallel:
    input ──┬──→ chain_a ──→ ┐
            │                 ├──→ {"summary": ..., "keywords": ...}
            └──→ chain_b ──→ ┘

  Real RAG pattern using all three:
    {"question": "..."}
         │
    RunnableParallel:
         ├── "context": retriever        ← fetch relevant docs
         └── "question": passthrough     ← forward user question
         │
         ▼
    {"context": [docs], "question": "..."}
         │
    prompt | model | parser
         │
         ▼
    Final answer
```

## 🌊 Streaming: Real-Time Token Output

```
╔═══════════════════════════════════════════════════════════════════════╗
║                   STREAMING                                           ║
╠═══════════════════════════════════════════════════════════════════════╣
║                                                                       ║
║  Without streaming:                                                   ║
║    User waits 5 seconds... then gets entire response at once.       ║
║                                                                       ║
║  With streaming:                                                      ║
║    User sees tokens appear one by one, like typing.                  ║
║    Feels 10× faster even though total time is the same!             ║
║                                                                       ║
║  In LangChain:                                                        ║
║    for chunk in chain.stream({"topic": "AI"}):                     ║
║        print(chunk, end="")                                          ║
║                                                                       ║
║  💡 Streaming works on ANY LCEL chain automatically!                ║
║     If you build with |, streaming comes for free.                   ║
║                                                                       ║
╚═══════════════════════════════════════════════════════════════════════╝
```

## 📦 Batch Processing: Multiple Inputs at Once

```
╔═══════════════════════════════════════════════════════════════════════╗
║                   BATCH PROCESSING                                    ║
╠═══════════════════════════════════════════════════════════════════════╣
║                                                                       ║
║  Instead of calling .invoke() in a loop (slow, sequential),          ║
║  use .batch() to process multiple inputs IN PARALLEL:                ║
║                                                                       ║
║  results = chain.batch([                                              ║
║      {"topic": "Python"},                                            ║
║      {"topic": "JavaScript"},                                        ║
║      {"topic": "Rust"},                                              ║
║  ])                                                                   ║
║                                                                       ║
║  This sends all 3 requests to the API simultaneously.               ║
║  3× faster than sequential invoke() calls!                           ║
║                                                                       ║
║  Use max_concurrency to control parallelism:                         ║
║    chain.batch(inputs, config={"max_concurrency": 5})               ║
║                                                                       ║
╚═══════════════════════════════════════════════════════════════════════╝
```

## 💡 Interview Prep: Common Questions

```
╔═══════════════════════════════════════════════════════════════════════╗
║           INTERVIEW QUESTIONS FOR THIS MODULE                        ║
╠═══════════════════════════════════════════════════════════════════════╣
║                                                                       ║
║  Q: Explain LCEL and the Runnable protocol.                          ║
║  A: LCEL uses the | pipe to compose components. Every component     ║
║     is a Runnable with invoke(), stream(), batch(), and async       ║
║     variants. Output of one feeds into the next. You get            ║
║     streaming and batching FREE from the composition.               ║
║                                                                       ║
║  Q: What's RunnableParallel and when do you use it?                 ║
║  A: Runs multiple Runnables concurrently on the same input.         ║
║     Returns a dict with each result. Classic use: RAG pattern       ║
║     where you retrieve context AND pass the question through       ║
║     at the same time.                                                ║
║                                                                       ║
║  Q: How does streaming work in LangChain?                            ║
║  A: Call chain.stream() instead of .invoke(). Returns an iterator   ║
║     yielding chunks (tokens) as they arrive. Works automatically    ║
║     on any LCEL chain. Essential for chat UIs and real-time apps.   ║
║                                                                       ║
║  Q: What's RunnableLambda for?                                       ║
║  A: Wraps any Python function as a Runnable so it can be used in    ║
║     LCEL chains. For custom data transformation, formatting,        ║
║     validation, or any logic between chain steps.                   ║
║                                                                       ║
╚═══════════════════════════════════════════════════════════════════════╝
```

## 📝 Quick Quiz

---

### Question 1:
What does `RunnablePassthrough.assign(summary=summary_chain)` do?

- A) Replaces all input with the summary
- B) Passes input through AND adds a new 'summary' key with chain result
- C) Deletes the input and only returns summary
- D) Creates a new variable in Python

<details><summary>Answer</summary>

**B) Passes input through AND adds a new 'summary' key with chain result**

`.assign()` keeps all existing keys from the input dict and adds new keys computed by the provided runnables. It's like dict.update() but reactive.

</details>

---

### Question 2:
You need to send the same text to 3 different LLMs and compare results. Which Runnable?

- A) RunnableLambda
- B) RunnableSequence
- C) RunnableParallel
- D) RunnablePassthrough

<details><summary>Answer</summary>

**C) RunnableParallel**

RunnableParallel runs multiple chains on the same input simultaneously. `{"gpt": gpt_chain, "claude": claude_chain, "gemini": gemini_chain}` would give you all 3 results in one call.

</details>
