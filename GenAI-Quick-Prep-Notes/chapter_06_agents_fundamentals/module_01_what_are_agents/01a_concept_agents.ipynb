# 🧭 Chapter 06-01a: From Chains to Agents — What Are AI Agents?

---

## 📖 The Story: The Intern vs The Senior Engineer

A **chain** is like an intern with a checklist. They follow steps in exact order: Step 1 → Step 2 → Step 3.

An **agent** is like a senior engineer. You say "deploy this to production" and they *figure out* the steps: check tests, review dependencies, build, deploy, verify, rollback if needed.

Agents **reason**, **decide**, and **act** — dynamically choosing what to do next.

---

## 🎯 What Is an AI Agent?

```
╔══════════════════════════════════════════════════════════════════════╗
║  AI AGENT                                                           ║
║                                                                      ║
║  A system where an LLM dynamically decides what actions to take     ║
║  and in what order, based on observation and reasoning.             ║
║                                                                      ║
║  Chain:  Input → Step1 → Step2 → Step3 → Output  (fixed)          ║
║  Agent:  Input → Think → Act → Observe → Think → Act → Output     ║
╚══════════════════════════════════════════════════════════════════════╝
```

---

## 🔄 The Evolution: Functions → Chains → Agents

```
Era 1: Functions
  translate(text) → fixed logic

Era 2: Chains (LangChain)
  prompt | llm | parser → fixed pipeline, flexible content

Era 3: Agents
  LLM decides: "I need to search → found info → need to calculate → done"
  Dynamic flow, unpredictable number of steps
```

| Feature | Chain | Agent |
|---------|-------|-------|
| Steps | Fixed | Dynamic |
| Tools | Pre-selected | Chosen at runtime |
| Loops | No | Yes (think → act → observe) |
| Complexity | Low | High |
| Predictability | High | Lower |
| Use when | Flow is known | Flow depends on input |

## 🧠 The ReAct Framework

```
╔══════════════════════════════════════════════════════════╗
║  ReAct = Reasoning + Acting                             ║
║                                                          ║
║  The agent alternates between:                          ║
║  • THOUGHT: "I need to find the weather data..."        ║
║  • ACTION:  search_weather("San Francisco")             ║
║  • OBSERVATION: "72°F, sunny"                            ║
║  • THOUGHT: "Now I can answer the user..."              ║
║  • FINAL ANSWER: "It's 72°F and sunny in SF"           ║
╚══════════════════════════════════════════════════════════╝
```

Example trace:
```
User: "What's the population of the country with the tallest building?"

Thought: I need to find which country has the tallest building.
Action: search("tallest building in the world")
Observation: Burj Khalifa in Dubai, UAE

Thought: Now I need the population of UAE.
Action: search("population of UAE")
Observation: ~9.9 million (2023)

Thought: I have the answer now.
Final Answer: The tallest building (Burj Khalifa) is in the UAE,
which has a population of approximately 9.9 million.
```

## 🏗️ Agent Architectures

### 1. ReAct (Iterative)
```
Think → Act → Observe → Think → Act → ... → Final Answer
```
Most common. One step at a time.

### 2. Plan-then-Execute
```
Plan: [Step1, Step2, Step3, Step4]
Execute: Step1 → Step2 → Step3 → Step4
Replan if needed
```
Better for complex, multi-step tasks.

### 3. Self-Reflection
```
Act → Evaluate result → "Was this good enough?" → Retry or finish
```
Adds a quality check after each action.

## 🧠 Memory Types for Agents

Agents need memory to maintain context across steps:

| Memory Type | What it Stores | Use Case |
|-------------|---------------|----------|
| **Buffer** | Full conversation history | Short conversations |
| **Window** | Last N messages | Long conversations |
| **Summary** | Compressed summary | Very long sessions |
| **Entity** | Facts about entities | "User likes Python" |
| **Vector** | Semantic retrieval of past messages | Large knowledge bases |

```
╔══════════════════════════════════════════════════════════╗
║  PRACTICAL MEMORY STRATEGY                               ║
║                                                          ║
║  Short session  → Buffer memory (keep everything)       ║
║  Long session   → Window (last 10) + Summary (older)    ║
║  Knowledge      → Vector memory (embed + retrieve)      ║
╚══════════════════════════════════════════════════════════╝
```

## 🎓 Interview Corner

**Q: What's the difference between a chain and an agent?**

<details>
<summary>💡 Click for Expert Answer</summary>

A **chain** has a predetermined execution flow — the sequence of steps is fixed at design time. A **prompt | model | parser** chain always runs those three steps.

An **agent** has a dynamic execution flow — the LLM decides at runtime what tool to call, whether to loop, and when to stop. The same agent might take 2 steps for one query and 10 steps for another.

Use chains when the workflow is predictable. Use agents when the workflow depends on the input or requires reasoning over multiple steps.

</details>

**Q: What can go wrong with agents?**

<details>
<summary>💡 Click for Expert Answer</summary>

1. **Infinite loops**: Agent keeps calling tools without converging
2. **Wrong tool selection**: Agent picks the wrong tool for the task
3. **Hallucinated tool calls**: Agent invents tool names or arguments
4. **Cost explosion**: Each loop iteration costs API calls
5. **Slow response**: Multiple sequential API calls add latency

Mitigations: max iterations limit, tool descriptions, structured outputs, cost monitoring, timeouts.

</details>
