## 📖 The Story: The Master Chef vs. The Home Cook

A general LLM is like a master chef who can cook any cuisine. But what if you need someone who *only* makes the best pizza? You don't retrain them from scratch — you **specialize** them: give them 1000 pizza recipes and let them practice.

That's fine-tuning: taking a powerful general model and specializing it for YOUR specific task.

---

## 🎯 What is Fine-Tuning?

```
╔══════════════════════════════════════════════════════════════════════╗
║  FINE-TUNING                                                        ║
║                                                                      ║
║  Taking a pre-trained LLM and training it further on YOUR specific ║
║  dataset to specialize its behavior, style, or knowledge.          ║
║                                                                      ║
║  Pre-training: Learn language (trillions of tokens, $millions)     ║
║  Fine-tuning:  Learn YOUR task (thousands of examples, $tens)      ║
╚══════════════════════════════════════════════════════════════════════╝
```

---

## 🤔 When to Fine-Tune vs. Alternatives

```
┌────────────────────────────────────────────────────────────┐
│  Decision Tree:                                            │
│                                                            │
│  Need specific knowledge? → RAG (retrieval)               │
│  Need specific format/style? → Fine-tuning                │
│  Need specific behavior? → Prompt engineering first        │
│  Prompt + RAG not enough? → Fine-tuning                   │
│  Need domain terminology? → Fine-tuning                   │
│  Budget < $50? → Prompt engineering + few-shot             │
└────────────────────────────────────────────────────────────┘
```

| Approach | Cost | Data Needed | Best For |
|----------|------|------------|----------|
| Prompt Engineering | Free | 0 | Quick fixes, formatting |
| Few-Shot Examples | Free | 3-10 | Pattern following |
| RAG | Low | Documents | Knowledge-heavy tasks |
| Fine-Tuning | Medium | 100-10K | Style, format, domain |
| Pre-Training | Very High | Billions | New language/domain |

## 🧠 LoRA — Low-Rank Adaptation

```
╔══════════════════════════════════════════════════════════════════════╗
║  LoRA (Low-Rank Adaptation)                                         ║
║                                                                      ║
║  Instead of updating ALL model weights (billions!), LoRA adds     ║
║  tiny trainable matrices alongside frozen weights.                 ║
║                                                                      ║
║  Full fine-tune: Update 7B parameters → needs 28GB+ VRAM          ║
║  LoRA: Update ~0.1% parameters → needs 8GB VRAM                   ║
║                                                                      ║
║  Original weights (frozen) + LoRA weights (trainable) = Output    ║
╚══════════════════════════════════════════════════════════════════════╝
```

```
    ┌─────────────────────────────┐
    │  Original Model (FROZEN)    │  7B parameters
    │  W (d × d matrix)           │  NOT updated
    └──────────┬──────────────────┘
               │
    ┌──────────▼──────────────────┐
    │  LoRA Adapter (TRAINABLE)   │  ~7M parameters
    │  A (d × r) × B (r × d)     │  r = rank (4-64)
    └──────────┬──────────────────┘
               │
    Output = W(x) + A·B(x)
```

---

## ⚡ QLoRA — Quantized LoRA

**QLoRA** = LoRA + 4-bit quantization. Even more memory efficient!

| Method | VRAM for 7B Model | Training Speed |
|--------|-------------------|----------------|
| Full Fine-Tune | ~28 GB | Slow |
| LoRA | ~8 GB | Fast |
| QLoRA | ~4 GB | Medium |

**QLoRA makes it possible to fine-tune a 7B model on a single consumer GPU (RTX 3090)!**

## 🎓 Interview Corner

**Q: When would you fine-tune vs. use RAG? Can you combine them?**

<details>
<summary>💡 Click for Expert Answer</summary>

**Fine-tune when**: You need consistent style/format, domain-specific terminology, or behavior change (e.g., always respond in medical language).

**RAG when**: You need access to specific documents, data changes frequently, or you need citations.

**Combine them**: Fine-tune for style + RAG for knowledge.
- Example: Fine-tune a model to write like a legal expert (style), then use RAG to inject specific case law (knowledge).
- This gives you a domain-expert model that can also access real-time data.

</details>
