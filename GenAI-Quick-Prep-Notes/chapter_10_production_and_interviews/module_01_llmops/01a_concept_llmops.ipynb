## 🔍 LangSmith — Observability for LLM Apps

```python
# Setup: Just set environment variables!
import os
os.environ["LANGCHAIN_TRACING_V2"] = "true"
os.environ["LANGCHAIN_API_KEY"] = "your-key"
os.environ["LANGCHAIN_PROJECT"] = "my-project"

# Now ALL LangChain calls are automatically traced!
# See: inputs, outputs, latency, tokens, errors
```

**What LangSmith shows you:**
- Every LLM call with input/output
- Token usage and costs per call
- Latency breakdown (which step is slow?)
- Error traces (where did it fail?)
- Custom evaluation datasets

## 💰 Cost Optimization Strategies

| Strategy | Savings | Effort |
|----------|---------|--------|
| Use smaller models (4o-mini vs 4o) | 90%+ | Low |
| Prompt caching | 50-80% | Low |
| Response caching (semantic) | 70-95% | Medium |
| Batch API (OpenAI) | 50% | Low |
| Shorter prompts | 20-40% | Medium |
| Fine-tuned smaller model | 80%+ | High |

## ⚡ Latency Optimization

```
┌─────────────────────────────────────────────────┐
│  Latency Budget (target: < 3 seconds)           │
│                                                  │
│  Embedding lookup:    ~50ms                      │
│  Vector search:       ~100ms                     │
│  LLM call:            ~1-3s (the bottleneck!)    │
│  Total:               ~1.5-3.5s                  │
└─────────────────────────────────────────────────┘
```

**Techniques:**
1. **Streaming** — Show tokens as they arrive
2. **Parallel tool calls** — Run independent tools concurrently
3. **Smaller models** — 4o-mini is 3x faster than 4o
4. **Pre-computation** — Cache embeddings, pre-chunk documents

## 🎓 Interview Corner

**Q: How would you reduce LLM costs by 80% in a production app?**

<details>
<summary>💡 Click for Expert Answer</summary>

1. **Model selection**: Use gpt-4o-mini instead of gpt-4o (90% cheaper, often good enough)
2. **Semantic caching**: Cache similar queries (Redis + embeddings)
3. **Prompt optimization**: Shorter system prompts, fewer examples
4. **Batch API**: 50% discount for non-real-time workloads
5. **Tiered approach**: Simple queries → small model, complex → large model
6. **Fine-tuning**: Replace long prompts with fine-tuned behavior

Combined, these typically achieve 80-95% cost reduction.
