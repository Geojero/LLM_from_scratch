## ✂️ Advanced Chunking Strategies

### 1. Fixed-Size Chunking (Basic)
```
[=====1000 chars=====][=====1000 chars=====][=====1000 chars=====]
         ^^^overlap^^^         ^^^overlap^^^
```
Simple but can split mid-sentence.

### 2. Semantic Chunking (Advanced)
```
[===Topic A===][======Topic B======][===Topic C===]
  detected by     embedding similarity     changes
```
Uses embeddings to detect where meaning shifts. Chunks are variable-sized.

### 3. Hierarchical Chunking
```
Parent: [==========Full Section==========]
Child:  [===Para 1===][===Para 2===][===Para 3===]
```
Search on small chunks (precise), return parent chunk (complete context).

### 4. Document-Aware Chunking
```
Markdown: Split by ## headers
Code:     Split by functions/classes
PDF:      Split by pages/sections
```

## 🔍 Query Transformation

The user's query is often not the best search query.

### Multi-Query Retrieval
```
User: "Why is my app slow?"
        │
        ▼ LLM generates 3 variations
  1. "application performance optimization"
  2. "diagnosing slow response times"
  3. "common causes of application latency"
        │
        ▼ All 3 search the vector DB
  Union of all results (deduplicated)
```

### HyDE (Hypothetical Document Embeddings)
```
User: "How to fix memory leaks?"
        │
        ▼ LLM generates a HYPOTHETICAL answer
  "Memory leaks occur when allocated memory is not freed...
   Common causes include circular references..."
        │
        ▼ Embed the hypothetical doc (not the question!)
  Search with this embedding → Better results!
```

Why? Questions and documents are phrased differently. HyDE makes the search query look more like a document.

## 🔀 Hybrid Search

```
╔══════════════════════════════════════════════════════════╗
║  HYBRID SEARCH = Vector Search + Keyword Search (BM25) ║
║                                                          ║
║  Vector: "ML model" matches "machine learning algorithm"║
║  BM25:   "torch.nn.Linear" matches exact code reference ║
║                                                          ║
║  Combined: Best of both worlds!                          ║
╚══════════════════════════════════════════════════════════╝
```

How to combine:
1. Get top-K from vector search
2. Get top-K from BM25 keyword search
3. **Reciprocal Rank Fusion (RRF)**: merge and re-score

```
RRF_score = Σ 1/(k + rank_i)
```

## 🏆 Re-Ranking

Initial retrieval is fast but rough. Re-ranking is slow but precise.

```
Step 1: Vector search → Top 50 chunks (fast, approximate)
Step 2: Cross-Encoder re-rank → Top 5 chunks (slow, precise)
```

**Cross-Encoder** sees the query AND document together (not separately), so it understands relevance better.

| Method | Speed | Quality | Use Case |
|--------|-------|---------|----------|
| Vector search only | ⚡ Fast | Good | Prototyping |
| + BM25 hybrid | ⚡ Fast | Better | Production |
| + Cross-encoder re-rank | 🐢 Slower | Best | High-stakes |

Popular re-rankers:
- Cohere Rerank API
- cross-encoder/ms-marco-MiniLM (local, free)
- Jina Reranker

## 📊 Self-Consistency & Multi-Query

### Multi-Query Approach
```
Original Q → LLM generates 3 paraphrases → 3 separate retrievals → Union
```

### Citation Tracking
```
Answer: "Revenue was $15M [Source: Q4 Report, p.12]"
```
Track which chunks contributed to which parts of the answer.

### Corrective RAG (CRAG)
```
1. Retrieve chunks
2. LLM evaluates: "Are these chunks relevant to the question?"
3. If NO → fall back to web search
4. If YES → generate answer
```

### GraphRAG
```
Instead of flat chunks, build a GRAPH of entities & relationships.
Traverse the graph to find connected information.
Better for questions spanning multiple documents.
```

## 🎓 Interview Corner

**Q: Your RAG system answers with wrong information. How would you debug it?**

<details>
<summary>💡 Click for Expert Answer</summary>

Systematic debugging:

1. **Check retrieval**: Are the right chunks being retrieved?
   - If NO → chunking problem, embedding model issue, or query-document mismatch
   - Try: better chunking, HyDE, multi-query, hybrid search

2. **Check context**: Is the answer in the retrieved context?
   - If NO → retrieval miss. Increase k, add re-ranking
   - If YES → generation problem

3. **Check generation**: Is the LLM ignoring the context?
   - Try: stronger system prompt, lower temperature, bigger model

4. **Check data**: Is the source data accurate and complete?

Use LangSmith to trace every step!
