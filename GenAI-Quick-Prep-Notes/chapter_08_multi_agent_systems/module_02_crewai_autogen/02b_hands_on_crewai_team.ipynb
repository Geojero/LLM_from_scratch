## 📖 The Mission

Build a CrewAI content creation crew that:
1. **Research Agent** — Researches a topic
2. **Writer Agent** — Writes a blog post
3. **Editor Agent** — Reviews and polishes

See how CrewAI handles multi-agent coordination with minimal code!

---

## 🧪 Step 1: Setup

````python
# !pip install crewai crewai-tools langchain-openai python-dotenv
````

````python
import os
from dotenv import load_dotenv
load_dotenv()

# Verify API key
assert os.getenv("OPENAI_API_KEY"), "Set OPENAI_API_KEY in .env file!"
print("✅ API key loaded!")
````

## 🧪 Step 2: Define Agents with Roles

````python
from crewai import Agent, Task, Crew, Process

# Agent 1: Researcher
researcher = Agent(
    role="Senior Research Analyst",
    goal="Find the most relevant and recent information about the given topic",
    backstory="""You are a seasoned research analyst with 10 years of experience.
    You excel at finding key insights, trends, and data points. You always verify
    information from multiple angles.""",
    verbose=True,
    allow_delegation=False
)

# Agent 2: Writer
writer = Agent(
    role="Content Writer",
    goal="Write an engaging and informative blog post based on research",
    backstory="""You are a creative content writer who can take complex topics
    and make them accessible and engaging. You write with clarity and punch.""",
    verbose=True,
    allow_delegation=False
)

# Agent 3: Editor
editor = Agent(
    role="Senior Editor",
    goal="Polish content to publication quality",
    backstory="""You are a meticulous editor who ensures every piece is clear,
    grammatically correct, well-structured, and engaging. You have a keen eye
    for detail.""",
    verbose=True,
    allow_delegation=False
)

print("✅ 3 Agents defined: Researcher, Writer, Editor")
````

## 🧪 Step 3: Define Tasks

````python
# Task 1: Research
research_task = Task(
    description="""Research the topic: 'How AI Agents are changing software development in 2024'.
    Find:
    - Key trends and statistics
    - Real-world examples and use cases
    - Expert opinions
    - Future predictions
    Compile findings in a structured research brief.""",
    expected_output="A comprehensive research brief with key findings, examples, and data points.",
    agent=researcher
)

# Task 2: Write
write_task = Task(
    description="""Using the research provided, write a blog post that:
    - Has a catchy title
    - Includes an engaging introduction
    - Covers 3-4 main points with examples
    - Has a compelling conclusion
    - Is 500-800 words""",
    expected_output="A complete blog post in markdown format, ready for editing.",
    agent=writer
)

# Task 3: Edit
edit_task = Task(
    description="""Edit and polish the blog post:
    - Fix any grammar or spelling issues
    - Improve sentence flow and readability
    - Ensure consistent tone
    - Add a meta description (for SEO)
    Return the final polished version.""",
    expected_output="The final, publication-ready blog post with meta description.",
    agent=editor
)

print("✅ 3 Tasks defined: Research → Write → Edit")
````

## 🧪 Step 4: Assemble & Run the Crew!

````python
# Create the crew
content_crew = Crew(
    agents=[researcher, writer, editor],
    tasks=[research_task, write_task, edit_task],
    process=Process.sequential,  # Tasks run in order
    verbose=True
)

print("🚀 Kicking off the Content Crew!")
print("=" * 60)

result = content_crew.kickoff()

print("\n" + "=" * 60)
print("✅ Crew finished!")
````

````python
# Print the final output
print("📄 FINAL OUTPUT")
print("=" * 60)
print(result)
````

## 🔍 Compare: Same Task in LangGraph vs CrewAI

| Aspect | LangGraph | CrewAI |
|--------|-----------|--------|
| **Lines of code** | ~80 | ~40 |
| **Setup time** | 30 min | 10 min |
| **Control** | Full graph control | Process.sequential/hierarchical |
| **Debugging** | Step-by-step state | Verbose logging |
| **Customization** | Any flow possible | Role/task based |

## ✍️ Practice Challenge

Try changing `Process.sequential` to `Process.hierarchical` and adding a `manager_llm`. In hierarchical mode, CrewAI auto-creates a manager that delegates tasks.

<details>
<summary>💡 Hint</summary>

    tasks=[research_task, write_task, edit_task],
