## 🚢 CrewAI — Build Agent Crews

**Philosophy**: Agents are crew members with **roles**, **goals**, and **backstories**.

```python
from crewai import Agent, Task, Crew

researcher = Agent(
    role="Senior Researcher",
    goal="Find the most relevant information",
    backstory="You are a seasoned research analyst...",
    tools=[search_tool]
)

writer = Agent(
    role="Content Writer",
    goal="Write engaging blog posts",
    backstory="You are a creative writer..."
)

task1 = Task(description="Research AI trends", agent=researcher)
task2 = Task(description="Write blog post", agent=writer)

crew = Crew(agents=[researcher, writer], tasks=[task1, task2])
result = crew.kickoff()
```

**Pros**: Super easy, role-based design, built-in task delegation
**Cons**: Less control over flow, opinionated, heavier abstraction

## 🤖 AutoGen — Conversational Agents

**Philosophy**: Agents communicate through **conversations** (chat-based).

```python
from autogen import AssistantAgent, UserProxyAgent

assistant = AssistantAgent(
    name="assistant",
    llm_config={"model": "gpt-4o-mini"}
)

user_proxy = UserProxyAgent(
    name="user",
    human_input_mode="NEVER",
    code_execution_config={"work_dir": "coding"}
)

user_proxy.initiate_chat(assistant, message="Write a Python script...")
```

**Pros**: Code execution built-in, group chat support, flexible
**Cons**: Conversation-heavy (can be verbose), less structured control

## 📊 Framework Comparison

| Feature | LangGraph | CrewAI | AutoGen |
|---------|-----------|--------|---------|
| **Abstraction** | Low (graphs) | High (roles) | Medium (chat) |
| **Control** | Maximum | Limited | Moderate |
| **Learning curve** | Steep | Easy | Medium |
| **Persistence** | ✅ Checkpoints | ❌ Basic | ❌ Manual |
| **HITL** | ✅ First-class | ✅ Basic | ✅ Built-in |
| **Code execution** | Manual | ✅ CodeAgent | ✅ Built-in |
| **Production ready** | ✅✅✅ | ✅✅ | ✅✅ |
| **Best for** | Complex flows | Quick prototypes | Code-heavy tasks |

### Decision Framework
```
Need maximum control?        → LangGraph
Need quick prototype?        → CrewAI
Need code execution?         → AutoGen
Production with persistence? → LangGraph
Simple role-based team?      → CrewAI
```

## 🎓 Interview Corner

**Q: You need to build a multi-agent system for a startup. Which framework and why?**

<details>
<summary>💡 Click for Expert Answer</summary>

**My recommendation**: Start with **CrewAI** for MVP/prototype (fast, easy, good enough). Migrate to **LangGraph** for production (control, persistence, monitoring).

Reasoning:
1. CrewAI gets you to demo in hours, not days
2. Once you validate the concept, rebuild in LangGraph for:
   - Checkpointing (crash recovery)
   - Human-in-the-loop (approval flows)
   - LangSmith integration (monitoring)
   - Streaming (real-time UI)
3. AutoGen if your use case is heavily code-generation focused
