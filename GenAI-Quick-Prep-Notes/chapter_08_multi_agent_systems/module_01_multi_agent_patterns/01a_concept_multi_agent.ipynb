## 📖 The Story: The Corporate Org Chart

A company doesn't have one person do everything. There's a CEO (supervisor) who delegates to VPs (team leads), who delegate to specialists (workers). Sometimes a VP hands off work to another VP's team.

Multi-agent AI systems work the same way. No single agent does everything — they **collaborate**.

---

## 🎯 Multi-Agent Patterns

```
╔══════════════════════════════════════════════════════════════════════╗
║  MULTI-AGENT SYSTEM                                                 ║
║                                                                      ║
║  Multiple specialized AI agents working together on complex        ║
║  tasks. Each agent has its own tools, prompts, and expertise.      ║
║                                                                      ║
║  Key insight: Specialization > Generalization for complex tasks.   ║
╚══════════════════════════════════════════════════════════════════════╝
```

## 🏗️ Pattern 1: Supervisor

One agent (supervisor) decides which worker agent to call.

```
         ┌──────────────┐
         │  Supervisor  │◄───────────────┐
         └──┬───┬───┬───┘               │
            │   │   │                   │
    ┌───────▼┐ ┌▼──────┐ ┌▼────────┐    │
    │ Coder  │ │Tester │ │Reviewer │────┘
    └────────┘ └───────┘ └─────────┘
```

**Pros**: Clear control, easy to debug, predictable routing
**Cons**: Supervisor is a bottleneck, single point of failure

---

## 🏗️ Pattern 2: Hierarchical Delegation

Multi-level supervision — supervisors managing supervisors.

```
              ┌─────────┐
              │   CEO   │
              └──┬───┬──┘
          ┌──────▼┐ ┌▼──────┐
          │VP Eng │ │VP Mkt │
          └┬────┬─┘ └┬────┬─┘
      ┌────▼┐ ┌▼───┐ ┌▼──┐ ┌▼───┐
      │Back │ │Front│ │SEO│ │Ads │
      └─────┘ └────┘ └───┘ └────┘
```

**Pros**: Scales well, modular teams, parallel work
**Cons**: Complex, potential communication overhead

---

## 🏗️ Pattern 3: Agent Handoff (Swarm)

Agents pass control directly to each other — no supervisor.

```
    ┌────────┐      ┌────────┐      ┌────────┐
    │Agent A │─────►│Agent B │─────►│Agent C │
    │(Triage)│      │(Sales) │      │(Support)│
    └────────┘      └────────┘      └────────┘
```

Each agent decides: "I'm done, hand off to Agent X"

**Pros**: Decentralized, flexible, each agent autonomous
**Cons**: Hard to debug, potential infinite loops, no central control

---

## 📊 When to Use Which Pattern

| Scenario | Best Pattern |
|----------|-------------|
| Clear workflow (code → test → review) | Supervisor |
| Large team, many specialties | Hierarchical |
| Dynamic routing, customer support | Handoff/Swarm |
| Research + writing | Supervisor or Pipeline |
| Chat with escalation | Handoff |

## 🛠️ Implementation in LangGraph

```python
# Supervisor pattern
def supervisor(state):
    response = llm.invoke([
        SystemMessage(content="""Choose the next worker: coder, tester, reviewer, or FINISH.
        Respond with ONLY the worker name."""),
        *state["messages"]
    ])
    return {"next_agent": response.content.strip().lower()}

def route_supervisor(state) -> str:
    next_agent = state["next_agent"]
    if next_agent == "finish":
        return END
    return next_agent

graph.add_conditional_edges("supervisor", route_supervisor)
# All workers route back to supervisor
for worker in ["coder", "tester", "reviewer"]:
    graph.add_edge(worker, "supervisor")
```

## 🎓 Interview Corner

**Q: Compare supervisor vs. handoff multi-agent patterns. When would you use each?**

<details>
<summary>💡 Click for Expert Answer</summary>

**Supervisor**:
- Central coordinator makes all routing decisions
- Best for: structured workflows, predictable tasks, when you want full control
- Example: Code generation pipeline (plan → code → test → review)

**Handoff**:
- Agents decide themselves who to hand off to
- Best for: dynamic scenarios, customer service, when next step depends on conversation
- Example: Support chat (triage → billing OR technical → escalation)

**Real-world**: Often hybrid — supervisor at top level, handoffs within teams.

</details>

## ✅ Key Takeaways

```
┌─────────────────────────────────────────────────────────────────────┐
│  📌 Supervisor: One coordinator, many workers (most common).       │
├─────────────────────────────────────────────────────────────────────┤
│  📌 Hierarchical: Multi-level delegation for large systems.        │
├─────────────────────────────────────────────────────────────────────┤
│  📌 Handoff/Swarm: Agents transfer control directly (OpenAI Swarm).│
├─────────────────────────────────────────────────────────────────────┤
│  📌 Workers route BACK to supervisor (cycle) until FINISH.         │
├─────────────────────────────────────────────────────────────────────┤
│  📌 Specialization > generalization for complex tasks.             │
└─────────────────────────────────────────────────────────────────────┘
```
