## 🤔 Quick Check: Are You Following?

**Question:** If an LLM predicts one word at a time, how does it write a 500-word essay?

<details>
<summary>👆 Click to see answer</summary>

**It loops!**

1. Predict word 1 → "The"
2. Now input is "The" → predict word 2 → "quick"
3. Now input is "The quick" → predict word 3 → "brown"
4. Keep going until it generates a special "STOP" token

This is called **autoregressive generation** — each new word depends on ALL previous words.

That's why LLMs can write coherent paragraphs — each word is chosen based on the full context of everything before it!

</details>

## 🗝️ Key Terminology Crash Course

Before we write any code, you MUST know these terms. They come up in **every interview:**

```

║               Used for: general tasks, chatbots                      ║
║                                                                       ║
║  temp = 1.5+ → Very creative, sometimes nonsensical                 ║
║               Used for: brainstorming, creative writing              ║
║                                                                       ║

║  The text the model generates BACK to you                            ║

║  When a model generates a response (as opposed to training)         ║
║  "Running inference" = asking the model to produce output            ║
║                                                                       ║
║  🎓 TRAINING vs FINE-TUNING                                           ║
║  Training: Building the model from scratch (costs millions $)       ║
║  Fine-tuning: Adjusting a trained model for YOUR task (affordable)  ║
║                                                                       ║
║  😵 HALLUCINATION                                                     ║
║  When the model confidently says something WRONG                    ║
║  "Gandhi was born in 1875" — sounds right, but it's 1869!           ║
```

### 🎉 Your VERY First LLM Call (Raw OpenAI)

Let's talk directly to GPT-4o using the OpenAI Python library:

````python
from openai import OpenAI

# Create the client
client = OpenAI()  # Automatically reads OPENAI_API_KEY from environment

# Make your first call!
response = client.chat.completions.create(
    model="gpt-4o-mini",              # The model to use
    messages=[
        {"role": "system", "content": "You are a helpful assistant."},
        {"role": "user", "content": "What is Generative AI? Explain in 2 sentences."}
    ],
    temperature=0.7                    # Creativity level
)

# Extract and print the response
answer = response.choices[0].message.content
print("🤖 LLM says:")
print(answer)
print(f"\n📊 Tokens used: {response.usage.total_tokens}")
````

**🎉 You just talked to a robot brain with 10 lines of code!**

### 🧪 Experiment: Play with Temperature!

````python
# Let's see how temperature affects output!

prompt = "Write a one-line tagline for a coffee shop"

for temp in [0.0, 0.7, 1.5]:
    print(f"\n🌡️ Temperature = {temp}")
    print("-" * 50)
    
    # Call 3 times at each temperature to see variation
    for i in range(3):
        response = client.chat.completions.create(
            model="gpt-4o-mini",
            messages=[{"role": "user", "content": prompt}],
            temperature=temp,
            max_tokens=30  # Keep responses short
        )
        print(f"  {i+1}. {response.choices[0].message.content}")
````

### 🔍 What Did You Notice?

```
┌─────────────────────────────────────────────────────────────────────┐
│  temp = 0.0 → Same answer every time (deterministic)               │
│  temp = 0.7 → Slightly different each time (balanced)              │
│  temp = 1.5 → Wildly different, sometimes weird (creative)         │
└─────────────────────────────────────────────────────────────────────┘
```

**Rule of thumb:**
- **Factual tasks (code, math, data):** Use temperature = 0
- **Creative tasks (writing, brainstorming):** Use temperature = 0.7-1.0
- **Never use above 1.5** unless you enjoy chaos 😄

### 🎭 Experiment: System Prompts Change EVERYTHING

````python
# The SAME question with DIFFERENT system prompts

user_question = "What is Python?"

system_prompts = [
    "You are a computer science professor explaining to beginners.",
    "You are a pirate who explains everything with nautical metaphors.",
    "You are a 5-year-old who just learned about computers.",
    "You are a stand-up comedian who makes everything funny."
]

for system in system_prompts:
    response = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=[
            {"role": "system", "content": system},
            {"role": "user", "content": user_question}
        ],
        temperature=0.7,
        max_tokens=100
    )
    
    print(f"\n🎤 System: \"{system}\"")
    print(f"💬 Response: {response.choices[0].message.content}")
    print("=" * 70)
````

**💡 Key Insight:** The system prompt is **incredibly powerful**. Same input, completely different outputs. This is the foundation of **prompt engineering** (Chapter 2).

## 🔢 Understanding Tokens (Interview Favorite!)

Let's actually SEE what tokens look like:

````python
import tiktoken

# Load the tokenizer used by GPT-4o
enc = tiktoken.encoding_for_model("gpt-4o")

# Let's tokenize some text!
texts = [
    "Hello world",
    "I love programming in Python",
    "chatgpt",
    "Generative AI is transforming the world",
    "🚀🤖🎉",  # Emojis take more tokens!
]

print("TEXT → TOKENS")
print("=" * 70)

for text in texts:
    tokens = enc.encode(text)
    token_strings = [enc.decode([t]) for t in tokens]
    
    print(f"\n📝 \"{text}\"")
    print(f"   Token count: {len(tokens)}")
    print(f"   Token IDs:   {tokens}")
    print(f"   Broken into: {token_strings}")
````

### 💰 Token = Money!

Here's a cost estimator:

````python
# Approximate pricing per million tokens (April 2026)
PRICING = {
    "gpt-4o":       {"input": 2.50, "output": 10.00},
    "gpt-4o-mini":  {"input": 0.15, "output": 0.60},
    "claude-4-sonnet": {"input": 3.00, "output": 15.00},
}

def estimate_cost(input_text, output_tokens=500, model="gpt-4o-mini"):
    enc = tiktoken.encoding_for_model("gpt-4o")
    input_tokens = len(enc.encode(input_text))
    
    prices = PRICING[model]
    input_cost = (input_tokens / 1_000_000) * prices["input"]
    output_cost = (output_tokens / 1_000_000) * prices["output"]
    total = input_cost + output_cost
    
    print(f"Model: {model}")
    print(f"Input tokens: {input_tokens}")
    print(f"Output tokens: {output_tokens} (estimated)")
    print(f"Input cost:  ${input_cost:.6f}")
    print(f"Output cost: ${output_cost:.6f}")
    print(f"Total cost:  ${total:.6f}")
    print(f"\n💡 At this rate, $1 gets you ~{int(1/total):,} similar calls!")

# How much does a typical call cost?
estimate_cost("What is Generative AI? Explain in detail.", output_tokens=200)
````

**Takeaway:** GPT-4o-mini is incredibly cheap! For learning and prototyping, the cost is almost negligible.

### Question 1:
What does an LLM do fundamentally?

- A) Understands human language perfectly
- B) Predicts the next token based on previous tokens
- C) Searches the internet for answers
- D) Stores all human knowledge in a database

<details><summary>Answer</summary>

**B) Predicts the next token based on previous tokens**

An LLM is essentially a next-token predictor. It doesn't truly "understand" — it finds statistical patterns from training data. It doesn't search the internet (unless given tools to do so). This distinction is critical for interviews!

</details>

---

### Question 2:
If temperature = 0, what happens?

- A) Model refuses to answer
- B) Model gives the most creative response
- C) Model always gives the same answer for the same input
- D) Model runs faster

<details><summary>Answer</summary>

**C) Model always gives the same answer for the same input**

Temperature=0 means it always picks the highest-probability token. The output becomes deterministic. This is important for applications where you need consistent, reproducible results (like code generation or data extraction).

</details>

---

### Question 3:
Why do LLMs sometimes "hallucinate" (make up facts)?

- A) They want to trick the user
- B) They have a bug in their code
- C) They generate probable-sounding text, not verified facts
- D) Their training data was wrong

<details><summary>Answer</summary>

**C) They generate probable-sounding text, not verified facts**

LLMs don't have a "truth database." They generate text that *statistically sounds right*. Sometimes that text happens to be factually wrong. This is why RAG (Chapter 5) exists — to give LLMs access to verified, up-to-date information!

</details>

### Question 4:
When would you choose an open-weights model over GPT-4o?

- A) When you need the absolute best quality
- B) When data privacy is critical and you need full control
- C) When you want the easiest setup
- D) When cost doesn't matter

```
╔═══════════════════════════════════════════════════════════════════════╗
║               INTERVIEW QUESTIONS FOR THIS MODULE                    ║
╠═══════════════════════════════════════════════════════════════════════╣
║                                                                       ║
║  Q: What is the difference between Traditional AI and Generative AI? ║
║  A: Traditional AI classifies/predicts. GenAI creates new content.   ║
║     Traditional = discriminative (label existing data)               ║
║     Generative = creative (produce new data)                         ║
║                                                                       ║
║  Q: How does an LLM generate text?                                   ║
║  A: Autoregressively — it predicts one token at a time, using       ║
║     all previous tokens as context. Each prediction uses the        ║
║     full self-attention mechanism to consider relationships          ║
║     between all tokens in the context window.                        ║
║                                                                       ║
║  Q: What is a token?                                                  ║
║  A: A subword unit, roughly 3-4 characters. Words are split into    ║
║     tokens using BPE (Byte Pair Encoding). Important because:       ║
║     - Pricing is per token                                           ║
║     - Context window is measured in tokens                           ║
║     - Different languages have different token efficiency            ║
║                                                                       ║
║  Q: Why do LLMs hallucinate?                                          ║
║  A: They are probabilistic text generators, not knowledge bases.    ║
║     They produce statistically likely continuations, which may      ║
║     be factually wrong. Mitigations: RAG, tool use, guardrails.    ║
║                                                                       ║
║  Q: When would you choose open-source over API models?               ║
║  A: Data privacy, regulatory compliance, cost at scale,             ║
║     fine-tuning needs, or avoiding vendor lock-in.                   ║
║                                                                       ║
╚═══════════════════════════════════════════════════════════════════════╝
```
