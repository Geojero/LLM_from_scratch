## 🔥 Exercise 2: System Prompts Change Everything

Ask "What is Python?" with **3 different system prompts:**
1. `"You are a teacher explaining to a 10-year-old."`
2. `"You are a strict interviewer."`
3. `"You are a pirate."`

Observe how the SAME question gets completely different answers.

````python
# ✍️ Exercise 2: YOUR CODE HERE
# Hint: Add a {"role": "system", "content": "..."} before the user message
````

<details>
<summary>✅ Solution</summary>

```python
question = "What is Python?"
personas = [
    "You are a teacher explaining to a 10-year-old.",
    "You are a strict interviewer.",
    "You are a pirate.",
]

for persona in personas:
    response = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=[
            {"role": "system", "content": persona},
            {"role": "user", "content": question}
        ],
        temperature=0.7,
        max_tokens=100
    )
    print(f"🎭 {persona}")
    print(response.choices[0].message.content)
    print("=" * 60)
```
</details>

## 🔥 Exercise 3: Temperature Experiment

Call the model **3 times** with the prompt `"Invent a name for a coffee shop"` at each of these temperatures: **0.0, 0.7, 1.5**

Observe: At `temp=0`, are all 3 answers the same?

````python
# ✍️ Exercise 3: YOUR CODE HERE
````

<details>
<summary>✅ Solution</summary>

```python
prompt = "Invent a name for a coffee shop"

for temp in [0.0, 0.7, 1.5]:
    print(f"\n🌡️ Temperature = {temp}")
    print("-" * 40)
    for i in range(3):
        response = client.chat.completions.create(
            model="gpt-4o-mini",
            messages=[{"role": "user", "content": prompt}],
            temperature=temp,
            max_tokens=20
        )
        print(f"  {i+1}. {response.choices[0].message.content}")

import tiktoken

enc = tiktoken.encoding_for_model("gpt-4o")

texts = ["Hello world", "chatgpt", "🚀🤖💡", "supercalifragilisticexpialidocious"]

for text in texts:
    tokens = enc.encode(text)
    pieces = [enc.decode([t]) for t in tokens]
    print(f"📝 \"{text}\" → {len(tokens)} tokens → {pieces}")
```

        model="gpt-4o-mini",
        messages=[
            {"role": "system", "content": f"You are a {persona}."},
