# 1. 🏗️ Classes & Objects

### Think of it this way...

A **class** is a cookie cutter. An **object** is the cookie. You design the cutter once (the class), and then stamp out as many cookies (objects) as you want. Each cookie can have different sprinkles (attribute values), but they all share the same shape (structure).

```
╔═══════════════════════════════════════════════════════╗
║              CLASS vs OBJECT                         ║
╠═══════════════════════════════════════════════════════╣
║  Class (Blueprint)        Object (Instance)          ║
║  ─────────────────        ─────────────────          ║
║  • Template/design        • Actual thing in memory   ║
║  • Defines attributes     • Has specific values      ║
║  • Defines methods        • Can call those methods   ║
║  • Created once           • Created many times       ║
║                                                      ║
║  class Dog:               buddy = Dog("Buddy", 3)    ║
║      name, age            buddy.name → "Buddy"       ║
║      bark()               buddy.bark() → "Woof!"     ║
╚═══════════════════════════════════════════════════════╝
```

````python
# === Classes & Objects — The Basics ===

class Dog:
    # Class variable — shared by ALL instances
    species = "Canis familiaris"
    dog_count = 0
    
    def __init__(self, name: str, age: int):
        # Instance variables — unique to EACH object
        self.name = name      # each dog has its own name
        self.age = age        # each dog has its own age
        Dog.dog_count += 1    # but they all share this counter
    
    def bark(self) -> str:
        return f"{self.name} says Woof!"
    
    def __str__(self) -> str:
        """Human-friendly — what users see (print, str())"""
        return f"{self.name}, {self.age} years old"
    
    def __repr__(self) -> str:
        """Developer-friendly — unambiguous, ideally can recreate the object"""
        return f"Dog(name='{self.name}', age={self.age})"


# Creating objects (instances)
buddy = Dog("Buddy", 3)
charlie = Dog("Charlie", 5)

print(buddy.bark())          # Instance method
print(buddy.species)         # Accessing class variable through instance
print(Dog.dog_count)         # Accessing class variable through class

# __str__ vs __repr__
print(str(buddy))            # Calls __str__ → "Buddy, 3 years old"
print(repr(buddy))           # Calls __repr__ → "Dog(name='Buddy', age=3)"
print(buddy)                 # print() calls __str__ by default
````

### 🧠 `__str__` vs `__repr__` — The Quick Rule

```
╔═══════════════════════════════════════════════════════════════╗
║  __str__                        __repr__                     ║
║  ──────                         ────────                     ║
║  For end users                  For developers               ║
║  Called by print(), str()       Called by repr(), debugger    ║
║  "Pretty" output               "Unambiguous" output          ║
║  Can be informal               Should ideally recreate obj   ║
║                                                              ║
║  If only one? Implement __repr__. Python falls back to it.   ║
╚═══════════════════════════════════════════════════════════════╝
```

### Instance vs Class vs Static Methods

Here's the thing — interviewers LOVE asking the difference between these three. They're testing if you understand how Python's method binding works.

````python
# === Instance, Class, and Static Methods ===

class Pizza:
    def __init__(self, size: str, toppings: list):
        self.size = size
        self.toppings = toppings
    
    # INSTANCE METHOD — needs self, works on a specific pizza
    def describe(self) -> str:
        return f"{self.size} pizza with {', '.join(self.toppings)}"
    
    # CLASS METHOD — gets cls (the class itself), not a specific instance
    # Perfect for alternative constructors (factory methods)
    @classmethod
    def margherita(cls) -> 'Pizza':
        return cls("Medium", ["mozzarella", "tomato", "basil"])
    
    @classmethod
    def pepperoni(cls) -> 'Pizza':
        return cls("Large", ["mozzarella", "pepperoni"])
    
    # STATIC METHOD — no self, no cls. Just a utility that lives in the class
    # It's basically a regular function that's namespaced under the class
    @staticmethod
    def is_valid_size(size: str) -> bool:
        return size in ["Small", "Medium", "Large"]


# Instance method — called on an object
p = Pizza("Large", ["mushrooms", "olives"])
print(p.describe())

# Class method — called on the class (alternative constructor)
m = Pizza.margherita()
print(m.describe())

# Static method — no instance or class needed
print(Pizza.is_valid_size("Medium"))   # True
print(Pizza.is_valid_size("Huge"))     # False
````

```
╔══════════════════════════════════════════════════════════════════════╗
║  METHOD TYPE       FIRST ARG    USE CASE                           ║
╠══════════════════════════════════════════════════════════════════════╣
║  Instance method   self         Works on a specific object         ║
║  Class method      cls          Alternative constructors/factories ║
║  Static method     (none)       Utility functions in class scope   ║
╚══════════════════════════════════════════════════════════════════════╝
```

# 2. 🔒 Encapsulation

### Think of it this way...

Your car has a steering wheel (public interface), but you can't directly touch the engine ECU while driving (private internals). The steering wheel gives you **controlled access** — you turn, and the car figures out the mechanics. That's encapsulation: **hiding internal state and exposing a clean interface.**

```
╔═══════════════════════════════════════════════════════════════╗
║           PYTHON ACCESS MODIFIERS                            ║
╠═══════════════════════════════════════════════════════════════╣
║  Prefix        Convention         Example                    ║
║  ──────        ──────────         ───────                    ║
║  (none)        Public             self.name                  ║
║  _             Protected          self._name (by convention) ║
║  __            Private            self.__name (name mangled) ║
║                                                              ║
║  ⚠️ Python has NO true private! __ just triggers              ║
║    name mangling: self.__name → self._ClassName__name        ║
╚═══════════════════════════════════════════════════════════════╝
```

So here's the thing about encapsulation in Python — it's more about **convention** than enforcement. Python trusts you. The `_` prefix is a gentleman's agreement that says "hey, don't touch this from outside." The `__` prefix is Python's way of making it *slightly harder* to access, but not impossible.

````python
# === Encapsulation — Access Levels & Name Mangling ===

class BankAccount:
    def __init__(self, owner: str, balance: float):
        self.owner = owner           # Public — anyone can access
        self._account_type = "savings"  # Protected — convention only, "please don't touch"
        self.__balance = balance      # Private — name mangled to _BankAccount__balance
    
    def deposit(self, amount: float):
        if amount > 0:
            self.__balance += amount
            print(f"Deposited ${amount}. New balance: ${self.__balance}")
        else:
            print("Invalid amount!")
    
    def get_balance(self) -> float:
        return self.__balance


acc = BankAccount("Sarthak", 1000)

# Public — works fine
print(acc.owner)

# Protected — works but you shouldn't (convention)
print(acc._account_type)

# Private — this FAILS
try:
    print(acc.__balance)
except AttributeError as e:
    print(f"Can't access __balance directly: {e}")

# But name mangling means you CAN still access it (Python is trusting)
print(f"Via name mangling: {acc._BankAccount__balance}")

# The RIGHT way — use the public method
print(f"Via method: {acc.get_balance()}")
````

### 🎯 The Pythonic Way: `@property` Decorator

In Java, you'd write `getBalance()` and `setBalance()`. In Python, we use `@property` — it lets you access attributes like normal variables, but with validation behind the scenes. This is **THE** Pythonic way to do getters/setters.

````python
# === @property — Getters/Setters the Pythonic Way ===

class Temperature:
    def __init__(self, celsius: float):
        self._celsius = celsius  # protected by convention
    
    @property
    def celsius(self) -> float:
        """Getter — called when you do temp.celsius"""
        return self._celsius
    
    @celsius.setter
    def celsius(self, value: float):
        """Setter — called when you do temp.celsius = 50"""
        if value < -273.15:
            raise ValueError("Temperature below absolute zero is impossible!")
        self._celsius = value
    
    @property
    def fahrenheit(self) -> float:
        """Read-only computed property — no setter defined"""
        return self._celsius * 9/5 + 32


temp = Temperature(25)
print(f"{temp.celsius}°C = {temp.fahrenheit}°F")   # Looks like attribute access

temp.celsius = 100               # Setter validates!
print(f"{temp.celsius}°C = {temp.fahrenheit}°F")

try:
    temp.celsius = -300          # Below absolute zero!
except ValueError as e:
    print(f"Validation caught it: {e}")

# fahrenheit is read only — no setter
try:
    temp.fahrenheit = 50
except AttributeError as e:
    print(f"Read-only: {e}")
````

# 3. 🎭 Abstraction

### Think of it this way...

When you use a TV remote, you press "Volume Up" — you don't care about the infrared signals, the circuit board, or the speaker amplifier. The remote **abstracts away** the complexity and gives you a simple interface. That's abstraction: **showing only what's necessary, hiding the how.**

Encapsulation hides *data*. Abstraction hides *complexity*.

```
╔══════════════════════════════════════════════════════════════╗
║  ENCAPSULATION vs ABSTRACTION                               ║
╠══════════════════════════════════════════════════════════════╣
║  Encapsulation                  Abstraction                  ║
║  ─────────────                  ───────────                  ║
║  Hides DATA (internal state)    Hides COMPLEXITY (impl.)     ║
║  HOW: access modifiers          HOW: abstract classes/intf.  ║
║  Focus: data protection         Focus: design simplification ║
║  Example: private balance       Example: .drive() hides      ║
║  with public deposit()          engine, transmission, etc.   ║
╚══════════════════════════════════════════════════════════════╝
```

````python
# === Abstraction with ABC ===
from abc import ABC, abstractmethod


class PaymentProcessor(ABC):
    """Abstract class — defines WHAT subclasses must do, not HOW.
    You CANNOT instantiate this directly."""
    
    @abstractmethod
    def process_payment(self, amount: float) -> bool:
        """Every payment processor MUST implement this"""
        pass
    
    @abstractmethod
    def refund(self, amount: float) -> bool:
        """Every payment processor MUST implement this"""
        pass
    
    def log_transaction(self, message: str):
        """Concrete method — shared by all subclasses (not abstract)"""
        print(f"[LOG] {message}")


class CreditCardProcessor(PaymentProcessor):
    def process_payment(self, amount: float) -> bool:
        self.log_transaction(f"Credit card charged ${amount}")
        return True
    
    def refund(self, amount: float) -> bool:
        self.log_transaction(f"Credit card refunded ${amount}")
        return True


class UPIProcessor(PaymentProcessor):
    def process_payment(self, amount: float) -> bool:
        self.log_transaction(f"UPI payment of ₹{amount}")
        return True
    
    def refund(self, amount: float) -> bool:
        self.log_transaction(f"UPI refund of ₹{amount}")
        return True


# Can't instantiate abstract class
try:
    p = PaymentProcessor()
except TypeError as e:
    print(f"Can't create abstract class: {e}")

# But concrete subclasses work fine
cc = CreditCardProcessor()
cc.process_payment(99.99)

upi = UPIProcessor()
upi.process_payment(500)
````

### Python Doesn't Have "Interfaces" — But You Can Fake It

Java has `interface`. Python doesn't. But an ABC with *only* abstract methods (no concrete ones) is essentially an interface.

```python
# This is Python's version of an "interface"
class Printable(ABC):
    @abstractmethod
    def print(self) -> str:
        pass

class Saveable(ABC):
    @abstractmethod 
    def save(self) -> None:
        pass

# A class can implement multiple "interfaces"
class Document(Printable, Saveable):
    def print(self) -> str:
        return "Printing document..."
    def save(self) -> None:
        print("Saving document...")
```

```
┌──────────────────────────────────────────────────────────────┐
│ 💡 Interview Tip:                                           │
│                                                             │
│ "Python doesn't have interfaces, but you can simulate them  │
│  using ABC with only abstract methods. Unlike Java, Python  │
│  supports multiple inheritance, so a class can 'implement'  │
│  multiple ABCs — which is how you get interface-like        │
│  behavior."                                                 │
│                                                             │
```

# 4. 🧬 Inheritance

### Think of it this way...

Inheritance is like a family tree. Your grandparents have certain traits → your parents inherit some and add their own → you inherit from your parents and add yours. Each generation **reuses** what came before and **extends** it.

```
╔══════════════════════════════════════════════════════════╗
║           TYPES OF INHERITANCE                          ║
╠══════════════════════════════════════════════════════════╣
║                                                         ║
║  Single:      A → B                                     ║
║  Multilevel:  A → B → C                                 ║
║  Multiple:    A, B → C  (C inherits from BOTH A and B)  ║
║  Hierarchical: A → B, A → C                             ║
║                                                         ║
║  ⚠️ Multiple inheritance can cause the Diamond Problem   ║
╚══════════════════════════════════════════════════════════╝
```

````python
# === Single & Multilevel Inheritance ===

class Animal:
    def __init__(self, name: str):
        self.name = name
    
    def speak(self) -> str:
        return "Some generic sound"


class Dog(Animal):  # Single inheritance
    def speak(self) -> str:
        return f"{self.name} says Woof!"


class GuideDog(Dog):  # Multilevel: Animal → Dog → GuideDog
    def __init__(self, name: str, handler: str):
        super().__init__(name)   # Calls Dog.__init__ → Animal.__init__
        self.handler = handler
    
    def guide(self) -> str:
        return f"{self.name} is guiding {self.handler}"


rex = GuideDog("Rex", "John")
print(rex.speak())    # Inherited from Dog
print(rex.guide())    # Own method
print(isinstance(rex, Animal))  # True — rex IS an Animal too
````

### 💎 The Diamond Problem & MRO

This is one of the **most asked** inheritance questions. Here's the scenario:

```
╔═══════════════════════════════════════════════════╗
║          THE DIAMOND PROBLEM                      ║
║                                                   ║
║              A                                    ║
║             / \                                   ║
║            B   C         D inherits from B and C  ║
║             \ /          Both B and C inherit A    ║
║              D                                    ║
║                                                   ║
║  When D calls a method, which path?               ║
║  D → B → A?  or  D → C → A?                      ║
║                                                   ║
║  Python solves this with C3 Linearization (MRO)   ║
║  MRO for D: D → B → C → A → object               ║
╚═══════════════════════════════════════════════════╝
```

````python
# === Diamond Problem & MRO (Method Resolution Order) ===

class A:
    def greet(self):
        return "Hello from A"

class B(A):
    def greet(self):
        return "Hello from B"

class C(A):
    def greet(self):
        return "Hello from C"

class D(B, C):  # Multiple inheritance — diamond!
    pass


d = D()
print(d.greet())    # "Hello from B" — B comes first in MRO

# Check the MRO (Method Resolution Order)
print("\nMRO:", [cls.__name__ for cls in D.__mro__])
# → D → B → C → A → object

# super() follows MRO, not just the parent!
class E(B, C):
    def greet(self):
        # super() goes to B (next in MRO), not A!
        return f"E says hi, then {super().greet()}"

e = E()
print(f"\n{e.greet()}")   # "E says hi, then Hello from B"
````

### ⚡ Composition vs Inheritance — "Prefer Composition"

This is a **critical** interview topic. The mantra is: *"Favor composition over inheritance."*

```
╔═════════════════════════════════════════════════════════════════╗
║  INHERITANCE (IS-A)              COMPOSITION (HAS-A)           ║
╠═════════════════════════════════════════════════════════════════╣
║  Dog IS-A Animal                 Car HAS-A Engine              ║
║  Tight coupling                  Loose coupling                ║
║  Changes in parent break child   Components can be swapped     ║
║  Hierarchy gets deep & messy     Flat and flexible             ║
║  Hard to change later            Easy to change later          ║
╚═════════════════════════════════════════════════════════════════╝
```

**Rule of thumb**: Use inheritance for genuine "is-a" relationships (Dog is-a Animal). Use composition when you want to USE functionality, not BE something.

````python
# === Composition over Inheritance ===

# BAD: Inheritance when it should be composition
# class Car(Engine):  ← A Car IS NOT an Engine!

# GOOD: Composition
class Engine:
    def __init__(self, horsepower: int):
        self.horsepower = horsepower
    
    def start(self) -> str:
        return f"Engine ({self.horsepower}hp) started"


class GPS:
    def navigate(self, destination: str) -> str:
        return f"Navigating to {destination}..."


class Car:
    def __init__(self, model: str, engine: Engine, gps: GPS = None):
        self.model = model
        self.engine = engine    # HAS-A engine
        self.gps = gps          # HAS-A gps (optional!)
    
    def drive(self, destination: str = None) -> str:
        result = f"{self.model}: {self.engine.start()}"
        if self.gps and destination:
            result += f" | {self.gps.navigate(destination)}"
        return result


# You can swap components easily!
v8 = Engine(450)
electric = Engine(670)
gps = GPS()

muscle_car = Car("Mustang", v8)
tesla = Car("Model S", electric, gps)

print(muscle_car.drive())
print(tesla.drive("Bangalore"))
````

```
┌──────────────────────────────────────────────────────────────┐
│ 💡 Interview Tip:                                           │
│                                                             │
│ "What is MRO?" → Method Resolution Order — the order        │
│ Python searches for methods in multiple inheritance.        │
│ Uses C3 linearization. Check with ClassName.__mro__         │
│                                                             │
│ "When to use inheritance vs composition?"                   │
│ → Inheritance for IS-A (Dog is Animal)                      │
│ → Composition for HAS-A (Car has Engine)                    │
│ → When in doubt, prefer composition — it's more flexible.  │

```

# 5. 🦆 Polymorphism

### Think of it this way...

Poly = many, morph = forms. Same interface, different behavior. Think of a USB port — you plug in a keyboard, mouse, or phone charger. The port doesn't care WHAT device it is. It just provides power and data. Each device does something different with it. That's polymorphism.

```
╔═══════════════════════════════════════════════════════════╗
║         TYPES OF POLYMORPHISM IN PYTHON                  ║
╠═══════════════════════════════════════════════════════════╣
║                                                          ║
║  1. Method Overriding  — Child redefines parent method   ║
║  2. Duck Typing        — "If it quacks like a duck..."   ║
║  3. Operator Overload  — Custom +, ==, len(), etc.       ║
║                                                          ║
║  ⚠️ Python does NOT have method overloading like Java     ║
║    (same method name, different params). Use *args or     ║
║    default params instead.                                ║
╚═══════════════════════════════════════════════════════════╝
```

````python
# === Method Overriding ===

class Shape:
    def area(self) -> float:
        raise NotImplementedError("Subclasses must implement area()")

class Circle(Shape):
    def __init__(self, radius: float):
        self.radius = radius
    def area(self) -> float:
        return 3.14159 * self.radius ** 2

class Rectangle(Shape):
    def __init__(self, width: float, height: float):
        self.width = width
        self.height = height
    def area(self) -> float:
        return self.width * self.height


# Same method name, different behavior — THAT'S polymorphism!
shapes = [Circle(5), Rectangle(4, 6), Circle(3)]

for shape in shapes:
    # We don't care WHAT shape it is — we just call .area()
    print(f"{shape.__class__.__name__}: area = {shape.area():.2f}")
````

````python
# === Duck Typing — "If it walks like a duck and quacks like a duck..." ===

# In Python, we don't check types. We check BEHAVIOR.
# If an object has the method we need, we use it. Period.

class Duck:
    def quack(self):
        return "Quack quack!"

class Person:
    def quack(self):
        return "I'm quacking like a duck!"

class RubberDuck:
    def quack(self):
        return "Squeak!"


def make_it_quack(thing):  # No type checking! Just call .quack()
    print(thing.quack())

# All of these work — Python doesn't care about the TYPE,
# only that the object has a .quack() method
make_it_quack(Duck())
make_it_quack(Person())
make_it_quack(RubberDuck())
````

````python
# === Operator Overloading (Dunder/Magic Methods) ===

class Vector:
    def __init__(self, x: float, y: float):
        self.x = x
        self.y = y
    
    def __add__(self, other: 'Vector') -> 'Vector':
        """v1 + v2"""
        return Vector(self.x + other.x, self.y + other.y)
    
    def __eq__(self, other: 'Vector') -> bool:
        """v1 == v2"""
        return self.x == other.x and self.y == other.y
    
    def __len__(self) -> int:
        """len(v) — returns magnitude as int"""
        return int((self.x**2 + self.y**2) ** 0.5)
    
    def __repr__(self) -> str:
        return f"Vector({self.x}, {self.y})"


v1 = Vector(2, 3)
v2 = Vector(4, 1)

print(f"v1 + v2 = {v1 + v2}")     # __add__
print(f"v1 == v2? {v1 == v2}")     # __eq__
print(f"len(v1) = {len(v1)}")      # __len__
print(f"v1 == Vector(2,3)? {v1 == Vector(2, 3)}")  # True!
````

# 6. 🏛️ SOLID Principles

SOLID is the bedrock of clean, maintainable code. Interviewers ask about these ALL the time — especially for mid/senior roles. Let's break each one down with real examples.

```
╔══════════════════════════════════════════════════════════════════╗
║  S — Single Responsibility    One class, one reason to change  ║
║  O — Open/Closed              Open for extension, closed for   ║
║                                modification                    ║
║  L — Liskov Substitution      Subtypes must be substitutable   ║
║  I — Interface Segregation    Many small interfaces > one fat  ║
║  D — Dependency Inversion     Depend on abstractions, not      ║
║                                concretions                     ║
╚══════════════════════════════════════════════════════════════════╝
```

## S — Single Responsibility Principle (SRP)

### Think of it this way...

A chef should cook. A waiter should serve. If the chef is also handling bills, cleaning tables, and managing inventory — something's wrong. Each person (class) should have **one job, one reason to change**.

````python
# === S: Single Responsibility Principle ===

# ❌ BAD — This class does TOO MANY things
class UserManagerBad:
    def create_user(self, name, email):
        # Creates user in DB
        pass
    
    def send_welcome_email(self, email):
        # Why is user management sending emails??
        pass
    
    def generate_report(self):
        # And generating reports too?? 😬
        pass


# ✅ GOOD — Each class has ONE responsibility
class UserService:
    """Only handles user CRUD operations"""
    def create_user(self, name: str, email: str) -> dict:
        user = {"name": name, "email": email}
        print(f"User created: {name}")
        return user

class EmailService:
    """Only handles email sending"""
    def send_welcome_email(self, email: str):
        print(f"Welcome email sent to {email}")

class ReportService:
    """Only handles report generation"""
    def generate_user_report(self):
        print("User report generated")


# Usage — each service does its OWN thing
user_svc = UserService()
email_svc = EmailService()

user = user_svc.create_user("Sarthak", "sarthak@example.com")
email_svc.send_welcome_email(user["email"])

print("\n✅ Now if email logic changes, only EmailService is modified.")
print("   UserService stays untouched. That's SRP!")
````

## O — Open/Closed Principle (OCP)

### Think of it this way...

Think of a plugin system. Your music player can play MP3 by default. To add FLAC support, you don't rewrite the player — you **add a FLAC plugin**. The player is **open for extension** (new plugins) but **closed for modification** (don't touch existing code).

````python
# === O: Open/Closed Principle ===
from abc import ABC, abstractmethod

# ❌ BAD — Every new discount type means modifying this class
class DiscountCalculatorBad:
    def calculate(self, price: float, discount_type: str) -> float:
        if discount_type == "percentage":
            return price * 0.9
        elif discount_type == "flat":
            return price - 10
        elif discount_type == "bogo":
            return price * 0.5
        # Every new type = modifying this method = violating OCP!


# ✅ GOOD — Add new discounts WITHOUT touching existing code
class Discount(ABC):
    @abstractmethod
    def apply(self, price: float) -> float:
        pass

class PercentageDiscount(Discount):
    def __init__(self, percent: float):
        self.percent = percent
    def apply(self, price: float) -> float:
        return price * (1 - self.percent / 100)

class FlatDiscount(Discount):
    def __init__(self, amount: float):
        self.amount = amount
    def apply(self, price: float) -> float:
        return max(0, price - self.amount)

# Adding a NEW discount — zero changes to existing classes!
class BOGODiscount(Discount):
    def apply(self, price: float) -> float:
        return price * 0.5


# The calculator doesn't need to know WHICH discount — just that it IS one
def checkout(price: float, discount: Discount) -> float:
    final = discount.apply(price)
    print(f"Original: ${price:.2f} → After discount: ${final:.2f}")
    return final

checkout(100, PercentageDiscount(20))
checkout(100, FlatDiscount(15))
checkout(100, BOGODiscount())
````

## L — Liskov Substitution Principle (LSP)

### Think of it this way...

If your code works with a `Bird` object, it should work with ANY subclass of `Bird` — `Sparrow`, `Eagle`, etc. — without breaking. If you have a `Penguin` subclass and calling `fly()` on it throws an error... **you've violated LSP**. The child must honor the parent's contract.

```
╔═══════════════════════════════════════════════════════════════╗
║  LISKOV SUBSTITUTION — The Rule:                             ║
║                                                              ║
║  If S is a subtype of T, then objects of type T can be       ║
║  replaced with objects of type S WITHOUT altering the        ║
║  correctness of the program.                                 ║
║                                                              ║
║  Simple: Child should be USABLE wherever parent is expected. ║
╚═══════════════════════════════════════════════════════════════╝
```

````python
# === L: Liskov Substitution Principle ===

# ❌ BAD — Penguin violates the Bird contract
class BirdBad:
    def fly(self):
        return "Flying!"

class PenguinBad(BirdBad):
    def fly(self):
        raise Exception("Penguins can't fly!")  # 💥 Breaks LSP!


# ✅ GOOD — Separate the contract properly
class Bird(ABC):
    @abstractmethod
    def move(self) -> str:
        pass

class FlyingBird(Bird):
    def move(self) -> str:
        return "Flying through the sky!"

class SwimmingBird(Bird):
    def move(self) -> str:
        return "Swimming in the ocean!"

class Eagle(FlyingBird):
    pass

class Penguin(SwimmingBird):
    pass


# Now ANY Bird can be used in this function — no surprises!
def make_bird_move(bird: Bird):
    print(f"{bird.__class__.__name__}: {bird.move()}")

make_bird_move(Eagle())
make_bird_move(Penguin())
# Both work! No exceptions! That's LSP. ✅
````

## I — Interface Segregation Principle (ISP)

### Think of it this way...

Imagine a restaurant menu with 500 items. Overwhelming, right? Now imagine separate menus: breakfast, lunch, dinner, desserts. Much better! ISP says: **don't force a class to implement methods it doesn't need**. Split fat interfaces into smaller, focused ones.

````python
# === I: Interface Segregation Principle ===

# ❌ BAD — One fat interface forces everything on everyone
class WorkerBad(ABC):
    @abstractmethod
    def work(self): pass
    @abstractmethod
    def eat(self): pass
    @abstractmethod
    def sleep(self): pass
    @abstractmethod
    def code(self): pass
    @abstractmethod
    def manage(self): pass

# A robot worker would have to implement eat() and sleep()?? 🤔


# ✅ GOOD — Small, focused interfaces
class Workable(ABC):
    @abstractmethod
    def work(self) -> str: pass

class Eatable(ABC):
    @abstractmethod
    def eat(self) -> str: pass

class Codeable(ABC):
    @abstractmethod
    def code(self) -> str: pass

class Manageable(ABC):
    @abstractmethod
    def manage(self) -> str: pass


# Each class implements ONLY what it needs
class Developer(Workable, Eatable, Codeable):
    def work(self) -> str: return "Working on sprint tasks"
    def eat(self) -> str: return "Eating lunch"
    def code(self) -> str: return "Writing Python code"

class Manager(Workable, Eatable, Manageable):
    def work(self) -> str: return "In meetings all day"
    def eat(self) -> str: return "Business lunch"
    def manage(self) -> str: return "Managing the team"

class Robot(Workable, Codeable):
    def work(self) -> str: return "Processing data 24/7"
    def code(self) -> str: return "Generating code"
    # No eat() or sleep() — Robot doesn't need them! ✅


dev = Developer()
robot = Robot()
print(f"Dev: {dev.code()}")
print(f"Robot: {robot.code()}")
print(f"\n✅ Robot doesn't need eat() or sleep() — and it doesn't have them!")
````

## D — Dependency Inversion Principle (DIP)

### Think of it this way...

Your laptop charger doesn't care if it's plugged into a wall outlet, a power strip, or a generator. It depends on the **abstraction** (electricity via a standard plug), not the **concrete source**. If your laptop only worked with one specific outlet, that would be terrible design.

```
╔══════════════════════════════════════════════════════════════╗
║  DEPENDENCY INVERSION                                       ║
║                                                             ║
║  ❌ High-level → depends on → Low-level (concrete)          ║
║  ✅ High-level → depends on → Abstraction ← Low-level      ║
║                                                             ║
║  Both high-level and low-level depend on abstractions.      ║
║  Abstractions don't depend on details.                      ║
╚══════════════════════════════════════════════════════════════╝
```

````python
# === D: Dependency Inversion Principle ===

# ❌ BAD — NotificationService is hardcoded to use Email
class EmailSenderBad:
    def send(self, message: str):
        print(f"Email: {message}")

class NotificationServiceBad:
    def __init__(self):
        self.sender = EmailSenderBad()  # 💥 Tightly coupled to Email!
    
    def notify(self, message: str):
        self.sender.send(message)
    # What if we want SMS? Push notifications? We'd have to MODIFY this class.


# ✅ GOOD — Depend on abstraction, inject the dependency
class MessageSender(ABC):
    @abstractmethod
    def send(self, message: str) -> None:
        pass

class EmailSender(MessageSender):
    def send(self, message: str):
        print(f"📧 Email: {message}")

class SMSSender(MessageSender):
    def send(self, message: str):
        print(f"📱 SMS: {message}")

class PushNotificationSender(MessageSender):
    def send(self, message: str):
        print(f"🔔 Push: {message}")


class NotificationService:
    def __init__(self, sender: MessageSender):  # Depends on ABSTRACTION!
        self.sender = sender
    
    def notify(self, message: str):
        self.sender.send(message)


# Now we can swap implementations without changing NotificationService!
email_notifier = NotificationService(EmailSender())
sms_notifier = NotificationService(SMSSender())
push_notifier = NotificationService(PushNotificationSender())

email_notifier.notify("Your order is confirmed!")
sms_notifier.notify("Your OTP is 847291")
push_notifier.notify("New message from Sarthak")

print("\n✅ NotificationService doesn't know or care HOW the message is sent.")
print("   It just knows the sender CAN send. That's DIP!")
````

### 🧠 The Big Picture

Everything connects:
- **Abstraction** enables **OCP** and **DIP** (you extend via abstractions)
- **Encapsulation** supports **SRP** (each class protects its own data)
- **Polymorphism** makes **LSP** possible (subtypes behave like parents)
- **Composition** often beats **Inheritance** for flexibility
- **DIP** = inject abstractions → automatically satisfies **OCP**

You don't memorize SOLID — you *understand* why each principle exists, and then the code writes itself.
