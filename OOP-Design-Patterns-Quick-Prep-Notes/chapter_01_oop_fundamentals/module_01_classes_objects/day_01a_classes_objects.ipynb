## 🎭 The Analogy

Think of a **Class** like a **cookie cutter** and **Objects** like the **cookies**:

- **Class** = The cookie cutter (blueprint/template) — defines shape, size
- **Object** = Each individual cookie — made from the cutter, but can have different toppings
- **Attributes** = Toppings (chocolate chips, sprinkles) — data each cookie carries
- **Methods** = Actions (eat, decorate, package) — what you can DO with a cookie

One cutter → many cookies. One class → many objects.

## 💻 Code — Build Classes from Memory

````python
# ============================================
# 1. Basic Class — The Foundation
# ============================================

class User:
    """A simple User class — know this cold."""
    
    def __init__(self, name: str, email: str):
        # Instance variables — each object gets its own copy
        self.name = name
        self.email = email
        self.is_active = True  # default value
    
    def deactivate(self):
        """Instance method — operates on self."""
        self.is_active = False
        print(f"{self.name} deactivated")
    
    def __str__(self):
        """Human-readable — for print() and str()."""
        return f"User({self.name}, {self.email})"
    
    def __repr__(self):
        """Developer-readable — for debugging, repr()."""
        return f"User(name='{self.name}', email='{self.email}', active={self.is_active})"


# Create objects
u1 = User("Sarthak", "sarthak@mail.com")
u2 = User("Alice", "alice@mail.com")

print(u1)           # calls __str__
print(repr(u2))     # calls __repr__
u1.deactivate()
print(f"u1 active? {u1.is_active}")  # False
print(f"u2 active? {u2.is_active}")  # True — separate objects!
````

````python
# ============================================
# 2. Instance vs Class Variables
# ============================================

class Employee:
    # CLASS variable — shared across ALL instances
    company = "Hire10x"
    employee_count = 0
    
    def __init__(self, name: str, salary: float):
        # INSTANCE variables — unique to each object
        self.name = name
        self.salary = salary
        Employee.employee_count += 1  # modify via CLASS name
    
    def __str__(self):
        return f"{self.name} @ {self.company} (₹{self.salary:,.0f})"


e1 = Employee("Sarthak", 1500000)
e2 = Employee("Alice", 1200000)

print(e1)  # Sarthak @ Hire10x
print(e2)  # Alice @ Hire10x
print(f"Total employees: {Employee.employee_count}")  # 2

# Change class variable — affects ALL instances
Employee.company = "Google"
print(e1)  # Sarthak @ Google
print(e2)  # Alice @ Google

# ⚠️ GOTCHA: Assigning via instance creates a NEW instance variable
e1.company = "Meta"  # creates instance var, doesn't change class var
print(e1)  # Sarthak @ Meta (instance var)
print(e2)  # Alice @ Google (still class var)
````

## 🔧 Instance vs Class vs Static Variables

```
╔═══════════════╦══════════════════════════════════════════════════╗
║  Type          ║  Description                                    ║
╠═══════════════╬══════════════════════════════════════════════════╣
║  Instance var  ║  self.name — unique per object                  ║
║  Class var     ║  company = "X" — shared by all objects          ║
║                ║  Modify via ClassName.var, not self.var         ║
╠═══════════════╬══════════════════════════════════════════════════╣
║  GOTCHA        ║  self.class_var = x creates NEW instance var!  ║
║                ║  Use ClassName.var = x to modify shared var     ║
╚═══════════════╩══════════════════════════════════════════════════╝
```

---

````python
# ============================================
# 3. Three Types of Methods
# ============================================

class Pizza:
    # Class variable
    restaurant = "Pizza Palace"
    
    def __init__(self, size: str, toppings: list):
        self.size = size
        self.toppings = toppings
    
    # INSTANCE METHOD — has access to self (instance) and cls (via self.__class__)
    def describe(self):
        return f"{self.size} pizza with {', '.join(self.toppings)} from {self.restaurant}"
    
    # CLASS METHOD — has access to cls (class), not instance
    # Use for: alternative constructors, modifying class state
    @classmethod
    def margherita(cls, size: str):
        """Alternative constructor — factory pattern."""
        return cls(size, ["mozzarella", "tomato", "basil"])
    
    @classmethod
    def change_restaurant(cls, new_name: str):
        cls.restaurant = new_name
    
    # STATIC METHOD — no access to self or cls
    # Use for: utility functions that logically belong to the class
    @staticmethod
    def calculate_price(size: str, num_toppings: int) -> float:
        """Pure function — doesn't need instance or class data."""
        base = {"small": 200, "medium": 350, "large": 500}
        return base.get(size, 350) + (num_toppings * 50)


# Instance method
p1 = Pizza("large", ["pepperoni", "mushroom"])
print(p1.describe())

# Class method as alternative constructor
p2 = Pizza.margherita("medium")
print(p2.describe())

# Static method — no instance needed
price = Pizza.calculate_price("large", 3)
print(f"Price: ₹{price}")
````

> **Interview Tip:** "I use `@classmethod` for factory methods (alternative constructors) and `@staticmethod` for utility functions that logically belong to the class but don't need instance or class data."

````python
# ============================================
# 4. __str__ vs __repr__ — Know the Difference
# ============================================

class BankAccount:
    def __init__(self, owner: str, balance: float):
        self.owner = owner
        self.balance = balance
    
    def __str__(self):
        """For end users — readable."""
        return f"{self.owner}'s account: ₹{self.balance:,.2f}"
    
    def __repr__(self):
        """For developers — unambiguous, ideally eval()-able."""
        return f"BankAccount(owner='{self.owner}', balance={self.balance})"


acc = BankAccount("Sarthak", 50000)

print(str(acc))    # __str__: Sarthak's account: ₹50,000.00
print(repr(acc))   # __repr__: BankAccount(owner='Sarthak', balance=50000)

# Where each is used:
print(acc)          # print() calls __str__
print([acc])        # In containers, Python calls __repr__
print(f"{acc}")     # f-strings call __str__
print(f"{acc!r}")   # !r forces __repr__
````

````python
# ============================================
# 5. Property Decorator — Pythonic Getters/Setters
# ============================================

class Temperature:
    def __init__(self, celsius: float):
        self._celsius = celsius  # convention: _ = "private"
    
    @property
    def celsius(self):
        """Getter — accessed like an attribute."""
        return self._celsius
    
    @celsius.setter
    def celsius(self, value: float):
        """Setter — validation on assignment."""
        if value < -273.15:
            raise ValueError("Below absolute zero!")
        self._celsius = value
    
    @property
    def fahrenheit(self):
        """Computed property — read-only."""
        return self._celsius * 9/5 + 32


t = Temperature(100)
print(f"{t.celsius}°C = {t.fahrenheit}°F")  # 100°C = 212.0°F

t.celsius = 0  # uses setter
print(f"{t.celsius}°C = {t.fahrenheit}°F")  # 0°C = 32.0°F

try:
    t.celsius = -300  # setter validates!
except ValueError as e:
    print(f"Error: {e}")
````

````python
# ============================================
# 6. Dunder Methods — Make Your Objects Pythonic
# ============================================

class Vector:
    """Interview favorite — implement a custom class with operators."""
    
    def __init__(self, x: float, y: float):
        self.x = x
        self.y = y
    
    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)
    
    def __sub__(self, other):
        return Vector(self.x - other.x, self.y - other.y)
    
    def __mul__(self, scalar):
        return Vector(self.x * scalar, self.y * scalar)
    
    def __eq__(self, other):
        return self.x == other.x and self.y == other.y
    
    def __len__(self):
        """Magnitude as integer."""
        return int((self.x**2 + self.y**2) ** 0.5)
    
    def __bool__(self):
        """Zero vector is falsy."""
        return self.x != 0 or self.y != 0
    
    def __repr__(self):
        return f"Vector({self.x}, {self.y})"


v1 = Vector(3, 4)
v2 = Vector(1, 2)

print(f"v1 + v2 = {v1 + v2}")     # Vector(4, 6)
print(f"v1 - v2 = {v1 - v2}")     # Vector(2, 2)
print(f"v1 * 3  = {v1 * 3}")      # Vector(9, 12)
print(f"v1 == v2? {v1 == v2}")     # False
print(f"|v1| = {len(v1)}")         # 5 (3-4-5 triangle)
print(f"bool(v1) = {bool(v1)}")    # True
print(f"bool(Vector(0,0)) = {bool(Vector(0, 0))}")  # False
````

## 📋 Essential Dunder Methods

```
╔══════════════════╦═══════════════════════════════════════════════╗
║  Dunder Method    ║  What It Does                                ║
╠══════════════════╬═══════════════════════════════════════════════╣
║  __init__         ║  Constructor — initialize instance           ║
║  __str__          ║  print(obj) — human-readable                 ║
║  __repr__         ║  repr(obj) — developer-readable              ║
║  __len__          ║  len(obj)                                     ║
║  __bool__         ║  bool(obj) — truthiness                      ║
║  __eq__           ║  obj1 == obj2                                 ║
║  __lt__/__gt__    ║  obj1 < obj2 / obj1 > obj2                   ║
║  __add__          ║  obj1 + obj2                                  ║
║  __getitem__      ║  obj[key]                                     ║
║  __setitem__      ║  obj[key] = val                               ║
║  __contains__     ║  x in obj                                     ║
║  __iter__         ║  for x in obj                                 ║
║  __call__         ║  obj() — make instance callable               ║
║  __hash__         ║  hash(obj) — needed for sets/dict keys        ║
╚══════════════════╩═══════════════════════════════════════════════╝
```

> **Interview Tip:** "If I implement `__eq__`, I should also implement `__hash__` if I want my objects usable as dict keys or in sets. Python sets `__hash__` to None when you define `__eq__`."

````python
# ============================================
# 7. Complete Interview Example — Design a Class
# ============================================
# Interview: "Design a BankAccount class with deposit, withdraw, transfer"

class BankAccount:
    _next_id = 1  # class variable for auto-incrementing IDs
    
    def __init__(self, owner: str, balance: float = 0):
        self.id = BankAccount._next_id
        BankAccount._next_id += 1
        self.owner = owner
        self._balance = balance  # "private" — use property
        self._transactions = []  # audit trail
    
    @property
    def balance(self):
        return self._balance
    
    def deposit(self, amount: float):
        if amount <= 0:
            raise ValueError("Deposit must be positive")
        self._balance += amount
        self._transactions.append(f"+₹{amount:,.2f}")
        return self
    
    def withdraw(self, amount: float):
        if amount <= 0:
            raise ValueError("Withdrawal must be positive")
        if amount > self._balance:
            raise ValueError("Insufficient funds")
        self._balance -= amount
        self._transactions.append(f"-₹{amount:,.2f}")
        return self
    
    def transfer(self, other: 'BankAccount', amount: float):
        self.withdraw(amount)
        other.deposit(amount)
        print(f"₹{amount:,.2f} transferred: {self.owner} → {other.owner}")
    
    @classmethod
    def from_dict(cls, data: dict):
        """Factory method — alternative constructor."""
        return cls(data["owner"], data.get("balance", 0))
    
    def __str__(self):
        return f"Account#{self.id} ({self.owner}): ₹{self._balance:,.2f}"
    
    def __repr__(self):
        return f"BankAccount(id={self.id}, owner='{self.owner}', balance={self._balance})"


# Demo
acc1 = BankAccount("Sarthak", 10000)
acc2 = BankAccount("Alice")

acc1.deposit(5000).withdraw(2000)  # method chaining!
print(acc1)  # ₹13,000

acc1.transfer(acc2, 3000)
print(acc1)  # ₹10,000
print(acc2)  # ₹3,000

# Factory method
acc3 = BankAccount.from_dict({"owner": "Bob", "balance": 50000})
print(acc3)
````

## 📝 Interview Questions — Practice These

| # | Question | Key Point |
|---|----------|------------|
| 1 | Class vs Object? | Class = blueprint (one in memory). Object = instance (many, each with own data) |
| 2 | What is `self`? | Reference to the current instance. Python passes it implicitly when you call a method |
| 3 | Instance vs class variable? | Instance = per object (self.x). Class = shared (Cls.x). Assigning via self creates new instance var |
| 4 | `@classmethod` vs `@staticmethod`? | classmethod gets cls, used for factory methods. staticmethod gets nothing, used for utility functions |
| 5 | `__str__` vs `__repr__`? | str = human-readable (print). repr = developer-readable (debug, ideally eval-able) |
| 6 | What is `@property`? | Pythonic way to add getters/setters. Access like attribute, but runs a method underneath for validation |
| 7 | What are dunder methods? | Special methods (__xx__) that let your objects work with Python operators (+, ==, len, in, etc.) |
| 8 | How to make a class iterable? | Implement __iter__ (returns iterator) and __next__ (returns next value) |
| 9 | What happens if you define __eq__ but not __hash__? | Python sets __hash__ to None → objects become unhashable → can't use in sets or as dict keys |
| 10 | What's the MRO? | Method Resolution Order — the order Python searches for methods in inheritance. Use ClassName.mro() to see it |
