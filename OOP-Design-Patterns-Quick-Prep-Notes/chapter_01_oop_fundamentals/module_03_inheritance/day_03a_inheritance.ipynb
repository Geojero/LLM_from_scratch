````python
# ============================================
# 1. Single Inheritance — The Foundation
# ============================================

class Animal:
    def __init__(self, name: str, species: str):
        self.name = name
        self.species = species
    
    def speak(self) -> str:
        return f"{self.name} makes a sound"
    
    def __str__(self):
        return f"{self.name} ({self.species})"


class Dog(Animal):
    def __init__(self, name: str, breed: str):
        # super() calls parent's __init__
        super().__init__(name, species="Dog")
        self.breed = breed  # child-specific attribute
    
    def speak(self) -> str:
        """Override parent method."""
        return f"{self.name} says Woof! 🐕"
    
    def fetch(self, item: str) -> str:
        """Child-specific method."""
        return f"{self.name} fetches the {item}"


# Dog inherits name, species, __str__ from Animal
# Dog overrides speak()
# Dog adds breed and fetch()

dog = Dog("Buddy", "Golden Retriever")
print(dog)              # __str__ inherited from Animal
print(dog.speak())      # overridden method
print(dog.fetch("ball")) # child-specific method
print(f"Is Dog an Animal? {isinstance(dog, Animal)}")  # True
````

````python
# ============================================
# 2. super() — Extending, Not Replacing
# ============================================

class Employee:
    def __init__(self, name: str, employee_id: int):
        self.name = name
        self.employee_id = employee_id
    
    def get_info(self) -> str:
        return f"ID: {self.employee_id}, Name: {self.name}"


class Manager(Employee):
    def __init__(self, name: str, employee_id: int, department: str):
        super().__init__(name, employee_id)  # reuse parent's init
        self.department = department
        self.team = []
    
    def get_info(self) -> str:
        # EXTEND parent's method, don't replace it
        base_info = super().get_info()
        return f"{base_info}, Dept: {self.department}, Team size: {len(self.team)}"
    
    def add_to_team(self, employee: Employee):
        self.team.append(employee)


emp = Employee("Alice", 101)
mgr = Manager("Sarthak", 100, "Engineering")
mgr.add_to_team(emp)

print(emp.get_info())  # base info only
print(mgr.get_info())  # extended info with dept and team
````

````python
# ============================================
# 3. Multiple Inheritance & The Diamond Problem
# ============================================

class A:
    def greet(self):
        return "Hello from A"

class B(A):
    def greet(self):
        return "Hello from B"

class C(A):
    def greet(self):
        return "Hello from C"

class D(B, C):  # Multiple inheritance — diamond!
    pass


d = D()
print(d.greet())  # "Hello from B" — WHY?

# MRO = Method Resolution Order
# Python uses C3 Linearization to determine the order
print(f"\nMRO of D: {[cls.__name__ for cls in D.__mro__]}")
# D → B → C → A → object
# Python searches LEFT to RIGHT in the inheritance list
# D(B, C) → B is checked before C

print("\n--- MRO explained ---")
for i, cls in enumerate(D.__mro__):
    print(f"  {i+1}. {cls.__name__}")
````

````python
# ============================================
# 4. Cooperative Multiple Inheritance with super()
# ============================================

class Base:
    def __init__(self):
        print("Base.__init__")

class Left(Base):
    def __init__(self):
        print("Left.__init__")
        super().__init__()  # calls NEXT in MRO, not necessarily Base!

class Right(Base):
    def __init__(self):
        print("Right.__init__")
        super().__init__()

class Child(Left, Right):
    def __init__(self):
        print("Child.__init__")
        super().__init__()


print("MRO:", [c.__name__ for c in Child.__mro__])
print()
c = Child()
# Output:
# Child.__init__
# Left.__init__    ← super() in Child calls Left (next in MRO)
# Right.__init__   ← super() in Left calls Right (NOT Base!)
# Base.__init__    ← super() in Right calls Base
````

````python
# ============================================
# 5. isinstance() vs type() vs issubclass()
# ============================================

class Animal:
    pass

class Dog(Animal):
    pass

class GoldenRetriever(Dog):
    pass

buddy = GoldenRetriever()

# isinstance() — checks inheritance chain (PREFERRED)
print("isinstance checks:")
print(f"  isinstance(buddy, GoldenRetriever): {isinstance(buddy, GoldenRetriever)}")  # True
print(f"  isinstance(buddy, Dog):             {isinstance(buddy, Dog)}")              # True
print(f"  isinstance(buddy, Animal):          {isinstance(buddy, Animal)}")           # True

# type() — checks EXACT type only
print(f"\ntype checks:")
print(f"  type(buddy) == GoldenRetriever: {type(buddy) == GoldenRetriever}")  # True
print(f"  type(buddy) == Dog:             {type(buddy) == Dog}")              # False!
print(f"  type(buddy) == Animal:          {type(buddy) == Animal}")           # False!

# issubclass() — checks class hierarchy
print(f"\nissubclass checks:")
print(f"  issubclass(GoldenRetriever, Dog):    {issubclass(GoldenRetriever, Dog)}")     # True
print(f"  issubclass(GoldenRetriever, Animal): {issubclass(GoldenRetriever, Animal)}")  # True
print(f"  issubclass(Dog, GoldenRetriever):    {issubclass(Dog, GoldenRetriever)}")     # False!
````

````python
# ============================================
# 6. Composition vs Inheritance
# ============================================
# Rule: "Favor composition over inheritance"
# Inheritance = IS-A (Dog IS AN Animal)
# Composition = HAS-A (Car HAS AN Engine)

# ❌ BAD — Inheritance where composition fits better
class Engine:
    def start(self):
        return "Engine started"

# class Car(Engine):  # Car IS-AN Engine? NO! Car HAS-AN Engine
#     pass

# ✅ GOOD — Composition
class Engine:
    def __init__(self, horsepower: int):
        self.horsepower = horsepower
    
    def start(self):
        return f"{self.horsepower}HP engine started 🔥"

class GPS:
    def navigate(self, destination: str):
        return f"Navigating to {destination} 🗺️"

class Car:
    def __init__(self, model: str, engine: Engine, gps: GPS = None):
        self.model = model
        self.engine = engine   # HAS-AN engine
        self.gps = gps         # HAS-A GPS (optional)
    
    def drive(self, destination: str = None):
        result = [f"{self.model}: {self.engine.start()}"]
        if destination and self.gps:
            result.append(self.gps.navigate(destination))
        return "\n".join(result)


car = Car("Tesla Model 3", Engine(283), GPS())
print(car.drive("Office"))

print("\n--- When to use each ---")
print("Inheritance: Dog IS-AN Animal (true taxonomy)")
print("Composition: Car HAS-AN Engine (pluggable parts)")
print("Tip: If in doubt, use composition. It's more flexible.")
````

```
│  INHERITANCE = Code Reuse via Parent-Child   │
│                                             │
│  Parent class  ──inherits──►  Child class   │
│  (Base/Super)                (Derived/Sub)  │
│                                             │

│  Child CAN override or extend them          │
└─────────────────────────────────────────────┘
```

**Quick mental model:** Inheritance = "My child IS-A type of me"

## 🧠 1. Inheritance Basics — Parent → Child → Grandchild

```
    ┌──────────┐
    │  Animal   │  ← Base class (has speak, eat)
    └────┬─────┘
         │ inherits
    ┌────▼─────┐
    │   Dog     │  ← Overrides speak(), adds fetch()
    └────┬─────┘
         │ inherits
  ┌──────▼────────┐
  │GoldenRetriever │  ← Overrides fetch(), adds be_cute()
  └───────────────┘
```

👉 **Key ideas:**
- `class Child(Parent):` — that's the syntax
- `super()` — calls the parent's method
- **Method overriding** — child redefines a parent method

````python
class Animal:
    def __init__(self, name, sound):
        self.name = name
        self.sound = sound
    
    def speak(self):
        return f"{self.name} says {self.sound}!"
    
    def eat(self):
        return f"{self.name} is eating."


class Dog(Animal):
    def __init__(self, name, breed):
        super().__init__(name, sound="Woof")  # 👈 super() calls Animal.__init__
        self.breed = breed
    
    def speak(self):  # 👈 Method override
        return f"{self.name} the {self.breed} says WOOF WOOF!"
    
    def fetch(self, item):
        return f"{self.name} fetches the {item}!"


class GoldenRetriever(Dog):
    def __init__(self, name):
        super().__init__(name, breed="Golden Retriever")  # 👈 calls Dog.__init__
    
    def fetch(self, item):  # 👈 Override Dog's fetch
        return f"{self.name} enthusiastically fetches the {item} and wants MORE!"
    
    def be_cute(self):
        return f"{self.name} tilts head. Maximum cuteness achieved."


# Test the chain
buddy = GoldenRetriever("Buddy")
print(buddy.speak())     # From Dog (overridden)
print(buddy.eat())       # From Animal (inherited all the way up)
print(buddy.fetch("ball"))  # From GoldenRetriever (overridden)
print(buddy.be_cute())   # GoldenRetriever's own
````

## 🎯 2. Types of Inheritance

### 2a. Single Inheritance
```
  [Parent] ──► [Child]
```
One parent, one child. Simplest form. (Dog → Animal above)

### 2b. Multilevel Inheritance
```
  [Grandparent] ──► [Parent] ──► [Child]
```
Chain of inheritance. (Animal → Dog → GoldenRetriever above)

### 2c. Hierarchical Inheritance
```
         [Animal]
        /    |    \
     [Dog] [Cat] [Bird]
```
One parent, MANY children.

### 2d. Multiple Inheritance
```
  [Father]   [Mother]
      \       /
       [Child]
```
One child, MANY parents. Python supports this! (Java/C# don't)

### 2e. Hybrid Inheritance
```
       [A]
      / \
    [B]  [C]
      \ /
      [D]
```
Mix of multiple + multilevel. Creates the **Diamond Problem** ⚠️

````python
# Quick examples of each type

# --- Hierarchical ---
class Animal:
    def breathe(self): return "Breathing..."

class Dog(Animal): pass
class Cat(Animal): pass
class Bird(Animal):
    def fly(self): return "Flying!"

print(Dog().breathe())   # Inherited
print(Bird().fly())      # Own method
print(Bird().breathe())  # Inherited

# --- Multiple Inheritance ---
class Father:
    def skill(self): return "Coding"

class Mother:
    def talent(self): return "Painting"

class Child(Father, Mother):  # 👈 Both parents!
    pass

kid = Child()
print(kid.skill())   # From Father
print(kid.talent())  # From Mother
````

## 🧠 3. `super()` Deep Dive

```
┌─────────────────────────────────────────────────┐
│  super() = "call the NEXT class in the MRO"     │
│                                                 │
│  NOT just "call my parent" — that's a           │
│  simplification that breaks with multiple       │
│  inheritance!                                   │
└─────────────────────────────────────────────────┘
```

👉 **Key insight:** `super()` follows the **MRO** (Method Resolution Order), not just the parent.

````python
# super() in single inheritance — straightforward
class Base:
    def __init__(self):
        print("Base.__init__")
        self.base_attr = "I'm from Base"

class Child(Base):
    def __init__(self):
        super().__init__()       # 👈 Calls Base.__init__
        print("Child.__init__")
        self.child_attr = "I'm from Child"

class GrandChild(Child):
    def __init__(self):
        super().__init__()       # 👈 Calls Child.__init__ (which calls Base)
        print("GrandChild.__init__")

print("--- Creating GrandChild ---")
gc = GrandChild()
print(f"Has base_attr: {gc.base_attr}")   # Inherited through chain
print(f"Has child_attr: {gc.child_attr}")  # Inherited from Child
````

````python
# super() in MULTIPLE inheritance — here's where it gets interesting
class A:
    def greet(self):
        print("Hello from A")

class B(A):
    def greet(self):
        print("Hello from B")
        super().greet()  # 👈 Does NOT always go to A!

class C(A):
    def greet(self):
        print("Hello from C")
        super().greet()

class D(B, C):  # Multiple inheritance
    def greet(self):
        print("Hello from D")
        super().greet()

print("--- MRO of D ---")
print([cls.__name__ for cls in D.__mro__])

print("\n--- D().greet() trace ---")
D().greet()
# D → B → C → A  (NOT D → B → A → C → A)
# super() follows MRO, so B's super() goes to C, not A!
````

## 🎯 4. Multiple Inheritance & MRO (C3 Linearization)

```
┌───────────────────────────────────────────────────┐
│  MRO = Method Resolution Order                    │
│                                                   │
│  "In what ORDER does Python search for a method?" │
│                                                   │
│  Algorithm: C3 Linearization                      │
│  Rule: Child before Parent, Left before Right     │
│        Preserve the order of parents              │
└───────────────────────────────────────────────────┘
```

### C3 Linearization — Simple Mental Model:
1. Start with the class itself
2. Go **left-to-right** through parents
3. A class appears only ONCE (first valid position)
4. A parent can't appear before its children
5. End with `object`

````python
# MRO with a complex hierarchy
#
#        object
#        /    \
#       A      B
#      / \    /
#     C   D  E
#      \ | /
#        F

class A: pass
class B: pass
class C(A): pass
class D(A): pass
class E(B): pass
class F(C, D, E): pass

print("MRO of F:")
print(" → ".join(cls.__name__ for cls in F.__mro__))
# F → C → D → A → E → B → object

# Two ways to check MRO:
print("\n.mro() method:", [c.__name__ for c in F.mro()])
print(".__mro__ attr: ", [c.__name__ for c in F.__mro__])
````

````python
# ⚠️ Invalid MRO — Python will REFUSE to create the class
# This happens when C3 can't find a consistent order

class X: pass
class Y: pass
class A(X, Y): pass
class B(Y, X): pass  # Reversed order!

try:
    # A says X before Y, B says Y before X — contradiction!
    class C(A, B): pass
except TypeError as e:
    print(f"TypeError: {e}")
    print("Python can't resolve conflicting parent orders!")
````

## 💎 5. The Diamond Problem

```
         ┌─────┐
         │  A   │    ← Base class
         └──┬──┘
           / \
    ┌─────┐   ┌─────┐
    │  B  │   │  C  │  ← Both inherit from A
    └──┬──┘   └──┬──┘
        \     /
       ┌──▼──┐
       │  D  │    ← Inherits from BOTH B and C
       └─────┘
```

**The problem:** When D calls a method, should it go through B's version or C's version? And should A's `__init__` be called TWICE?

**Python's answer:** MRO + `super()` ensures A is called exactly ONCE.

````python
# Diamond Problem — WITHOUT super() (the WRONG way)
class A:
    def __init__(self):
        print("A.__init__")

class B(A):
    def __init__(self):
        A.__init__(self)  # 👈 Directly calling parent
        print("B.__init__")

class C(A):
    def __init__(self):
        A.__init__(self)  # 👈 Directly calling parent
        print("C.__init__")

class D(B, C):
    def __init__(self):
        B.__init__(self)
        C.__init__(self)
        print("D.__init__")

print("--- WITHOUT super() --- (A called TWICE! Bad!)")
d = D()
````

````python
# Diamond Problem — WITH super() (the RIGHT way)
class A:
    def __init__(self):
        print("A.__init__")
        super().__init__()  # 👈 Even A calls super() (goes to object)

class B(A):
    def __init__(self):
        print("B.__init__")
        super().__init__()  # 👈 Follows MRO → goes to C, not A!

class C(A):
    def __init__(self):
        print("C.__init__")
        super().__init__()  # 👈 Follows MRO → goes to A

class D(B, C):
    def __init__(self):
        print("D.__init__")
        super().__init__()  # 👈 Follows MRO → goes to B

print("--- WITH super() --- (A called only ONCE! Correct!)")
d = D()

print("\nMRO:", " → ".join(cls.__name__ for cls in D.__mro__))

# Trace: D → B → C → A → object
# Each class called exactly once! 🎉
````

## 🧩 6. Mixins — Plug-and-Play Behavior

```
┌───────────────────────────────────────────────┐
│  Mixin = Small class that adds ONE behavior   │
│                                               │
│  Rules:                                       │
│  • NOT meant to be instantiated alone         │
│  • Provides a single, focused feature         │
│  • Name ends with 'Mixin' (convention)        │
│  • Uses multiple inheritance to "mix in"      │
└───────────────────────────────────────────────┘

  [LoggingMixin]  [SerializationMixin]  [Base]
        \              |                /
         \             |               /
          ────────── [MyClass] ───────
```

````python
import json
from datetime import datetime


# --- Mixin 1: Logging ---
class LoggingMixin:
    """Adds logging capability to any class."""
    def log(self, message):
        timestamp = datetime.now().strftime("%H:%M:%S")
        print(f"[{timestamp}] {self.__class__.__name__}: {message}")


# --- Mixin 2: Serialization ---
class SerializationMixin:
    """Adds JSON serialization to any class."""
    def to_json(self):
        return json.dumps(self.__dict__, indent=2, default=str)
    
    @classmethod
    def from_json(cls, json_str):
        data = json.loads(json_str)
        obj = cls.__new__(cls)
        obj.__dict__.update(data)
        return obj


# --- Mixin 3: Validation ---
class ValidationMixin:
    """Adds field validation."""
    def validate(self):
        for attr, value in self.__dict__.items():
            if value is None:
                raise ValueError(f"{attr} cannot be None")
        return True


# --- Main class uses mixins ---
class User(LoggingMixin, SerializationMixin, ValidationMixin):
    def __init__(self, name, email):
        self.name = name
        self.email = email
        self.log(f"Created user: {name}")  # 👈 From LoggingMixin


user = User("Sarthak", "sarthak@example.com")
print(user.to_json())         # 👈 From SerializationMixin
print(user.validate())        # 👈 From ValidationMixin

print("\nMRO:", " → ".join(c.__name__ for c in User.__mro__))
````

## 👉 7. `isinstance()` vs `issubclass()`

```
isinstance(obj, Class)    → Is this OBJECT an instance of this class (or its parents)?
issubclass(ChildCls, ParentCls) → Is this CLASS a subclass of another class?

         isinstance  → checks OBJECTS
         issubclass  → checks CLASSES
```

````python
class Animal: pass
class Dog(Animal): pass
class Cat(Animal): pass
class GoldenRetriever(Dog): pass

buddy = GoldenRetriever()

# isinstance — checks the OBJECT
print("=== isinstance ===")
print(f"buddy is GoldenRetriever: {isinstance(buddy, GoldenRetriever)}")  # True
print(f"buddy is Dog:             {isinstance(buddy, Dog)}")              # True (parent)
print(f"buddy is Animal:          {isinstance(buddy, Animal)}")           # True (grandparent)
print(f"buddy is Cat:             {isinstance(buddy, Cat)}")              # False

# issubclass — checks the CLASS
print("\n=== issubclass ===")
print(f"GoldenRetriever < Dog:    {issubclass(GoldenRetriever, Dog)}")     # True
print(f"GoldenRetriever < Animal: {issubclass(GoldenRetriever, Animal)}")  # True
print(f"Dog < Cat:                {issubclass(Dog, Cat)}")                 # False
print(f"Dog < Dog:                {issubclass(Dog, Dog)}")                 # True (itself!)

# 👉 Pro tip: isinstance() with tuple checks multiple types
print(f"\nbuddy is Dog or Cat: {isinstance(buddy, (Dog, Cat))}")  # True
````

## 🔄 8. Composition vs Inheritance

```
┌─────────────────────┐    ┌─────────────────────┐
│    INHERITANCE       │    │    COMPOSITION       │
│    "IS-A"            │    │    "HAS-A"           │
│                     │    │                     │
│  Dog IS-A Animal    │    │  Car HAS-A Engine   │
│  class Dog(Animal)  │    │  self.engine = ..   │
│                     │    │                     │
│  Tight coupling     │    │  Loose coupling     │
│  Hard to change     │    │  Easy to swap       │
│  Can break children │    │  Flexible            │
└─────────────────────┘    └─────────────────────┘

  🏆 "Favor composition over inheritance"
     — Gang of Four (Design Patterns book)
```

````python
# ❌ BAD: Inheritance where composition is better
# Why? Because a Car is NOT an Engine. It HAS an Engine.

class Engine:
    def start(self): return "Engine started"
    def stop(self): return "Engine stopped"

# Bad: Car inherits Engine
class BadCar(Engine):
    def drive(self): return "Driving..."

# What if we want to swap engines? Can't! Stuck with parent's implementation.

print("--- BAD (Inheritance) ---")
bad_car = BadCar()
print(bad_car.start())  # Works but semantically wrong
````

````python
# ✅ GOOD: Composition — Car HAS-A Engine

class ElectricEngine:
    def start(self): return "⚡ Electric engine humming"
    def stop(self): return "⚡ Electric engine off"

class GasEngine:
    def start(self): return "🔥 Gas engine roaring"
    def stop(self): return "🔥 Gas engine off"

class Car:
    def __init__(self, engine):  # 👈 INJECT the engine (composition)
        self.engine = engine      # Car HAS-A engine
    
    def start(self):
        return self.engine.start()
    
    def drive(self):
        return f"{self.engine.start()} → Driving!"

print("--- GOOD (Composition) ---")
tesla = Car(ElectricEngine())     # Easy to swap!
mustang = Car(GasEngine())

print(f"Tesla:   {tesla.drive()}")
print(f"Mustang: {mustang.drive()}")

# 👉 Want to swap Mustang to electric? Just change the engine!
mustang.engine = ElectricEngine()
print(f"Mustang (converted): {mustang.drive()}")
````

### When to use which?

| Use **Inheritance** when... | Use **Composition** when... |
|---|---|
| True IS-A relationship | HAS-A relationship |
| `Dog IS-A Animal` ✅ | `Car HAS-A Engine` ✅ |
| Shared behavior across family | Need flexibility to swap parts |
| Polymorphism needed | Runtime behavior changes needed |
| Stable, unlikely to change | Likely to change/extend |

## 🗺️ 9. Inheritance Pattern Map

```
┌──────────────────── INHERITANCE CHEATSHEET ────────────────────┐
│                                                                │
│  SYNTAX:  class Child(Parent):                                 │
│           class Child(Parent1, Parent2):  # multiple           │
│                                                                │
│  super():  Calls next class in MRO (not just parent!)          │
│  MRO:      C3 Linearization → left-to-right, child-first      │
│                                                                │
│  ┌─────────────────┐  ┌─────────────────┐                     │
│  │ isinstance(obj) │  │ issubclass(cls) │                     │
│  │ checks OBJECTS   │  │ checks CLASSES  │                     │
│  └─────────────────┘  └─────────────────┘                     │
│                                                                │
│  TYPES:                                                        │
│  ┌──────────┬───────────┬──────────────┬─────────┐            │
│  │ Single   │ Multiple  │ Multilevel   │ Hybrid  │            │
│  │ A→B      │ A,B→C     │ A→B→C        │ Diamond │            │
│  └──────────┴───────────┴──────────────┴─────────┘            │
│                                                                │
│  DIAMOND FIX: Use super() everywhere → MRO handles it         │
│  MIXINS:      Small, focused, plug-in behavior classes         │
│  BEST PRACTICE: Favor composition (HAS-A) over inheritance    │
│                                                                │
│  ⚠️  GOTCHAS:                                                  │
│  • Forgetting super().__init__() → missing parent attrs        │
│  • Direct Parent.__init__() → breaks diamond pattern           │
│  • Deep hierarchies → hard to debug                            │
│  • Multiple inheritance → check MRO before shipping            │
└────────────────────────────────────────────────────────────────┘
```

## 💡 10. Interview Quick-Fire — Inheritance

**Q1: What is the difference between `super().__init__()` and `Parent.__init__(self)`?**
> `super()` follows MRO — works correctly with multiple inheritance. Direct call bypasses MRO and can cause double-init in diamonds.

**Q2: What is MRO? How does Python compute it?**
> Method Resolution Order — the order Python searches for methods. Computed via **C3 Linearization**: child before parent, left before right, consistent ordering.

**Q3: What is the Diamond Problem? How does Python solve it?**
> When D inherits from B and C, both inheriting from A. Without MRO, A's init runs twice. Python solves it with C3 MRO + `super()` — each class called exactly once.

**Q4: What's the difference between `isinstance()` and `issubclass()`?**
> `isinstance(obj, cls)` checks if an **object** is an instance. `issubclass(cls1, cls2)` checks if a **class** is a subclass. Both check the full inheritance chain.

**Q5: When would you use a Mixin?**
> When you want to add a single, reusable behavior (logging, serialization) to multiple unrelated classes without creating a deep hierarchy. Mixins are meant to be combined, not stand alone.

**Q6: Why "favor composition over inheritance"?**
> Composition is more flexible (swap parts at runtime), loose coupling (changes don't break children), and avoids fragile base class problem. Use inheritance only for true IS-A relationships.

**Q7: Can Python's MRO ever fail?**
> Yes! If parent order is contradictory (class A(X,Y) and class B(Y,X), then class C(A,B)), Python raises `TypeError` — inconsistent MRO.

**Q8: What happens if you forget `super().__init__()` in a child class?**
> Parent's `__init__` never runs → parent's attributes won't exist → `AttributeError` when accessing them. Always call `super().__init__()` unless you intentionally want to skip parent init.
