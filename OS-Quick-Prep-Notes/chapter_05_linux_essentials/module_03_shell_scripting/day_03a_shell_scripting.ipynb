╠══════════════════════════════════════════════════════════════╣
║                                                              ║
║  VARIABLES                                                   ║
║    name="value"      — NO spaces around =                   ║

║  CONDITIONS                                                  ║
║    [ -f file ]       — file exists?                         ║

║    [ -z "$var" ]     — string empty?                       ║
║    [ "$a" = "$b" ]   — string equal?                       ║
║    [ $a -eq $b ]     — numeric equal?                      ║
║                                                              ║
║  TEXT TRIO                                                   ║
║    grep  — search/filter lines                               ║

````python
# ============================================
# 1. Bash Script Patterns (as Python reference)
# ============================================

# Since Jupyter runs Python, we'll show bash scripts as strings
# and explain each pattern

bash_patterns = {
    "Basic Script Structure": '''
#!/bin/bash
set -euo pipefail    # Exit on error, undefined vars, pipe fails

# Variables
NAME="deploy"
TIMESTAMP=$(date +%Y%m%d_%H%M%S)
LOG_FILE="/var/log/${NAME}_${TIMESTAMP}.log"

echo "Starting $NAME at $TIMESTAMP"
''',
    "If/Else & Functions": '''
#!/bin/bash

check_service() {
    local service=$1
    if systemctl is-active --quiet "$service"; then
        echo "✅ $service is running"
        return 0
    else
        echo "❌ $service is down!"
        return 1
    fi
}

for svc in nginx postgresql redis; do
    check_service "$svc"
done
''',
    "Loops & Arrays": '''
#!/bin/bash

# Array
SERVERS=("web01" "web02" "db01")

# For loop
for server in "${SERVERS[@]}"; do
    echo "Pinging $server..."
    ping -c 1 "$server" > /dev/null 2>&1 && echo "UP" || echo "DOWN"
done

# While loop (read file line by line)
while IFS= read -r line; do
    echo "Processing: $line"
done < input.txt

# Counter
count=0
while [ $count -lt 5 ]; do
    echo "Attempt $((count + 1))"
    count=$((count + 1))
done
''',
}

for title, script in bash_patterns.items():
    print(f"  {'═' * 55}")
    print(f"  📝 {title}")
    print(f"  {'═' * 55}")
    for line in script.strip().split('\n'):
        print(f"  {line}")
    print()
````

````python
# ============================================
# 2. grep / sed / awk — The Text Processing Trio
# ============================================

grep_examples = [
    ('grep "error" app.log',          'Find lines containing "error"'),
    ('grep -i "error" app.log',       'Case-insensitive search'),
    ('grep -r "TODO" src/',           'Recursive search in directory'),
    ('grep -n "def " *.py',           'Show line numbers'),
    ('grep -c "ERROR" app.log',       'Count matching lines'),
    ('grep -v "DEBUG" app.log',       'Invert (exclude DEBUG lines)'),
    ('grep -E "err|warn|fatal" log',  'Extended regex (OR)'),
    ('grep -l "password" *.conf',     'List files with matches only'),
]

sed_examples = [
    ('sed "s/old/new/" file',         'Replace first occurrence per line'),
    ('sed "s/old/new/g" file',        'Replace ALL occurrences'),
    ('sed -i "s/old/new/g" file',     'Edit file in-place'),
    ('sed -n "5,10p" file',           'Print lines 5-10'),
    ('sed "/^#/d" file',              'Delete comment lines'),
    ('sed "3d" file',                 'Delete line 3'),
]

awk_examples = [
    ('awk \'{print $1}\' file',       'Print first column'),
    ('awk -F: \'{print $1}\' /etc/passwd', 'Custom delimiter (:)'),
    ('awk \'{sum+=$2} END{print sum}\'', 'Sum column 2'),
    ('awk \'$3 > 100\' file',         'Filter: column 3 > 100'),
    ('awk \'NR==5\' file',            'Print line 5'),
    ('awk \'{print NR, $0}\' file',   'Add line numbers'),
]

for name, examples in [("grep", grep_examples), ("sed", sed_examples), ("awk", awk_examples)]:
    print(f"  {'═' * 60}")
    print(f"  🔍 {name.upper()} — {'search' if name=='grep' else 'edit' if name=='sed' else 'process'}")
    print(f"  {'═' * 60}")
    for cmd, desc in examples:
        print(f"    {cmd:<45} # {desc}")
    print()
````

````python
# ============================================
# 3. Must-Know One-Liners for Interviews
# ============================================

one_liners = [
    ("Find large files (>100MB)",
     'find / -type f -size +100M 2>/dev/null'),
    
    ("Disk usage by directory",
     'du -sh /var/* | sort -rh | head -10'),
    
    ("Count lines of Python code",
     'find . -name "*.py" | xargs wc -l | tail -1'),
    
    ("Find process using port 8080",
     'lsof -i :8080'),
    
    ("Watch log file in real-time",
     'tail -f /var/log/app.log | grep --color ERROR'),
    
    ("Find files modified in last 24h",
     'find . -mtime -1 -type f'),
    
    ("Replace string in all files recursively",
     'find . -type f -name "*.py" -exec sed -i "s/old/new/g" {} +'),
    
    ("Top 10 memory-consuming processes",
     'ps aux --sort=-%mem | head -11'),
    
    ("Extract unique IPs from access log",
     'awk \'{print $1}\' access.log | sort -u | wc -l'),
    
    ("Quick HTTP server",
     'python3 -m http.server 8000'),
    
    ("SSH tunnel (port forward)",
     'ssh -L 3306:localhost:3306 user@remote'),
    
    ("Generate random password",
     'openssl rand -base64 24'),
]

print("  🚀 Must-Know Linux One-Liners")
print("  " + "═" * 65)
for i, (desc, cmd) in enumerate(one_liners, 1):
    print(f"\n  {i:2d}. {desc}")
    print(f"      $ {cmd}")
````

## 📝 Interview Questions

| # | Question | Key Point |
|---|----------|------------|
| 1 | Difference between ` ` and $( )? | Both = command substitution. $() is preferred (nestable, readable) |
| 2 | What does set -euo pipefail do? | -e: exit on error. -u: error on undefined vars. -o pipefail: pipe errors |
| 3 | grep vs awk vs sed? | grep: search. sed: stream edit. awk: column processing & computation |
| 4 | How to debug a bash script? | `bash -x script.sh` (trace mode). Or add `set -x` in script |
| 5 | $@ vs $*? | $@: each arg as separate word (use this!). $*: all args as single string |
