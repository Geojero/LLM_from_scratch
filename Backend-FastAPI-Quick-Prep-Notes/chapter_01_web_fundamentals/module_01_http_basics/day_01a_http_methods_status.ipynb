## 🎭 The Analogy

Think of HTTP like ordering at a restaurant:

- **GET** → "Can I see the menu?" (just reading)
- **POST** → "I'd like to place an order" (creating something new)
- **PUT** → "Actually, change my entire order to this" (full replacement)
- **PATCH** → "Just add extra cheese to my pizza" (partial update)
- **DELETE** → "Cancel my order" (remove it)

The waiter (server) responds with a **status code** — like a thumbs up (200), "we don't have that" (404), or "kitchen is on fire" (500).

### 🤔 What's **Idempotent**?

**Making the same request N times produces the same result as making it once.**

- `GET /users/1` → Always returns user 1. ✅ Idempotent
- `DELETE /users/1` → First call deletes, second call gets 404, but the **state** is the same. ✅ Idempotent
- `POST /users` → Each call creates a NEW user. ❌ NOT idempotent

> **Interview Tip:** "PUT is idempotent because sending the same PUT twice results in the same resource state. POST is NOT because it creates a new resource each time."

### PUT vs PATCH — The Key Difference

```python
# PUT: Replace the ENTIRE resource
# You must send ALL fields
PUT /users/1
{"name": "Sarthak", "email": "s@mail.com", "age": 25}

# PATCH: Update only SPECIFIC fields
# Send only what changed
PATCH /users/1
{"age": 26}
```

## 📊 Status Codes — The Interview Favorites

### 2xx — Success ✅
```
╔══════╦══════════════════════════════════════════════════════════╗
║ Code ║ Meaning                                                  ║
╠══════╬══════════════════════════════════════════════════════════╣
║ 200  ║ OK — Request succeeded (GET, PUT, PATCH)                ║
║ 201  ║ Created — New resource created (POST)                   ║
║ 204  ║ No Content — Success but no body (DELETE)               ║
╚══════╩══════════════════════════════════════════════════════════╝
```

### 3xx — Redirection 🔄
```
╔══════╦══════════════════════════════════════════════════════════╗
║ 301  ║ Moved Permanently — URL changed forever (SEO!)         ║
║ 302  ║ Found — Temporarily at different URL                    ║
║ 304  ║ Not Modified — Use your cached version                  ║
╚══════╩══════════════════════════════════════════════════════════╝
```

> **Interview Classic:** "301 vs 302?" → 301 = permanent (search engines update). 302 = temporary (like maintenance page).

### 4xx — Client Error ❌
```
╔══════╦══════════════════════════════════════════════════════════╗
║ 400  ║ Bad Request — Malformed syntax, invalid data            ║
║ 401  ║ Unauthorized — "Who are you?" (not logged in)          ║
║ 403  ║ Forbidden — "I know who you are, but NO" (no perms)   ║
║ 404  ║ Not Found — Resource doesn't exist                      ║
║ 405  ║ Method Not Allowed — Can't POST to this URL             ║
║ 409  ║ Conflict — Resource state conflict (duplicate email)    ║
║ 422  ║ Unprocessable Entity — Valid JSON but bad values        ║
║ 429  ║ Too Many Requests — Rate limited                        ║
╚══════╩══════════════════════════════════════════════════════════╝
```

> **Interview Classic:** "401 vs 403?" → 401 = identity unknown (not authenticated). 403 = identity known but permission denied (not authorized).

### 5xx — Server Error 🔥
```
╔══════╦══════════════════════════════════════════════════════════╗
║ 500  ║ Internal Server Error — Bug in our code                 ║
║ 502  ║ Bad Gateway — Upstream server failed                    ║
║ 503  ║ Service Unavailable — Server overloaded/maintenance     ║
║ 504  ║ Gateway Timeout — Upstream server too slow              ║
╚══════╩══════════════════════════════════════════════════════════╝
```

## 💻 Code — Making HTTP Requests from Memory

You need to be able to write these **without looking anything up**. Let's drill.

````python
import requests

# ============================================
# GET — Retrieve data
# ============================================
response = requests.get("https://jsonplaceholder.typicode.com/posts/1")
print(f"Status: {response.status_code}")  # 200
print(f"JSON: {response.json()}")          # dict
print(f"Headers: {response.headers['Content-Type']}")  # application/json
````

````python
# ============================================
# GET with query parameters
# ============================================
# This hits: /posts?userId=1&_limit=3
params = {"userId": 1, "_limit": 3}
response = requests.get(
    "https://jsonplaceholder.typicode.com/posts",
    params=params
)
print(f"URL built: {response.url}")  # see the query string
print(f"Got {len(response.json())} posts")
````

````python
# ============================================
# POST — Create a new resource
# ============================================
new_post = {
    "title": "My Post",
    "body": "Hello from interview prep!",
    "userId": 1
}
response = requests.post(
    "https://jsonplaceholder.typicode.com/posts",
    json=new_post  # automatically sets Content-Type: application/json
)
print(f"Status: {response.status_code}")  # 201 Created
print(f"Created: {response.json()}")
````

````python
# ============================================
# PUT — Full replacement
# ============================================
updated_post = {
    "id": 1,
    "title": "Updated Title",
    "body": "Completely new body",
    "userId": 1
}
response = requests.put(
    "https://jsonplaceholder.typicode.com/posts/1",
    json=updated_post
)
print(f"Status: {response.status_code}")  # 200
print(f"Updated: {response.json()}")
````

````python
# ============================================
# PATCH — Partial update
# ============================================
partial_update = {"title": "Only Title Changed"}
response = requests.patch(
    "https://jsonplaceholder.typicode.com/posts/1",
    json=partial_update
)
print(f"Status: {response.status_code}")  # 200
print(f"Patched: {response.json()}")
````

````python
# ============================================
# DELETE — Remove resource
# ============================================
response = requests.delete("https://jsonplaceholder.typicode.com/posts/1")
print(f"Status: {response.status_code}")  # 200
print(f"Body: {response.json()}")  # empty dict {}
````

````python
# ============================================
# Custom headers + Auth
# ============================================
headers = {
    "Authorization": "Bearer my-secret-token-123",
    "Accept": "application/json",
    "X-Request-ID": "abc-123"
}
response = requests.get(
    "https://jsonplaceholder.typicode.com/posts/1",
    headers=headers
)
print(f"Request headers sent: {response.request.headers}")
````

````python
# ============================================
# Error handling — production pattern
# ============================================
import requests
from requests.exceptions import HTTPError, ConnectionError, Timeout

def safe_api_call(url, method="GET", **kwargs):
    """Pattern you should know for interviews."""
    try:
        response = requests.request(method, url, timeout=10, **kwargs)
        response.raise_for_status()  # Raises HTTPError for 4xx/5xx
        return response.json()
    except HTTPError as e:
        print(f"HTTP Error: {e.response.status_code} - {e.response.text}")
    except ConnectionError:
        print("Connection failed — is the server running?")
    except Timeout:
        print("Request timed out")
    except Exception as e:
        print(f"Unexpected error: {e}")
    return None

# Test it
result = safe_api_call("https://jsonplaceholder.typicode.com/posts/1")
print(f"Result: {result['title'] if result else 'Failed'}")
````

````python
# ============================================
# Session — persistent connection + cookies
# ============================================
# In interviews: "I'd use a Session for connection pooling 
# and persistent headers across requests"

session = requests.Session()
session.headers.update({"Authorization": "Bearer token123"})

# All requests now include the auth header
r1 = session.get("https://jsonplaceholder.typicode.com/posts/1")
r2 = session.get("https://jsonplaceholder.typicode.com/posts/2")

print(f"Post 1: {r1.json()['title']}")
print(f"Post 2: {r2.json()['title']}")

session.close()  # always close when done
````

## 📝 Interview Questions — Practice These

| # | Question | Key Point |
|---|----------|-----------|
| 1 | GET vs POST? | GET reads (cacheable, idempotent), POST creates (has body, not idempotent) |
| 2 | PUT vs PATCH? | PUT replaces entire resource, PATCH updates specific fields |
| 3 | 401 vs 403? | 401 = not authenticated (who are you?), 403 = not authorized (no permission) |
| 4 | 301 vs 302? | 301 = permanent redirect (SEO transfers), 302 = temporary |
| 5 | What is idempotency? | Same request N times = same result as once. GET, PUT, DELETE are idempotent |
| 6 | POST vs PUT for creation? | POST creates at server-chosen URL (/users → /users/123). PUT creates at client-chosen URL (/users/123) |
| 7 | What's 204? | No Content — success but nothing to return (DELETE) |
| 8 | What's 422? | Unprocessable Entity — valid JSON but semantically wrong (age: -5) |
| 9 | What's 429? | Too Many Requests — rate limited |
| 10 | HTTP vs HTTPS? | HTTPS = HTTP + TLS encryption. Data encrypted in transit |
