## 🎯 Problem 1: Decode String (LC #394)

Given `"3[a2[c]]"`, return `"accaccacc"`.

**Stack-based approach:** When you see `[`, push current string and count onto stack. When you see `]`, pop and repeat.

Think of it like function calls: `3[a2[c]]` → the inner `2[c]` resolves to `cc`, then `a + cc = acc`, then repeat 3x.

```
  Processing: 3 [ a 2 [ c ] ]
  
  See '3': num = 3
  See '[': push ("", 3) onto stack, reset
  See 'a': current = "a"
  See '2': num = 2
  See '[': push ("a", 2) onto stack, reset
  See 'c': current = "c"
  See ']': pop ("a", 2) → current = "a" + "c"*2 = "acc"
  See ']': pop ("", 3)  → current = "" + "acc"*3 = "accaccacc"
```

````python
def decodeString(s):
    """
    Stack: push (prev_string, count) on '[', pop on ']'.
    Time: O(output length), Space: O(nesting depth)
    """
    stack = []  # (previous_string, repeat_count)
    current = ""
    num = 0
    
    for ch in s:
        if ch.isdigit():
            num = num * 10 + int(ch)  # Handle multi-digit numbers
        elif ch == '[':
            stack.append((current, num))
            current = ""
            num = 0
        elif ch == ']':
            prev, count = stack.pop()
            current = prev + current * count
        else:
            current += ch
    
    return current

print(decodeString("3[a2[c]]"))      # "accaccacc"
print(decodeString("2[abc]3[cd]ef"))  # "abcabccdcdcdef"
````

## 🎯 Problem 2: Basic Calculator II (LC #227)

Evaluate `"3+2*2"` = 7. Supports `+`, `-`, `*`, `/` (no parentheses).

**Stack approach:** Process left to right. For `*` and `/`, immediately compute with the top of stack. For `+` and `-`, just push the number (with sign).

```
  "3+2*2"
  Stack after processing:
    see 3, op='+': push +3  → [3]
    see 2, op='+': push +2  → [3, 2]
    see 2, op='*': pop 2, push 2*2  → [3, 4]
  Result = sum(stack) = 7
```

````python
def calculate(s):
    """
    Stack: handle * / immediately, defer + -.
    Time: O(n), Space: O(n)
    """
    stack = []
    num = 0
    op = '+'
    
    for i, ch in enumerate(s):
        if ch.isdigit():
            num = num * 10 + int(ch)
        
        if ch in '+-*/' or i == len(s) - 1:
            if op == '+':
                stack.append(num)
            elif op == '-':
                stack.append(-num)
            elif op == '*':
                stack.append(stack.pop() * num)
            elif op == '/':
                stack.append(int(stack.pop() / num))  # Truncate toward zero
            op = ch
            num = 0
    
    return sum(stack)

print(calculate("3+2*2"))     # 7
print(calculate(" 3/2 "))     # 1
print(calculate(" 3+5 / 2 ")) # 5
````

## 🎯 Problem 3: Encode and Decode Strings (LC #271)

Design encode/decode for a list of strings. The trick is handling strings that contain any character (including delimiters).

**Length-prefix encoding:** `"hello"` → `"5#hello"`. The `#` separates length from content. No ambiguity possible.

````python
def encode(strs):
    """Encode: length + '#' + string."""
    return ''.join(f"{len(s)}#{s}" for s in strs)

def decode(s):
    """Decode: read length, skip '#', read that many chars."""
    result = []
    i = 0
    while i < len(s):
        j = s.index('#', i)
        length = int(s[i:j])
        result.append(s[j+1:j+1+length])
        i = j + 1 + length
    return result

original = ["hello", "world", "a#b", ""]
encoded = encode(original)
print(f"Encoded: {encoded}")
print(f"Decoded: {decode(encoded)}")  # ['hello', 'world', 'a#b', '']
````
