## Building Strings Efficiently

Many interview problems ask you to **construct** a result string character by character. Remember the golden rule: never concatenate in a loop. Use a **list** and `join` at the end.

This module covers problems where you're doing math-on-strings, converting between representations, or building output character by character.

## 🎯 Problem 1: Add Binary (LC #67)

Add two binary strings and return the sum as a binary string.

**Same pattern as adding two numbers on paper:** Process from right to left, track carry.

```
    1 0 1 1
  +   1 1 0
  ─────────
  1 0 0 0 1
  
  Position 0: 1+0 = 1, carry 0
  Position 1: 1+1 = 0, carry 1
  Position 2: 0+1+1(carry) = 0, carry 1
  Position 3: 1+0+1(carry) = 0, carry 1
  Position 4: 0+0+1(carry) = 1
```

````python
def addBinary(a, b):
    """
    Right-to-left addition with carry.
    Time: O(max(n,m)), Space: O(max(n,m))
    """
    result = []
    carry = 0
    i, j = len(a) - 1, len(b) - 1
    
    while i >= 0 or j >= 0 or carry:
        total = carry
        if i >= 0:
            total += int(a[i])
            i -= 1
        if j >= 0:
            total += int(b[j])
            j -= 1
        result.append(str(total % 2))
        carry = total // 2
    
    return ''.join(reversed(result))

print(addBinary("1011", "110"))  # "10001"
````

## 🎯 Problem 2: Multiply Strings (LC #43)

Multiply two numbers represented as strings (too large for int in some languages).

**Grade school multiplication:** Each digit `num1[i] * num2[j]` contributes to position `i + j` and `i + j + 1` in the result.

```
     1 2 3
   ×   4 5
   ───────
   positions: result has at most len(num1)+len(num2) digits
   num1[2]*num2[1] = 3*4 = 12 → goes to result[3] and result[4]
```

````python
def multiply(num1, num2):
    """
    Grade-school multiplication.
    Time: O(n*m), Space: O(n+m)
    """
    m, n = len(num1), len(num2)
    result = [0] * (m + n)
    
    for i in range(m - 1, -1, -1):
        for j in range(n - 1, -1, -1):
            mul = int(num1[i]) * int(num2[j])
            p1, p2 = i + j, i + j + 1
            total = mul + result[p2]
            
            result[p2] = total % 10
            result[p1] += total // 10
    
    # Remove leading zeros
    result_str = ''.join(map(str, result)).lstrip('0')
    return result_str or '0'

print(multiply("123", "45"))  # "5535"
print(multiply("0", "0"))    # "0"
````

## 🎯 Problem 3: Integer to Roman (LC #12)

Convert integer to Roman numeral string.

**Greedy from largest to smallest.** Use a lookup table that includes the subtractive cases (like 4=IV, 9=IX).

````python
def intToRoman(num):
    """
    Greedy with subtractive pairs. O(1) since max is 3999.
    """
    pairs = [
        (1000, 'M'), (900, 'CM'), (500, 'D'), (400, 'CD'),
        (100, 'C'),  (90, 'XC'),  (50, 'L'),  (40, 'XL'),
        (10, 'X'),   (9, 'IX'),   (5, 'V'),   (4, 'IV'),
        (1, 'I')
    ]
    
    result = []
    for value, symbol in pairs:
        while num >= value:
            result.append(symbol)
            num -= value
    
    return ''.join(result)

print(intToRoman(1994))  # "MCMXCIV"
print(intToRoman(58))    # "LVIII"
````
