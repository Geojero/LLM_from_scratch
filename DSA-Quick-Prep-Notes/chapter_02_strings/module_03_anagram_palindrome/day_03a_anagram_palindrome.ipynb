## Two of the Most Common Interview Topics

**Anagram:** Same characters, different order. `"listen"` → `"silent"`  
**Palindrome:** Reads the same forwards and backwards. `"racecar"`

These two categories cover probably 30% of all string interview questions. The good news? There's a small set of patterns that handles almost all of them.

```
╔═══════════════════════════════════════════════════════════════════════╗
║  ANAGRAM CORE TOOL: Character frequency count                      ║
║  → Counter(s1) == Counter(s2) → Anagram!                          ║
║                                                                     ║
║  PALINDROME CORE TOOL: Two pointers OR expand from center         ║
║  → Check: left pointer, right pointer, move inward                ║
║  → Find: expand from each center (2n-1 centers)                   ║
╚═══════════════════════════════════════════════════════════════════════╝
```

## 🎯 Problem 1: Valid Anagram (LC #242)

Are two strings anagrams of each other?

**If two strings have the exact same character frequencies, they're anagrams.** That's it. No sorting needed (though sorting works too).

````python
from collections import Counter

def isAnagram(s, t):
    """Counter comparison. O(n) time, O(1) space (26 chars max)."""
    return Counter(s) == Counter(t)

print(isAnagram("anagram", "nagaram"))  # True
print(isAnagram("rat", "car"))          # False
````

## 🎯 Problem 2: Group Anagrams (LC #49)

Group strings that are anagrams of each other.

**Key insight:** All anagrams produce the same sorted string (or same character count tuple). Use that as a **dictionary key**.

```
  "eat" → sorted → "aet"  ─┐
  "tea" → sorted → "aet"  ─┤→ Same group!
  "ate" → sorted → "aet"  ─┘
  "tan" → sorted → "ant"  ─┐→ Different group
  "nat" → sorted → "ant"  ─┘
```

````python
from collections import defaultdict

def groupAnagrams(strs):
    """
    Group by sorted string as key.
    Time: O(n * k log k) where k = max string length
    """
    groups = defaultdict(list)
    for s in strs:
        key = tuple(sorted(s))  # or use char count tuple for O(n*k)
        groups[key].append(s)
    return list(groups.values())

print(groupAnagrams(["eat","tea","tan","ate","nat","bat"]))
# [["eat","tea","ate"],["tan","nat"],["bat"]]
````

## 🎯 Problem 3: Longest Palindromic Substring (LC #5)

This is a **classic** interview problem. The trick? Instead of checking every substring (O(n³)), **expand from each center**.

A palindrome mirrors around its center. There are `2n - 1` possible centers (n single chars + n-1 between chars for even-length palindromes).

```
  "babad"
   ↑ expand from 'a' at index 1: "bab" ✓, "xbabx" ✗
     ↑ expand from 'b' at index 2: "aba" ✓, "babax" ✗
  Both give length 3, return either "bab" or "aba"
```

````python
def longestPalindrome(s):
    """
    Expand around center.
    Time: O(n²), Space: O(1)
    """
    def expand(left, right):
        while left >= 0 and right < len(s) and s[left] == s[right]:
            left -= 1
            right += 1
        return s[left+1:right]  # Last valid palindrome
    
    result = ""
    for i in range(len(s)):
        # Odd length palindrome (center = single char)
        odd = expand(i, i)
        # Even length palindrome (center = between two chars)
        even = expand(i, i + 1)
        
        result = max(result, odd, even, key=len)
    
    return result

print(longestPalindrome("babad"))   # "bab" or "aba"
print(longestPalindrome("cbbd"))    # "bb"
````

## 🎯 Problem 4: Palindrome Partitioning (LC #131)

Partition `s` so that every substring is a palindrome. Return all possible partitions.

**Backtracking:** At each position, try every possible palindrome starting there. If it's a palindrome, recurse on the rest.

````python
def partition(s):
    """
    Backtracking: try every palindrome prefix, recurse on suffix.
    """
    def is_palindrome(sub):
        return sub == sub[::-1]
    
    def backtrack(start, path):
        if start == len(s):
            result.append(path[:])
            return
        
        for end in range(start + 1, len(s) + 1):
            substring = s[start:end]
            if is_palindrome(substring):
                path.append(substring)
                backtrack(end, path)
                path.pop()
    
    result = []
    backtrack(0, [])
    return result

print(partition("aab"))  # [["a","a","b"],["aa","b"]]
````
