## Why Strings Get Their Own Chapter

You might think: "Strings are just arrays of characters, right?" Technically yes, but in Python they're **immutable**. That one difference changes everything about how you approach string problems.

When you do `s = s + 'x'`, Python doesn't append — it creates an **entirely new string**. Do that in a loop and you've got O(n²) without even realizing it. The fix? Work with **lists**, then `''.join()` at the end.

```
╔═══════════════════════════════════════════════════════════════════════╗
║  🔑 Python Strings Are Immutable                                   ║
║                                                                     ║
║  s[i] = 'x'      → TypeError!                                     ║
║  s += 'x'         → Creates NEW string each time → O(n) per op    ║
║  list(s) → modify → ''.join(lst)  → This is the pattern!          ║
╚═══════════════════════════════════════════════════════════════════════╝
```

### Essential Python String Operations (Quick Reference)

| Operation | Syntax | Time |
|-----------|--------|------|
| Access char | `s[i]` | O(1) |
| Slice | `s[i:j]` | O(j-i) |
| Concatenate | `s1 + s2` | O(len(s1)+len(s2)) |
| Find | `s.find(sub)` | O(n*m) |
| Split | `s.split()` | O(n) |
| Join | `''.join(lst)` | O(total chars) |
| Reverse | `s[::-1]` | O(n) |
| Count char | `s.count(c)` | O(n) |

## 🎯 Problem 1: Reverse String (LC #344)

Reverse a character array **in-place**.

This is the "hello world" of string problems, but it introduces the **two-pointer swap** pattern that shows up everywhere.

**Approach:** Left pointer at start, right pointer at end. Swap and move inward.

```
  ['h','e','l','l','o']
    L               R     → swap h↔o
  ['o','e','l','l','h']
       L         R        → swap e↔l
  ['o','l','l','e','h']
          L   R           → L >= R, stop!
```

````python
def reverseString(s):
    """
    Two pointer swap. O(n) time, O(1) space.
    """
    left, right = 0, len(s) - 1
    while left < right:
        s[left], s[right] = s[right], s[left]
        left += 1
        right -= 1

chars = ['h','e','l','l','o']
reverseString(chars)
print(chars)  # ['o','l','l','e','h']
````

## 🎯 Problem 2: Reverse Words in a String (LC #151)

Given `"  the sky  is blue  "`, return `"blue is sky the"`.

**Why it's tricky:** Extra spaces everywhere — leading, trailing, between words. You need to handle all of them.

**Pythonic approach:** `split()` handles all whitespace automatically, then reverse and join.

**Interview follow-up:** "Can you do it in-place with O(1) extra space?" → Reverse entire string, then reverse each word individually.

````python
def reverseWords(s):
    """
    split() handles multiple spaces. Reverse the list. Join.
    Time: O(n), Space: O(n)
    """
    return ' '.join(s.split()[::-1])

print(reverseWords("  the sky  is blue  "))  # "blue is sky the"

# Follow-up: In-place approach (if given char array)
def reverseWords_inplace(s):
    """Reverse all, then reverse each word."""
    # Step 1: Clean up spaces and convert to list
    words = s.split()
    chars = list(' '.join(words))
    
    # Step 2: Reverse entire array
    chars.reverse()
    
    # Step 3: Reverse each word back
    n = len(chars)
    start = 0
    for i in range(n + 1):
        if i == n or chars[i] == ' ':
            # Reverse chars[start:i]
            left, right = start, i - 1
            while left < right:
                chars[left], chars[right] = chars[right], chars[left]
                left += 1
                right -= 1
            start = i + 1
    
    return ''.join(chars)

print(reverseWords_inplace("the sky is blue"))  # "blue is sky the"
````

## 🎯 Problem 3: Valid Palindrome (LC #125)

Check if a string is a palindrome, considering only alphanumeric characters and ignoring case.

**Two pointers from both ends.** Skip non-alphanumeric characters. Compare lowercase versions.

````python
def isPalindrome(s):
    """
    Two pointers, skip non-alphanumeric.
    Time: O(n), Space: O(1)
    """
    left, right = 0, len(s) - 1
    while left < right:
        while left < right and not s[left].isalnum():
            left += 1
        while left < right and not s[right].isalnum():
            right -= 1
        if s[left].lower() != s[right].lower():
            return False
        left += 1
        right -= 1
    return True

print(isPalindrome("A man, a plan, a canal: Panama"))  # True
print(isPalindrome("race a car"))                       # False
````

## 🎯 Problem 4: String to Integer (atoi) (LC #8)

This is less about algorithms and more about **careful edge case handling** — which is exactly what interviewers test.

Steps: strip whitespace → check sign → read digits → clamp to 32-bit range.

````python
def myAtoi(s):
    """
    Careful state machine: whitespace → sign → digits → clamp.
    Time: O(n), Space: O(1)
    """
    s = s.lstrip()  # Strip leading whitespace
    if not s:
        return 0
    
    sign = 1
    idx = 0
    
    if s[0] in ('+', '-'):
        sign = -1 if s[0] == '-' else 1
        idx = 1
    
    result = 0
    INT_MAX, INT_MIN = 2**31 - 1, -(2**31)
    
    while idx < len(s) and s[idx].isdigit():
        result = result * 10 + int(s[idx])
        idx += 1
    
    result *= sign
    return max(INT_MIN, min(INT_MAX, result))  # Clamp

print(myAtoi("42"))            # 42
print(myAtoi("   -42"))        # -42
print(myAtoi("4193 with words"))  # 4193
print(myAtoi("-91283472332"))  # -2147483648 (clamped)
````
