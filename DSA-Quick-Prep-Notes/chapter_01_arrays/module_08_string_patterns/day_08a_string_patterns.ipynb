## Strings Are Just Arrays With Attitude

Most string problems are really array problems in disguise — two pointer, sliding window, hashing — but with some string-specific tricks:

- **Palindrome checks** (two pointer from edges or expand from center)
- **Anagram / permutation** (frequency counting)
- **String building** (use lists, not concatenation — O(n) vs O(n²))
- **Character math** (`ord('a')`, `chr(97)`, `ord(ch) - ord('a')` for indexing)

## 🎯 Problem 1: Valid Anagram (LC #242)

Classic warm-up. Shows up as a screening question.

```
Input: s = "anagram", t = "nagaram"  →  True
Input: s = "rat", t = "car"           →  False
```

````python
from collections import Counter

# One-liner
def isAnagram(s, t):
    return Counter(s) == Counter(t)

# Manual approach (what interviewers want to see you understand)
def isAnagram_manual(s, t):
    if len(s) != len(t):
        return False
    
    count = [0] * 26  # 26 lowercase letters
    for i in range(len(s)):
        count[ord(s[i]) - ord('a')] += 1
        count[ord(t[i]) - ord('a')] -= 1
    
    return all(c == 0 for c in count)

print(isAnagram("anagram", "nagaram"))  # True
print(isAnagram("rat", "car"))           # False
````

## 🎯 Problem 2: Longest Palindromic Substring (LC #5)

Top 10 most asked interview question. The trick is **expand from center**.

```
Input: "babad"  →  Output: "bab" (or "aba")
Input: "cbbd"   →  Output: "bb"
```

For each position, try expanding outward from that center. Handle both odd-length (single center) and even-length (two centers) palindromes.

````python
def longestPalindrome(s):
    """
    Expand around center for each position.
    Time: O(n²), Space: O(1)
    """
    def expand(left, right):
        """Expand outward while palindrome."""
        while left >= 0 and right < len(s) and s[left] == s[right]:
            left -= 1
            right += 1
        return s[left+1:right]  # Went one too far in each direction
    
    result = ""
    for i in range(len(s)):
        # Odd length: single center
        odd = expand(i, i)
        # Even length: two centers
        even = expand(i, i + 1)
        
        result = max(result, odd, even, key=len)
    
    return result

print(longestPalindrome("babad"))  # "bab" or "aba"
print(longestPalindrome("cbbd"))   # "bb"
print(longestPalindrome("a"))      # "a"
````

## 🎯 Problem 3: Group Anagrams (LC #49)

```
Input: ["eat","tea","tan","ate","nat","bat"]
Output: [["bat"],["nat","tan"],["ate","eat","tea"]]
```

**Key insight:** Anagrams have the same sorted characters. Use sorted string as a dictionary key.

````python
from collections import defaultdict

def groupAnagrams(strs):
    """
    Group by sorted characters as key.
    Time: O(n * k log k) where k = max word length
    """
    groups = defaultdict(list)
    
    for s in strs:
        key = tuple(sorted(s))  # "eat" → ('a','e','t')
        groups[key].append(s)
    
    return list(groups.values())

# Alternative: use character count as key (O(n * k) — avoids sorting)
def groupAnagrams_v2(strs):
    groups = defaultdict(list)
    for s in strs:
        count = [0] * 26
        for ch in s:
            count[ord(ch) - ord('a')] += 1
        groups[tuple(count)].append(s)
    return list(groups.values())

result = groupAnagrams(["eat","tea","tan","ate","nat","bat"])
for group in result:
    print(group)
````

## 🎯 Problem 4: Encode and Decode Strings (LC #271 / LintCode #659)

Design an algorithm to encode a list of strings to a single string, and decode it back. No delimiter is safe (strings can contain ANY character).

**Trick:** Prefix each string with its length + a separator character.

```
["hello", "world"] → "5#hello5#world"
```

````python
def encode(strs):
    """Encode list of strings to single string."""
    return ''.join(f"{len(s)}#{s}" for s in strs)

def decode(s):
    """Decode single string back to list of strings."""
    result = []
    i = 0
    while i < len(s):
        # Find the '#' separator
        j = s.index('#', i)
        length = int(s[i:j])
        result.append(s[j+1 : j+1+length])
        i = j + 1 + length
    return result

original = ["hello", "world", "foo#bar", "", "123"]
encoded = encode(original)
decoded = decode(encoded)
print(f"Original: {original}")
print(f"Encoded:  {encoded}")
print(f"Decoded:  {decoded}")
print(f"Match:    {original == decoded}")
````

## 💡 Python String Tips for Interviews

Quick reference for things that come up constantly:

````python
# === String building: use list + join (NOT string concatenation) ===
# BAD: O(n²)
# result = ""
# for ch in chars: result += ch

# GOOD: O(n)
chars = ['h', 'e', 'l', 'l', 'o']
result = ''.join(chars)
print(result)  # hello

# === Character ↔ number ===
print(ord('a'))           # 97
print(chr(97))            # 'a'
print(ord('z') - ord('a'))  # 25

# === Check types ===
print("abc123".isalnum())   # True
print("abc".isalpha())      # True
print("123".isdigit())      # True

# === Reverse a string ===
print("hello"[::-1])        # olleh

# === String is immutable → convert to list for in-place ops ===
s = list("hello")
s[0] = 'H'
print(''.join(s))  # Hello
````
