## 🤔 Let's Start Simple

Imagine two people are searching for a name in a phone book with 1,000 pages.

**Person A** starts from page 1 and flips through every single page.

**Person B** opens to the middle, checks if the name comes before or after, and eliminates half the book. Then does it again. And again.

| | Person A | Person B |
|---|---|---|
| 1,000 pages | ~1,000 checks | ~10 checks |
| 1,000,000 pages | ~1,000,000 checks | ~20 checks |
| 1 billion pages | ~1,000,000,000 checks | ~30 checks |

See the pattern? Person A's work grows **linearly** — double the input, double the work. Person B's work grows **logarithmically** — double the input, just one extra check.

**Big O notation** is how we express this difference.
- Person A: **O(n)** — linear
- Person B: **O(log n)** — logarithmic

## 📐 What Big O Actually Means

Big O describes **how your code scales** as the input grows. It answers one question:

> "If I give you 10x more data, how much slower does it get?"

We **don't care** about:
- Exact number of operations (whether it's 2n or 5n, it's still O(n))
- Small inputs (O(n²) might be fine for n=10)
- Constants (O(2n) = O(n), the 2 gets dropped)

We **do care** about:
- The **dominant term** — the part that grows fastest
- **Worst case** behavior (unless specified otherwise)

```
╔═══════════════════════════════════════════════════════════════════════╗
║  SIMPLIFICATION RULES                                                ║
╠═══════════════════════════════════════════════════════════════════════╣
║                                                                       ║
║  1. Drop constants: O(3n) → O(n)                                    ║
║  2. Drop smaller terms: O(n² + n) → O(n²)                          ║
║  3. Only the dominant term matters: O(n³ + n² + n) → O(n³)          ║
║                                                                       ║
║  Why? Because as n gets HUGE, only the biggest term matters.        ║
║  n² + n when n = 1,000,000:                                        ║
║  = 1,000,000,000,000 + 1,000,000                                   ║
║  The +1,000,000 is basically nothing!                               ║
║                                                                       ║
╚═══════════════════════════════════════════════════════════════════════╝
```

## 📊 The Complexity Classes — Know Them Cold

From best to worst:

```
Speed:  FAST ←────────────────────────────────────→ SLOW

        O(1)  O(log n)  O(n)  O(n log n)  O(n²)  O(2ⁿ)  O(n!)
        
n=10:    1      3        10      33         100    1024   3628800
n=100:   1      7       100     664       10000   1.2×10³⁰  LOL
n=1000:  1     10      1000    9966     1000000   ∞        ∞
```

Let me show you what each one looks like in real code...

````python
# === O(1) — Constant Time ===
# No matter how big the input, same number of operations

def get_first(nums):
    return nums[0]  # Always 1 operation, whether list has 10 or 10 million items

def is_even(n):
    return n % 2 == 0  # Always 1 operation

# Dictionary/set lookups are O(1) average
d = {"key": "value"}
val = d["key"]  # O(1) — doesn't matter if dict has 10 or 10 million entries

print("O(1) examples — instant regardless of input size")
````

````python
# === O(log n) — Logarithmic ===
# Halving the problem each step. Binary search is the classic example.

def binary_search(nums, target):
    left, right = 0, len(nums) - 1
    
    while left <= right:
        mid = (left + right) // 2
        if nums[mid] == target:
            return mid
        elif nums[mid] < target:
            left = mid + 1   # Eliminate left half
        else:
            right = mid - 1  # Eliminate right half
    
    return -1

# For 1 billion items, this only takes ~30 steps!
nums = list(range(1000000))
print("Found at index:", binary_search(nums, 777777))
````

````python
# === O(n) — Linear ===
# Look at each element once. Double the input = double the time.

def find_max(nums):
    max_val = nums[0]
    for num in nums:       # One pass through the array
        if num > max_val:
            max_val = num
    return max_val

# Any single for-loop over the input is O(n)
print("Max:", find_max([3, 7, 2, 9, 4]))
````

````python
# === O(n log n) — Linearithmic ===
# This is your sorting complexity. Any comparison-based sort can't do better.

nums = [5, 2, 8, 1, 9, 3]
nums.sort()  # Python uses Timsort — O(n log n)
print("Sorted:", nums)

# Whenever you see "sort first, then solve" — that's O(n log n) minimum
````

````python
# === O(n²) — Quadratic ===
# Nested loops over the same data. This is what interviewers want you to optimize FROM.

def has_pair_with_sum_brute(nums, target):
    """Check if any two numbers sum to target. BRUTE FORCE: O(n²)"""
    for i in range(len(nums)):
        for j in range(i + 1, len(nums)):  # Nested loop!
            if nums[i] + nums[j] == target:
                return True
    return False

def has_pair_with_sum_optimal(nums, target):
    """Same problem, OPTIMAL: O(n) using a set"""
    seen = set()
    for num in nums:
        if target - num in seen:  # O(1) lookup
            return True
        seen.add(num)
    return False

nums = [2, 7, 11, 15]
print("Brute O(n²):", has_pair_with_sum_brute(nums, 9))
print("Optimal O(n):", has_pair_with_sum_optimal(nums, 9))
# Same answer, VASTLY different performance
````

### 💡 A Quick Rule of Thumb for Interviews

When you see the input constraint, this tells you what complexity is expected:

```
╔═══════════════════════════════════════════════════════════════════════╗
║  INPUT SIZE (n)      →      EXPECTED COMPLEXITY                      ║
╠═══════════════════════════════════════════════════════════════════════╣
║                                                                       ║
║  n ≤ 10              →      O(n!) or O(2ⁿ)   [brute force OK]       ║
║  n ≤ 20              →      O(2ⁿ)            [backtracking]          ║
║  n ≤ 500             →      O(n³)            [3 nested loops]        ║
║  n ≤ 5,000           →      O(n²)            [2 nested loops]        ║
║  n ≤ 100,000         →      O(n log n)       [sort + something]      ║
║  n ≤ 1,000,000       →      O(n)             [single pass]           ║
║  n ≤ 10⁹             →      O(log n) or O(1) [binary search/math]   ║
║                                                                       ║
║  Rule: ~10⁸ operations per second is the rough limit                 ║
║                                                                       ║
╚═══════════════════════════════════════════════════════════════════════╝
```

So if a problem says `1 ≤ n ≤ 10⁵`, and you write O(n²) code... that's 10¹⁰ operations. Too slow. You need O(n log n) or better.

## 🧠 Space Complexity — Don't Forget This One

Time gets all the attention, but interviewers ask about space too.

**Space complexity** = how much **extra memory** your solution uses (not counting the input).

| What You Use | Space |
|---|---|
| A few variables | O(1) |
| A hash set/dict of n items | O(n) |
| A 2D matrix n×m | O(n×m) |
| Recursive call stack depth d | O(d) |

**Pro tip:** When the interviewer says "can you do it in-place?" they want O(1) extra space.

````python
# Space O(1) — only using a few variables, no matter how big the input
def find_max_o1_space(nums):
    max_val = nums[0]  # Just one variable
    for num in nums:
        max_val = max(max_val, num)
    return max_val

# Space O(n) — creating extra data structure that grows with input
def find_duplicates_on_space(nums):
    seen = set()   # This set can grow up to n items
    dups = []
    for num in nums:
        if num in seen:
            dups.append(num)
        seen.add(num)
    return dups

print("Max:", find_max_o1_space([3, 7, 2, 9]))
print("Dups:", find_duplicates_on_space([1, 2, 3, 2, 4, 3]))
````

## 🎯 The UMPIRE Framework — Your Problem-Solving Playbook

Every interview problem should be approached the same way. Don't just jump into coding — that's the #1 mistake candidates make.

```
╔═══════════════════════════════════════════════════════════════════════╗
║                      THE UMPIRE FRAMEWORK                             ║
╠═══════════════════════════════════════════════════════════════════════╣
║                                                                       ║
║  U — UNDERSTAND the problem                                          ║
║      • Read it twice. Ask clarifying questions.                      ║
║      • What are the inputs? Outputs? Constraints?                    ║
║      • Walk through examples by hand.                                ║
║                                                                       ║
║  M — MATCH to known patterns                                         ║

║      • "Find pair" → Hash set. "Count frequency" → Hash map.        ║

```

Let me show you this in action...

### 🎯 UMPIRE in Action: Two Sum

**Problem:** Given an array of integers `nums` and an integer `target`, return indices of the two numbers that add up to `target`.

**U — Understand:**
- Input: array + target number
- Output: two indices
- Each input has exactly one solution
- Can't use the same element twice

**M — Match:**
- "Find pair that sums to target" → Hash map pattern!
- We need indices, not just values → store index in hash map

**P — Plan:**
- Walk through array. For each number, check if `target - num` is in our map.
- If yes: return both indices. If no: store current number and index.

**I — Implement:**

````python
def twoSum(nums, target):
    seen = {}  # value → index
    
    for i, num in enumerate(nums):
        complement = target - num
        
        if complement in seen:
            return [seen[complement], i]
        
        seen[num] = i
    
    return []  # No solution found

# R — Review with examples:
print(twoSum([2, 7, 11, 15], 9))   # [0, 1] → 2 + 7 = 9
print(twoSum([3, 2, 4], 6))         # [1, 2] → 2 + 4 = 6
print(twoSum([3, 3], 6))             # [0, 1] → 3 + 3 = 6

# E — Evaluate:
# Time: O(n) — single pass
# Space: O(n) — hash map stores up to n elements
````

See how clean that is? You didn't just solve it — you **demonstrated your thinking process**. That's what gets you hired.

## 📊 How to Quickly Spot Complexity

Here's a cheat sheet for counting complexity by looking at the code structure:

```
┌─────────────────────────────────────────────────────────────────────┐
│  CODE PATTERN                         →   TIME COMPLEXITY           │
├─────────────────────────────────────────────────────────────────────┤
│  No loops, just math/lookup           →   O(1)                     │
│  Single loop: for i in range(n)       →   O(n)                     │
│  Two separate loops (not nested)      →   O(n) + O(n) = O(n)      │
│  Nested loops: for i... for j...      →   O(n²)                    │
│  Loop that halves: while n > 0: n//2  →   O(log n)                │
│  Sort + single loop                   →   O(n log n)              │
│  Recursive with 2 branches            →   O(2ⁿ) (usually)         │
│  Recursive with memoization           →   O(states × work/state)  │
└─────────────────────────────────────────────────────────────────────┘
```

## 📝 Quick Quiz

**Q1.** What's the time complexity?
```python
for i in range(n):
    for j in range(n):
        print(i, j)
```

<details><summary>Answer</summary>

**O(n²)** — nested loop, both go up to n.
</details>

---

**Q2.** What's the time complexity?
```python
for i in range(n):
    print(i)
for j in range(n):
    print(j)
```

<details><summary>Answer</summary>

**O(n)** — two separate loops = O(n) + O(n) = O(2n) = O(n). Drop the constant!
</details>

---

**Q3.** What's the time complexity?
```python
i = n
while i > 0:
    print(i)
    i = i // 2
```

<details><summary>Answer</summary>

**O(log n)** — halving each time. How many times can you halve n? log₂(n) times.
</details>

---

**Q4.** If n = 10⁵, which complexities will pass within the time limit?

<details><summary>Answer</summary>

- O(n log n) = ~1.7 million ops ✅
- O(n²) = 10 billion ops ❌ (too slow!)
- O(n) = 100,000 ops ✅

So you need O(n log n) or better.
</details>
