## The Pattern That Turns O(n²) into O(n)

Two pointer is probably the most **bang-for-your-buck** pattern in all of DSA. Once you see it, you'll notice it everywhere.

The idea is dead simple: instead of using nested loops to check pairs, you use **two pointers** that move towards each other (or in the same direction) to eliminate unnecessary work.

**When to use it:**
- Array/string is **sorted** (or can be sorted)
- Looking for **pairs/triplets** that satisfy a condition
- "Find two numbers that sum to X"
- Removing duplicates in-place
- Palindrome checking

---

## 🧠 The Core Idea

Picture this: you have a **sorted array** and need to find two numbers that add up to a target.

```
nums = [1, 3, 5, 7, 9, 11]    target = 12
        ↑                  ↑
       left              right

sum = 1 + 11 = 12 → FOUND! ✅
```

But what if the sum isn't right?

```
sum = 1 + 11 = 12   → Too big?  Move right pointer LEFT  (smaller sum)
sum = 1 + 9 = 10    → Too small? Move left pointer RIGHT  (bigger sum)
```

Why does this work? Because the array is **sorted**. Moving left pointer right makes the sum bigger. Moving right pointer left makes it smaller. You're guaranteed not to miss the answer.

## 🎯 Problem 1: Two Sum II — Sorted Array (LC #167)

```
Given a SORTED array and a target, return the 1-indexed positions
of two numbers that add up to target.

Input: numbers = [2, 7, 11, 15], target = 9
Output: [1, 2]  (numbers[0] + numbers[1] = 2 + 7 = 9)
```

````python
def twoSum(numbers, target):
    """
    Two Sum II - Input Array Is Sorted
    Time: O(n), Space: O(1)
    """
    left, right = 0, len(numbers) - 1
    
    while left < right:
        current_sum = numbers[left] + numbers[right]
        
        if current_sum == target:
            return [left + 1, right + 1]  # 1-indexed
        elif current_sum < target:
            left += 1   # Need bigger sum → move left pointer right
        else:
            right -= 1  # Need smaller sum → move right pointer left
    
    return []

# Test
print(twoSum([2, 7, 11, 15], 9))    # [1, 2]
print(twoSum([2, 3, 4], 6))          # [1, 3]
print(twoSum([-1, 0], -1))            # [1, 2]
````

**Why not just use a hash map like regular Two Sum?** You could! But the interviewer specifically wants O(1) space here. Two pointer gives you that because the array is already sorted.

## 🎯 Problem 2: 3Sum (LC #15)

This is one of the **most asked** interview problems. It's basically Two Sum II inside a loop.

```
Given an array, find all unique triplets [a, b, c] such that a + b + c = 0.

Input: nums = [-1, 0, 1, 2, -1, -4]
Output: [[-1, -1, 2], [-1, 0, 1]]
```

**The trick:** Fix one number, then use two pointer on the rest.

```
Sort: [-4, -1, -1, 0, 1, 2]

Fix nums[i] = -1, then find two numbers in the rest that sum to +1
         ↑
   fixed (i)    [remaining array → two pointer]
```

````python
def threeSum(nums):
    """
    3Sum — Find all unique triplets that sum to zero.
    Time: O(n²), Space: O(1) extra (ignoring output)
    """
    nums.sort()  # Sort first!
    result = []
    
    for i in range(len(nums) - 2):
        # Skip duplicates for the first number
        if i > 0 and nums[i] == nums[i - 1]:
            continue
        
        # Two pointer on the remaining array
        left, right = i + 1, len(nums) - 1
        
        while left < right:
            total = nums[i] + nums[left] + nums[right]
            
            if total == 0:
                result.append([nums[i], nums[left], nums[right]])
                # Skip duplicates
                while left < right and nums[left] == nums[left + 1]:
                    left += 1
                while left < right and nums[right] == nums[right - 1]:
                    right -= 1
                left += 1
                right -= 1
            elif total < 0:
                left += 1
            else:
                right -= 1
    
    return result

print(threeSum([-1, 0, 1, 2, -1, -4]))
# [[-1, -1, 2], [-1, 0, 1]]
````

**The duplicate-skipping part is the hardest bit.** Make sure you understand why we do `while left < right and nums[left] == nums[left+1]: left += 1`. Without this, you'd get duplicate triplets in your answer.

## 🎯 Problem 3: Container With Most Water (LC #11)

```
Given heights, find two lines that form a container holding the most water.
Area = min(height[left], height[right]) × (right - left)

Input: height = [1, 8, 6, 2, 5, 4, 8, 3, 7]
Output: 49
```

**Why two pointer?** Start with the widest container (left=0, right=end). The only way to get more water is to find taller walls, so move the shorter wall inward.

````python
def maxArea(height):
    """
    Container With Most Water
    Time: O(n), Space: O(1)
    """
    left, right = 0, len(height) - 1
    max_water = 0
    
    while left < right:
        # Calculate water with current pointers
        width = right - left
        h = min(height[left], height[right])
        max_water = max(max_water, width * h)
        
        # Move the shorter wall — it's the bottleneck
        if height[left] < height[right]:
            left += 1
        else:
            right -= 1
    
    return max_water

print(maxArea([1, 8, 6, 2, 5, 4, 8, 3, 7]))  # 49
print(maxArea([1, 1]))                          # 1
````

## 🎯 Problem 4: Valid Palindrome (LC #125)

This is the **same-direction-from-both-ends** flavor of two pointer.

```
Check if a string is a palindrome (ignoring non-alphanumeric + case).

Input: "A man, a plan, a canal: Panama"
Output: True
```

````python
def isPalindrome(s):
    """
    Valid Palindrome
    Time: O(n), Space: O(1)
    """
    left, right = 0, len(s) - 1
    
    while left < right:
        # Skip non-alphanumeric characters
        while left < right and not s[left].isalnum():
            left += 1
        while left < right and not s[right].isalnum():
            right -= 1
        
        if s[left].lower() != s[right].lower():
            return False
        
        left += 1
        right -= 1
    
    return True

print(isPalindrome("A man, a plan, a canal: Panama"))  # True
print(isPalindrome("race a car"))                       # False
print(isPalindrome(" "))                                 # True
````

## 🎯 Problem 5: Remove Duplicates from Sorted Array (LC #26)

This is the **slow-fast pointer** variation — both pointers move in the same direction.

```
Remove duplicates IN-PLACE from sorted array. Return the number of unique elements.

Input: nums = [1, 1, 2, 2, 3]
Output: 3, nums = [1, 2, 3, _, _]
```

````python
def removeDuplicates(nums):
    """
    Remove Duplicates from Sorted Array (in-place)
    Time: O(n), Space: O(1)
    
    slow = position to place next unique element
    fast = scanner that looks for new unique elements
    """
    if not nums:
        return 0
    
    slow = 0  # Points to last unique element
    
    for fast in range(1, len(nums)):
        if nums[fast] != nums[slow]:  # Found a new unique element!
            slow += 1
            nums[slow] = nums[fast]
    
    return slow + 1  # Number of unique elements

nums = [1, 1, 2, 2, 3]
k = removeDuplicates(nums)
print(f"Unique count: {k}, Array: {nums[:k]}")

nums2 = [0, 0, 1, 1, 1, 2, 2, 3, 3, 4]
k2 = removeDuplicates(nums2)
print(f"Unique count: {k2}, Array: {nums2[:k2]}")
````

## 🗺️ Two Pointer Pattern Map

```
╔═══════════════════════════════════════════════════════════════════════╗
║  TWO POINTER VARIATIONS                                              ║
╠═══════════════════════════════════════════════════════════════════════╣
║                                                                       ║
║  1. OPPOSITE ENDS (converging)                                       ║
║     left = 0, right = n-1, move towards each other                   ║
║     → Two Sum sorted, Container Water, Palindrome, Trapping Rain     ║
║                                                                       ║
║  2. SAME DIRECTION (slow-fast)                                       ║
║     slow & fast both start left, fast moves ahead                    ║
║     → Remove duplicates, Move zeros, Partition                       ║
║                                                                       ║
║  3. FIX ONE + TWO POINTER                                            ║
║     Fix element with outer loop, two pointer on rest                 ║
║     → 3Sum, 4Sum, closest sum                                        ║
║                                                                       ║
╚═══════════════════════════════════════════════════════════════════════╝
```

├─────────────────────────────────────────────────────────────────────┤
│  📌 Opposite ends: start from both sides, converge to middle      │
├─────────────────────────────────────────────────────────────────────┤
