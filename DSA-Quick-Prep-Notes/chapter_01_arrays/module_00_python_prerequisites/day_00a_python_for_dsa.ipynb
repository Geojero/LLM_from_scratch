## 📋 1. Lists — Your Most-Used Data Structure

90% of DSA problems involve lists (arrays). Here's everything you need to know, fast.

### The Operations That Matter

````python
# === CREATING ===
nums = [1, 2, 3, 4, 5]
zeros = [0] * 5                    # [0, 0, 0, 0, 0] — you'll use this A LOT for DP tables
grid = [[0] * 3 for _ in range(3)] # 3x3 grid — NEVER do [[0]*3]*3 (shared reference trap!)

print("zeros:", zeros)
print("grid:", grid)

# === ACCESSING ===
print("\nFirst:", nums[0])     # 1
print("Last:", nums[-1])       # 5 — negative indexing is your friend
print("Second last:", nums[-2]) # 4
````

````python
# === SLICING — super important, know this cold ===
nums = [10, 20, 30, 40, 50, 60, 70]

print("nums[1:4]:", nums[1:4])    # [20, 30, 40] — start inclusive, end EXCLUSIVE
print("nums[:3]:", nums[:3])       # [10, 20, 30] — first 3
print("nums[3:]:", nums[3:])       # [40, 50, 60, 70] — from index 3 onwards
print("nums[::2]:", nums[::2])     # [10, 30, 50, 70] — every 2nd element
print("nums[::-1]:", nums[::-1])   # [70, 60, 50, 40, 30, 20, 10] — REVERSE!
````

````python
# === MODIFYING ===
nums = [1, 2, 3]

nums.append(4)         # Add to end: [1, 2, 3, 4] — O(1)
nums.pop()             # Remove from end: [1, 2, 3] — O(1)
nums.pop(0)            # Remove from front: [2, 3] — O(n)! Careful!
nums.insert(0, 1)      # Insert at index 0: [1, 2, 3] — O(n)!

# Here's the thing — append/pop from END is O(1), but from FRONT is O(n)
# If you need fast front operations, use collections.deque instead

print("After operations:", nums)
````

````python
# === LIST COMPREHENSIONS — write cleaner code, impress the interviewer ===

# Instead of:
squares = []
for i in range(5):
    squares.append(i * i)

# Do this:
squares = [i * i for i in range(5)]
print("Squares:", squares)  # [0, 1, 4, 9, 16]

# With condition:
evens = [x for x in range(10) if x % 2 == 0]
print("Evens:", evens)  # [0, 2, 4, 6, 8]

# Flatten 2D list:
matrix = [[1, 2], [3, 4], [5, 6]]
flat = [num for row in matrix for num in row]
print("Flattened:", flat)  # [1, 2, 3, 4, 5, 6]
````

### ⚠️ The #1 List Trap in Interviews

This one bites people all the time:

```python
# WRONG — all rows share the SAME list object!
grid = [[0] * 3] * 3
grid[0][0] = 1
print(grid)  # [[1, 0, 0], [1, 0, 0], [1, 0, 0]]  ← ALL rows changed!

# RIGHT — each row is an independent list
grid = [[0] * 3 for _ in range(3)]
grid[0][0] = 1
print(grid)  # [[1, 0, 0], [0, 0, 0], [0, 0, 0]]  ← Only first row changed
```

**Rule of thumb:** Always use list comprehension for 2D arrays.

## 🗂️ 2. Dictionaries & Sets — Your O(1) Weapons

If lists are your bread, dictionaries and sets are your butter. You'll reach for these whenever a problem says "find", "count", "check if exists".

````python
# === DICTIONARY — key-value pairs, O(1) average lookup ===

# Counting frequency — you'll do this in SO many problems
nums = [1, 2, 2, 3, 3, 3]

# Method 1: Manual counting
freq = {}
for num in nums:
    freq[num] = freq.get(num, 0) + 1  # .get() avoids KeyError
print("Frequency:", freq)  # {1: 1, 2: 2, 3: 3}

# Method 2: Counter (even faster to write)
from collections import Counter
freq = Counter(nums)
print("Counter:", freq)  # Counter({3: 3, 2: 2, 1: 1})
print("Most common:", freq.most_common(2))  # [(3, 3), (2, 2)]
````

````python
# === defaultdict — no more checking "if key in dict" ===
from collections import defaultdict

# Grouping — another super common pattern
words = ["eat", "tea", "tan", "ate", "nat", "bat"]

# Group anagrams together
groups = defaultdict(list)
print(groups)
for word in words:
    key = "".join(sorted(word))  # "eat" → "aet", "tea" → "aet"
    groups[key].append(word)

print("Anagram groups:")
for key, group in groups.items():
    print(f"  {key}: {group}")
````

````python
# === SET — unique elements, O(1) lookup, great for "have I seen this?" ===

# Check for duplicates
nums = [1, 2, 3, 2, 4]
print("Has duplicates?", len(nums) != len(set(nums)))  # True

# Set operations you'll use:
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

print("Union:", a | b)         # {1, 2, 3, 4, 5, 6}
print("Intersection:", a & b)  # {3, 4}
print("Difference:", a - b)    # {1, 2}

# "Seen" set pattern — used in Two Sum, graph traversal, etc.
seen = set()
for num in [3, 1, 4, 1, 5]:
    if num in seen:  # O(1) check!
        print(f"Duplicate found: {num}")
    seen.add(num)
````

## 📝 3. Strings — Immutable but Important

Here's the thing about Python strings that trips people up: **you can't modify them in-place**. Every "modification" creates a new string. So when you need to build a string character by character, use a **list** and join at the end.

````python
# Strings are IMMUTABLE — can't do s[0] = 'X'
s = "hello"

# Converting to list for manipulation
chars = list(s)        # ['h', 'e', 'l', 'l', 'o']
chars[0] = 'H'
s = "".join(chars)     # "Hello"
print(s)

# String methods you'll use constantly:
s = "  Hello, World!  "
print(s.strip())       # "Hello, World!" — remove whitespace
print(s.lower())       # "  hello, world!  "
print(s.split(","))    # ['  Hello', ' World!  ']
print("abc".isalpha()) # True
print("123".isdigit()) # True
print(ord('a'))        # 97 — ASCII value (useful for char math)
print(chr(97))         # 'a' — back to character
````

````python
# === Building strings efficiently ===

# SLOW — creates new string each time: O(n²)
# result = ""
# for char in "hello":
#     result += char.upper()

# FAST — use list then join: O(n)
result = []
for char in "hello":
    result.append(char.upper())
result = "".join(result)
print(result)  # "HELLO"

# Even faster — list comprehension:
result = "".join([char.upper() for char in "hello"])
print(result)  # "HELLO"
````

## 🔄 4. Sorting & Lambda — The Interview Power Combo

Sorting comes up everywhere. Knowing how to sort with custom keys is a must.

````python
# Basic sorting
nums = [3, 1, 4, 1, 5, 9]
print("sorted():", sorted(nums))  # Returns NEW list, original unchanged

nums.sort()                        # Sorts IN-PLACE, returns None
print("nums.sort():", nums)

# Reverse sort
print("Descending:", sorted([3, 1, 4], reverse=True))  # [4, 3, 1]
````

````python
# === Custom sorting with lambda — this is where it gets powerful ===

# Sort by second element (like sorting intervals by end time)
intervals = [[3, 7], [1, 5], [2, 3], [4, 8]]
intervals.sort(key=lambda x: x[1])  # Sort by end time
print("By end time:", intervals)

# Sort by length
words = ["banana", "pie", "apple", "kiwi"]
words.sort(key=lambda w: len(w))
print("By length:", words)

# Sort by multiple criteria (sort by length, then alphabetically)
words = ["banana", "pie", "apple", "kiwi", "fig"]
words.sort(key=lambda w: (len(w), w))  # Tuple comparison!
print("By length then alpha:", words)
````

## 🧰 5. Collections You Need to Know

Python's `collections` module is a goldmine for DSA. Here are the ones you'll actually use:

````python
from collections import deque, Counter, defaultdict, OrderedDict
import heapq

# === deque — double-ended queue, O(1) from both ends ===
# Use this for BFS, sliding window, and any time you need fast front operations
q = deque([1, 2, 3])
q.append(4)       # Right end: [1, 2, 3, 4]
q.appendleft(0)   # Left end: [0, 1, 2, 3, 4]
q.pop()            # From right: 4
q.popleft()        # From left: 0
print("deque:", q)

# === heapq — min-heap, O(log n) push/pop ===
# Python only has MIN heap. For max heap, negate values.
heap = []
heapq.heappush(heap, 3)
heapq.heappush(heap, 1)
heapq.heappush(heap, 2)
print("Min:", heapq.heappop(heap))  # 1 (smallest)

# Max heap trick:
max_heap = []
for num in [3, 1, 2]:
    heapq.heappush(max_heap, -num)  # Negate!
print("Max:", -heapq.heappop(max_heap))  # 3 (largest)
````

## ⚡ 6. Python Tricks That Save Time in Interviews

These are the things that make your code look clean and your interviewer nod approvingly.

````python
# === Swap without temp variable ===
a, b = 1, 2
a, b = b, a  # Python magic
print(f"After swap: a={a}, b={b}")

# === Multiple assignment ===
x = y = z = 0
left, right = 0, len([1,2,3]) - 1  # Two pointer init

# === Infinity — useful for min/max tracking ===
min_val = float('inf')   # Positive infinity
max_val = float('-inf')  # Negative infinity

# === Enumerate — when you need both index AND value ===
fruits = ["apple", "banana", "cherry"]
for i, fruit in enumerate(fruits):
    print(f"  {i}: {fruit}")

# === Zip — iterate two lists together ===
names = ["Alice", "Bob"]
scores = [95, 87]
for name, score in zip(names, scores):
    print(f"  {name}: {score}")
````

````python
# === Unpacking — clean and readable ===
point = (3, 5)
x, y = point

# Works great with intervals/pairs:
intervals = [[1, 3], [2, 5], [4, 7]]
for start, end in intervals:
    print(f"  [{start}, {end}]")

# === Ternary (inline if-else) ===
x = 10
result = "even" if x % 2 == 0 else "odd"
print(f"{x} is {result}")

# === any() and all() — check conditions across a list ===
nums = [2, 4, 6, 8]
print("All even?", all(n % 2 == 0 for n in nums))  # True
print("Any > 5?", any(n > 5 for n in nums))        # True
````
