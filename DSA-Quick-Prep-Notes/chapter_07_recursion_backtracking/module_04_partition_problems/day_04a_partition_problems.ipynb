## The Partition Pattern

Try every possible split point in a string/array. If the left part is valid, recurse on the right part.

```
s = "aab"
Try splits: "a" | "ab"  →  "a" valid palindrome → recurse on "ab"
            "aa" | "b"  →  "aa" valid palindrome → recurse on "b"
            "aab"       →  not palindrome → skip
```

---

## 🎯 Problem 1: Palindrome Partitioning (LC #131)

Partition string so every substring is a palindrome. Return all possible partitions.

````python
def partition(s):
    """
    Backtracking: try each prefix, if palindrome → recurse on rest.
    Time: O(n * 2^n), Space: O(n)
    """
    result = []
    
    def is_palindrome(sub):
        return sub == sub[::-1]
    
    def backtrack(start, path):
        if start == len(s):
            result.append(path[:])
            return
        for end in range(start + 1, len(s) + 1):
            prefix = s[start:end]
            if is_palindrome(prefix):
                path.append(prefix)
                backtrack(end, path)
                path.pop()
    
    backtrack(0, [])
    return result

print(partition("aab"))
# [['a', 'a', 'b'], ['aa', 'b']]
````

## 🎯 Problem 2: Restore IP Addresses (LC #93)

Insert dots to create valid IP addresses from a digit string.

**Constraints per segment:** 1-3 digits, value 0-255, no leading zeros (except "0" itself).

````python
def restoreIpAddresses(s):
    """
    Backtracking: place 3 dots, validate each segment.
    Time: O(1) bounded by IP format, Space: O(1)
    """
    result = []
    
    def backtrack(start, segments):
        if len(segments) == 4:
            if start == len(s):  # Used all digits
                result.append(".".join(segments))
            return
        
        for length in range(1, 4):  # Try 1, 2, or 3 digits
            if start + length > len(s):
                break
            segment = s[start:start + length]
            # Validate: no leading zeros, value <= 255
            if (len(segment) > 1 and segment[0] == '0') or int(segment) > 255:
                continue
            segments.append(segment)
            backtrack(start + length, segments)
            segments.pop()
    
    backtrack(0, [])
    return result

print(restoreIpAddresses("25525511135"))
# ['255.255.11.135', '255.255.111.35']
````
