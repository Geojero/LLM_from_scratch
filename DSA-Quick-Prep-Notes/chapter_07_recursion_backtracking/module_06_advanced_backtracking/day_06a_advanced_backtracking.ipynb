## Making Backtracking Fast

Backtracking's worst case is exponential. The real skill is **pruning** — cutting branches early so you never explore dead ends.

**Pruning strategies:**
1. **Sort + early termination** — if current candidate too large, all future ones are too
2. **Constraint propagation** — use sets/arrays for O(1) validity checks
3. **Bound checking** — if remaining capacity can't possibly improve, stop
4. **Symmetry breaking** — don't explore rotations/reflections of the same solution

## 🎯 Problem 1: Partition to K Equal Sum Subsets (LC #698)

Can you divide the array into K subsets with equal sum?

**This is a hard backtracking problem.** We assign each number to one of K buckets. Key optimizations:
- Sort descending (larger numbers fail faster → prune sooner)
- Skip identical bucket values (symmetry breaking)
- If a bucket is empty and we skip it, all subsequent empty buckets would be the same

````python
def canPartitionKSubsets(nums, k):
    """
    Assign numbers to K buckets with heavy pruning.
    Time: O(k * 2^n) with pruning, Space: O(n)
    """
    total = sum(nums)
    if total % k != 0:
        return False
    target = total // k
    nums.sort(reverse=True)  # Prune: big numbers first fail faster
    if nums[0] > target:
        return False
    
    buckets = [0] * k
    
    def backtrack(idx):
        if idx == len(nums):
            return all(b == target for b in buckets)
        
        seen = set()  # Symmetry breaking: don't try same bucket value twice
        for i in range(k):
            if buckets[i] + nums[idx] > target:
                continue  # Pruning: would overflow
            if buckets[i] in seen:
                continue  # Symmetry: same state already tried
            seen.add(buckets[i])
            
            buckets[i] += nums[idx]
            if backtrack(idx + 1):
                return True
            buckets[i] -= nums[idx]
            
            if buckets[i] == 0:
                break  # If empty bucket didn't work, no other empty will
        
        return False
    
    return backtrack(0)

print(canPartitionKSubsets([4, 3, 2, 3, 5, 2, 1], 4))  # True (target=5)
print(canPartitionKSubsets([1, 2, 3, 4], 3))             # False
````

## 🎯 Problem 2: Letter Combinations of a Phone Number (LC #17)

Given a phone digit string, return all possible letter combinations.

**Simple backtracking with multiple choice arrays.** Each digit maps to 3-4 letters.

````python
def letterCombinations(digits):
    """
    BFS-like backtracking over phone pad.
    Time: O(4^n * n), Space: O(n)
    """
    if not digits:
        return []
    
    phone = {
        '2': 'abc', '3': 'def', '4': 'ghi', '5': 'jkl',
        '6': 'mno', '7': 'pqrs', '8': 'tuv', '9': 'wxyz'
    }
    result = []
    
    def backtrack(idx, path):
        if idx == len(digits):
            result.append("".join(path))
            return
        for letter in phone[digits[idx]]:
            path.append(letter)
            backtrack(idx + 1, path)
            path.pop()
    
    backtrack(0, [])
    return result

print(letterCombinations("23"))
# ['ad','ae','af','bd','be','bf','cd','ce','cf']
````
