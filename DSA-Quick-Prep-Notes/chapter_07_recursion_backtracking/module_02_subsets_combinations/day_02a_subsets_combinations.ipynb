# 📦 Day 02: Subsets & Combinations — Include or Exclude

## The Core Idea

For each element, you have a binary choice: **include it or exclude it**. This gives you 2^n subsets.

```
nums = [1, 2, 3]

                  []
              /        \
          [1]            []
         /    \        /    \
      [1,2]  [1]    [2]    []
      / \    / \    / \    / \
  [123][12][13][1][23][2][3]  []
```

**Backtracking template for subsets:**
```python
def backtrack(start, path):
    result.append(path[:])  # Every path is a valid subset
    for i in range(start, len(nums)):
        path.append(nums[i])
        backtrack(i + 1, path)  # i+1: don't reuse
        path.pop()              # Unchoose
```

## 🎯 Problem 1: Subsets (LC #78)

Generate all possible subsets of an array with distinct elements.

````python
def subsets(nums):
    """
    Backtracking: at each level, choose to include next elements.
    Time: O(n * 2^n), Space: O(n)
    """
    result = []
    
    def backtrack(start, path):
        result.append(path[:])  # Every path is a valid subset
        for i in range(start, len(nums)):
            path.append(nums[i])
            backtrack(i + 1, path)
            path.pop()
    
    backtrack(0, [])
    return result

print(subsets([1, 2, 3]))
# [[], [1], [1,2], [1,2,3], [1,3], [2], [2,3], [3]]
````

## 🎯 Problem 2: Subsets II — With Duplicates (LC #90)

Same but input may have duplicates. Skip duplicate branches.

**Key trick:** Sort first. Then if `nums[i] == nums[i-1]` and we didn't use `i-1` in this branch, skip `i` (it would create a duplicate subset).

````python
def subsetsWithDup(nums):
    """
    Sort + skip duplicates at same level.
    Time: O(n * 2^n), Space: O(n)
    """
    nums.sort()  # Sort to group duplicates
    result = []
    
    def backtrack(start, path):
        result.append(path[:])
        for i in range(start, len(nums)):
            # Skip duplicates at the same decision level
            if i > start and nums[i] == nums[i-1]:
                continue
            path.append(nums[i])
            backtrack(i + 1, path)
            path.pop()
    
    backtrack(0, [])
    return result

print(subsetsWithDup([1, 2, 2]))
# [[], [1], [1,2], [1,2,2], [2], [2,2]]
````

## 🎯 Problem 3: Combination Sum (LC #39)

Find combinations that sum to target. **Elements can be reused.**

**Difference from subsets:** We have a target. We stop when sum ≥ target. And we can reuse (pass `i` not `i+1`).

````python
def combinationSum(candidates, target):
    """
    Backtracking with reuse (pass i, not i+1).
    Time: O(n^(target/min)), Space: O(target/min)
    """
    result = []
    candidates.sort()
    
    def backtrack(start, path, remaining):
        if remaining == 0:
            result.append(path[:])
            return
        for i in range(start, len(candidates)):
            if candidates[i] > remaining:
                break  # Pruning: sorted, so all future too large
            path.append(candidates[i])
            backtrack(i, path, remaining - candidates[i])  # i, not i+1 → reuse
            path.pop()
    
    backtrack(0, [], target)
    return result

print(combinationSum([2, 3, 6, 7], 7))  # [[2,2,3], [7]]
````

## 🎯 Problem 4: Combination Sum II (LC #40)

Same but each number used at most once. Input may have duplicates.

**Combine both tricks:** `i+1` (no reuse) + skip duplicates at same level.

````python
def combinationSum2(candidates, target):
    """
    No reuse + skip duplicates.
    """
    candidates.sort()
    result = []
    
    def backtrack(start, path, remaining):
        if remaining == 0:
            result.append(path[:])
            return
        for i in range(start, len(candidates)):
            if candidates[i] > remaining:
                break
            if i > start and candidates[i] == candidates[i-1]:
                continue  # Skip duplicates
            path.append(candidates[i])
            backtrack(i + 1, path, remaining - candidates[i])
            path.pop()
    
    backtrack(0, [], target)
    return result

print(combinationSum2([10,1,2,7,6,1,5], 8))
# [[1,1,6], [1,2,5], [1,7], [2,6]]
````

## 🗺️ Subsets/Combinations Cheat Sheet

```
╔══════════════════════════════════════════════════════════════════════╗
║  SUBSETS & COMBINATIONS                                             ║
╠══════════════════════════════════════════════════════════════════════╣
║                                                                      ║
║  Subsets:           backtrack(i+1), collect at every node           ║
║  Subsets w/ dups:   sort + skip if nums[i]==nums[i-1] & i>start    ║
║  Combination Sum:   backtrack(i) → reuse allowed                   ║
║  Comb Sum no reuse: backtrack(i+1) + skip dups                     ║
║                                                                      ║
║  PRUNING: sort array + break when candidate > remaining            ║
║                                                                      ║
╚══════════════════════════════════════════════════════════════════════╝
```

## ✅ Key Takeaways

```
┌──────────────────────────────────────────────────────────────────────┐
│  📌 Subsets = collect every node in the recursion tree              │
├──────────────────────────────────────────────────────────────────────┤
│  📌 Reuse → pass i. No reuse → pass i+1.                         │
├──────────────────────────────────────────────────────────────────────┤
│  📌 Duplicates: sort + skip same value at same level               │
├──────────────────────────────────────────────────────────────────────┤
│  📌 Always sort + prune (break when > remaining) for efficiency    │
└──────────────────────────────────────────────────────────────────────┘
```
