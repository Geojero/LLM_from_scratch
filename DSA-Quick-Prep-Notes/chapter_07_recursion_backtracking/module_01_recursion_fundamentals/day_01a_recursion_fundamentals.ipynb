## How to Think Recursively

Recursion is NOT about tracing every call. It's about **trust.**

1. **Define what the function does** (not how)
2. **Find the base case** (smallest input where answer is obvious)
3. **Assume recursion works for smaller input** — how do you build from that?

```
Think: "If a magic function solves the smaller problem,
        how do I use that to solve the current problem?"
```

### Example: Reverse a linked list
- Base: empty or single node → return it
- Assume `reverse(head.next)` correctly reverses the rest
- Now just make `head.next.next = head; head.next = None`

## 🎯 Problem 1: Power Function (LC #50)

Compute `x^n` efficiently.

**Naive:** Multiply x, n times → O(n). 

**Recursive insight:** x^n = (x^(n/2))² → O(log n)!

````python
def myPow(x, n):
    """
    Fast exponentiation: x^n = (x^(n/2))^2
    Time: O(log n), Space: O(log n) call stack
    """
    if n == 0:
        return 1
    if n < 0:
        return 1 / myPow(x, -n)
    
    half = myPow(x, n // 2)
    if n % 2 == 0:
        return half * half
    else:
        return half * half * x

print(myPow(2.0, 10))   # 1024.0
print(myPow(2.0, -2))   # 0.25
````

## 🎯 Problem 2: Generate Parentheses (LC #22)

Generate all valid combinations of `n` pairs of parentheses.

**Recursive choice at each position:**
- Add `(` if we have open parens left
- Add `)` if close count < open count (keeps it valid)

````python
def generateParenthesis(n):
    """
    Recursion with validity constraint.
    Time: O(4^n / √n) Catalan, Space: O(n)
    """
    result = []
    
    def build(path, open_count, close_count):
        if len(path) == 2 * n:
            result.append("".join(path))
            return
        if open_count < n:
            path.append('(')
            build(path, open_count + 1, close_count)
            path.pop()  # Backtrack
        if close_count < open_count:
            path.append(')')
            build(path, open_count, close_count + 1)
            path.pop()  # Backtrack
    
    build([], 0, 0)
    return result

print(generateParenthesis(3))
# ['((()))', '(()())', '(())()', '()(())', '()()()']
````

**This is our first taste of backtracking:** make a choice → recurse → undo the choice (pop). This "choose-explore-unchoose" is the backtracking template.
