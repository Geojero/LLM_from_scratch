## 🎯 Problem 1: Design HashMap (LC #706)

Implement put, get, remove without using built-in hash tables.

**Array of buckets + chaining (linked list per bucket).** Use modulo for hashing.

````python
class MyHashMap:
    def __init__(self):
        self.size = 1000
        self.buckets = [[] for _ in range(self.size)]
    
    def _hash(self, key):
        return key % self.size
    
    def put(self, key, value):
        h = self._hash(key)
        for i, (k, v) in enumerate(self.buckets[h]):
            if k == key:
                self.buckets[h][i] = (key, value)
                return
        self.buckets[h].append((key, value))
    
    def get(self, key):
        h = self._hash(key)
        for k, v in self.buckets[h]:
            if k == key:
                return v
        return -1
    
    def remove(self, key):
        h = self._hash(key)
        self.buckets[h] = [(k, v) for k, v in self.buckets[h] if k != key]

hm = MyHashMap()
hm.put(1, 1); hm.put(2, 2)
print(hm.get(1))    # 1
print(hm.get(3))    # -1
hm.remove(2)
print(hm.get(2))    # -1
````

## 🎯 Problem 2: Implement Queue using Stacks (LC #232)

**Two stacks:** `in_stack` for push, `out_stack` for pop/peek. Transfer when `out_stack` is empty.

Each element gets moved at most twice → **amortized O(1)** per operation.

````python
class MyQueue:
    def __init__(self):
        self.in_stack = []   # Push here
        self.out_stack = []  # Pop from here
    
    def push(self, x):
        self.in_stack.append(x)
    
    def _transfer(self):
        if not self.out_stack:
            while self.in_stack:
                self.out_stack.append(self.in_stack.pop())
    
    def pop(self):
        self._transfer()
        return self.out_stack.pop()
    
    def peek(self):
        self._transfer()
        return self.out_stack[-1]
    
    def empty(self):
        return not self.in_stack and not self.out_stack

q = MyQueue()
q.push(1); q.push(2)
print(q.peek())  # 1
print(q.pop())   # 1
print(q.empty()) # False
````

╠═══════════════════════════════════════════════════════════════════════╣

║  HASHMAP FROM SCRATCH: Array of buckets + chaining for collisions    ║
║  → hash(key) % size → bucket index → linked list search             ║
║  → Key decisions: hash function, collision handling, resizing        ║
║                                                                       ║
║  STACK ↔ QUEUE CONVERSION:                                           ║
║  → Queue from 2 stacks: lazy transfer (amortized O(1))             ║
║  → Stack from queue: rotate on push                                  ║
║                                                                       ║
║  GENERAL APPROACH:                                                    ║
║  1. Identify required operations and their time complexity           ║
║  2. Choose underlying data structures to meet those constraints      ║
║  3. Often combine two structures (hashmap + list, etc.)              ║
