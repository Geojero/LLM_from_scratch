## How to Approach Design Problems

Design problems in coding interviews are NOT system design. They ask you to **implement a class** with specific operations, each with optimal time complexity.

**Framework:**
1. List all required operations and their complexity targets
2. Pick data structures that give those complexities
3. Combine multiple data structures if needed (HashMap + LinkedList, etc.)

---

## Common Combos

```
╔══════════════════════════════════════════════════════════════════════╗
║  Need O(1) lookup + O(1) insert/delete → HashMap + DLL           ║
║  Need O(1) insert + O(1) random       → List + HashMap            ║

║  Need ordered iteration                → BST or sorted container  ║

```

## 🎯 Problem 1: Min Stack (LC #155)

Stack that supports push, pop, top, and getMin — all in O(1).

**Store (value, current_min) pairs.** Each entry remembers the min at that point in time.

````python
class MinStack:
    def __init__(self):
        self.stack = []  # Each entry: (val, min_so_far)
    
    def push(self, val):
        curr_min = min(val, self.stack[-1][1] if self.stack else val)
        self.stack.append((val, curr_min))
    
    def pop(self):
        self.stack.pop()
    
    def top(self):
        return self.stack[-1][0]
    
    def getMin(self):
        return self.stack[-1][1]

ms = MinStack()
ms.push(-2); ms.push(0); ms.push(-3)
print(ms.getMin())  # -3
ms.pop()
print(ms.getMin())  # -2
````

## 🎯 Problem 2: Insert Delete GetRandom O(1) (LC #380)

All three operations in O(1).

**HashMap (val → index) + List.** Insert at end. Delete by swapping with last. Random via random index.

````python
import random

class RandomizedSet:
    def __init__(self):
        self.val_to_idx = {}  # val → index in list
        self.vals = []         # values
    
    def insert(self, val):
        if val in self.val_to_idx:
            return False
        self.val_to_idx[val] = len(self.vals)
        self.vals.append(val)
        return True
    
    def remove(self, val):
        if val not in self.val_to_idx:
            return False
        idx = self.val_to_idx[val]
        last = self.vals[-1]
        # Swap with last
        self.vals[idx] = last
        self.val_to_idx[last] = idx
        # Remove last
        self.vals.pop()
        del self.val_to_idx[val]
        return True
    
    def getRandom(self):
        return random.choice(self.vals)

rs = RandomizedSet()
print(rs.insert(1))  # True
print(rs.remove(2))  # False
print(rs.insert(2))  # True
print(rs.getRandom()) # 1 or 2
````
