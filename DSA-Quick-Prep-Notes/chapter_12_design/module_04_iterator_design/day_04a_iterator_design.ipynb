## Iterator Pattern

Iterator problems ask you to lazily produce elements one at a time from a complex structure. The key is managing state — typically with a **stack** that simulates the traversal.

---

## 🎯 Problem 1: Flatten Nested List Iterator (LC #341)

Given a nested list of integers, implement an iterator that returns them in order.

**Stack-based approach:** Push elements in reverse order. On `hasNext`, keep flattening the top until it's an integer.

````python
# Simulating NestedInteger interface
class NestedInteger:
    def __init__(self, val=None, lst=None):
        self._val = val
        self._list = lst or []
    def isInteger(self): return self._val is not None
    def getInteger(self): return self._val
    def getList(self): return self._list

class NestedIterator:
    def __init__(self, nestedList):
        self.stack = nestedList[::-1]  # Reverse so first element is on top
    
    def next(self):
        return self.stack.pop().getInteger()
    
    def hasNext(self):
        while self.stack and not self.stack[-1].isInteger():
            # Flatten: pop list, push its elements in reverse
            nested = self.stack.pop().getList()
            self.stack.extend(nested[::-1])
        return len(self.stack) > 0

# Test: [[1,1],2,[1,1]]
nl = [
    NestedInteger(lst=[NestedInteger(1), NestedInteger(1)]),
    NestedInteger(2),
    NestedInteger(lst=[NestedInteger(1), NestedInteger(1)])
]
it = NestedIterator(nl)
result = []
while it.hasNext():
    result.append(it.next())
print(result)  # [1,1,2,1,1]
````

## 🎯 Problem 2: Peeking Iterator (LC #284)

Add `peek()` to an existing iterator without advancing it.

**Cache the next value.** On `peek`, return cached. On `next`, return cached and advance.

````python
class PeekingIterator:
    def __init__(self, iterator):
        self.iterator = iterator
        self._next = next(iterator, None)
    
    def peek(self):
        return self._next
    
    def next(self):
        val = self._next
        self._next = next(self.iterator, None)
        return val
    
    def hasNext(self):
        return self._next is not None

it = PeekingIterator(iter([1,2,3]))
print(it.peek())    # 1
print(it.next())    # 1
print(it.next())    # 2
print(it.hasNext()) # True
````

```
║  → Key: hasNext() does the "flattening" work, next() just pops     ║
║                                                                       ║
║  PEEKING ITERATOR: Cache the next element for look-ahead             ║
║  → Wrap existing iterator, store peeked value                        ║
║  → peek() returns cached, next() clears cache and advances          ║
║                                                                       ║
║  GENERAL ITERATOR APPROACH:                                           ║
║  1. Maintain internal state (stack, pointer, cached value)           ║
║  2. hasNext() may need to advance state lazily                       ║
║  3. next() returns current and advances                               ║
║                                                                       ║

```
