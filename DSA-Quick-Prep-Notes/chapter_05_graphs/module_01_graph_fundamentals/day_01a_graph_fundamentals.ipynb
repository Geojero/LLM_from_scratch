## Trees Were Just a Warm-Up

Trees are graphs with restrictions (no cycles, one root, one path between nodes). Graphs remove all restrictions — any node can connect to any other node, cycles can exist, and there's no "root."

```
Directed:    A → B → C       (edges have direction)
             ↑       |
             └───────┘

Undirected:  A — B — C       (edges go both ways)
             |       |
             └───────┘

Weighted:    A --5-- B --3-- C   (edges have costs)
```

## 📐 Graph Representation — Pick the Right One

**Adjacency List** (99% of problems): Dictionary mapping each node to its neighbors.

**Adjacency Matrix**: 2D array, `matrix[i][j] = 1` if edge exists. Good for dense graphs.

**Grid as implicit graph**: 2D matrix where each cell connects to its 4 (or 8) neighbors.

````python
from collections import defaultdict, deque

# =====================================
# Adjacency List — The Go-To Format
# =====================================

# From edge list
edges = [[0,1], [0,2], [1,3], [2,3]]

# Undirected graph
graph = defaultdict(list)
for u, v in edges:
    graph[u].append(v)
    graph[v].append(u)  # Remove this line for directed

print("Adjacency List:")
for node in sorted(graph):
    print(f"  {node} → {graph[node]}")

# Grid as graph (implicit)
grid = [
    [1, 1, 0],
    [1, 0, 0],
    [0, 0, 1]
]
# Neighbors of (r, c) = 4-directional
directions = [(0,1), (0,-1), (1,0), (-1,0)]
r, c = 0, 0
neighbors = [(r+dr, c+dc) for dr, dc in directions 
             if 0 <= r+dr < len(grid) and 0 <= c+dc < len(grid[0])]
print(f"\nNeighbors of (0,0) in grid: {neighbors}")
````

## 🎯 BFS vs DFS on Graphs — When to Use Which

| Use BFS When | Use DFS When |
|---|---|
| Shortest path (unweighted) | Explore all paths / backtracking |
| Level-by-level processing | Connected components |
| Min steps / moves | Cycle detection (directed) |
| Closest node | Topological sort |

**Critical difference from trees:** Graphs can have cycles! So you **must track visited nodes**, or you'll loop forever.

````python
# =====================================
#  BFS Template for Graphs
# =====================================
def bfs(graph, start):
    visited = {start}
    queue = deque([start])
    order = []
    
    while queue:
        node = queue.popleft()
        order.append(node)
        for neighbor in graph[node]:
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)
    return order

# =====================================
#  DFS Template for Graphs
# =====================================
def dfs(graph, start):
    visited = set()
    order = []
    
    def explore(node):
        visited.add(node)
        order.append(node)
        for neighbor in graph[node]:
            if neighbor not in visited:
                explore(neighbor)
    
    explore(start)
    return order

print(f"BFS from 0: {bfs(graph, 0)}")   # [0, 1, 2, 3]
print(f"DFS from 0: {dfs(graph, 0)}")   # [0, 1, 3, 2]
````

## 🎯 Problem: Clone Graph (LC #133)

Given a connected undirected graph, return a deep copy. Each node has a value and a list of neighbors.

**Approach:** BFS/DFS + hashmap (old_node → new_node). When visiting a neighbor, either create a new clone or use the existing one.

````python
class Node:
    def __init__(self, val=0, neighbors=None):
        self.val = val
        self.neighbors = neighbors if neighbors else []

def cloneGraph(node):
    """
    BFS + hashmap for cloning.
    Time: O(V + E), Space: O(V)
    """
    if not node:
        return None
    
    clones = {node: Node(node.val)}  # old → new
    queue = deque([node])
    
    while queue:
        curr = queue.popleft()
        for neighbor in curr.neighbors:
            if neighbor not in clones:
                clones[neighbor] = Node(neighbor.val)
                queue.append(neighbor)
            clones[curr].neighbors.append(clones[neighbor])
    
    return clones[node]

# Build test graph: 1 — 2 — 3 — 4 — 1
n1, n2, n3, n4 = Node(1), Node(2), Node(3), Node(4)
n1.neighbors = [n2, n4]
n2.neighbors = [n1, n3]
n3.neighbors = [n2, n4]
n4.neighbors = [n3, n1]

clone = cloneGraph(n1)
print(f"Original node 1 neighbors: {[n.val for n in n1.neighbors]}")
print(f"Cloned node 1 neighbors:   {[n.val for n in clone.neighbors]}")
print(f"Same object? {clone is n1}")  # False (deep copy)
````

## 🗺️ Graph Fundamentals Pattern Map

```
╔══════════════════════════════════════════════════════════════════════╗
║  GRAPH FUNDAMENTALS                                                 ║
╠══════════════════════════════════════════════════════════════════════╣
║                                                                      ║
║  REPRESENTATION: Adjacency list (defaultdict(list)) for 99% cases  ║
║  GRID: Implicit graph, neighbors = 4 directions with bounds check  ║
║                                                                      ║
║  BFS: Queue + visited set. Use for shortest path (unweighted).     ║
║  DFS: Recursion + visited set. Use for components, paths, cycles.  ║
║                                                                      ║
║  ALWAYS: Track visited to avoid infinite loops in cycles!           ║
║  TIME: O(V + E) for both BFS and DFS                               ║
║                                                                      ║
╚══════════════════════════════════════════════════════════════════════╝
```
