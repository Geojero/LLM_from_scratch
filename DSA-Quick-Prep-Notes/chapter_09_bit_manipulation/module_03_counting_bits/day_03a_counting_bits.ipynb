## The Counting Pattern

"How many 1-bits?" appears everywhere. Two key tools:
- **Brian Kernighan's:** `n &= (n-1)` removes rightmost set bit → count iterations
- **DP relation:** `bits[i] = bits[i >> 1] + (i & 1)` — take parent's count and add the last bit

---

## 🎯 Problem 1: Counting Bits (LC #338)

For every number from 0 to n, count the number of 1s in its binary. Return as array.

**DP approach:** `i` in binary = `i >> 1` (drop last bit) + `i & 1` (what was that last bit?).

````python
def countBits(n):
    """
    DP: bits[i] = bits[i >> 1] + (i & 1)
    Time: O(n), Space: O(n)
    """
    dp = [0] * (n + 1)
    for i in range(1, n + 1):
        dp[i] = dp[i >> 1] + (i & 1)
    return dp

print(countBits(5))  # [0, 1, 1, 2, 1, 2]
````

**Why this works:**
```
5 = 101 → bits[5] = bits[2] + 1 = 1 + 1 = 2  ✓
4 = 100 → bits[4] = bits[2] + 0 = 1 + 0 = 1  ✓
3 = 011 → bits[3] = bits[1] + 1 = 1 + 1 = 2  ✓
```

## 🎯 Problem 2: Power of Two (LC #231)

Check if `n` is a power of 2.

**Insight:** Powers of 2 have exactly ONE set bit: `1, 10, 100, 1000...`  
`n & (n-1)` clears that one bit → result is 0.

````python
def isPowerOfTwo(n):
    """
    One set bit means n & (n-1) == 0.
    Time: O(1), Space: O(1)
    """
    return n > 0 and (n & (n - 1)) == 0

print(isPowerOfTwo(16))  # True
print(isPowerOfTwo(18))  # False
print(isPowerOfTwo(1))   # True
````

## 🎯 Problem 3: Hamming Distance (LC #461)

Number of bit positions where two integers differ = count set bits of `x ^ y`.

````python
def hammingDistance(x, y):
    """
    XOR gives differing bits, then count set bits.
    Time: O(1), Space: O(1)
    """
    xor = x ^ y
    count = 0
    while xor:
        xor &= (xor - 1)  # Kernighan's
        count += 1
    return count

print(hammingDistance(1, 4))  # 2 (001 vs 100)
````
