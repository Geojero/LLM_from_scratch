## Why Bits Matter

Every number in a computer is already stored as bits. Bit manipulation lets you do in O(1) what loops do in O(n). In interviews, it shows up as a "clever trick" — once you know the tricks, these problems are free points.

## 🔑 Essential Bit Tricks

These come up again and again. Memorize them.

````python
n = 12  # 1100 in binary

# 1. Check if power of 2
print(f"Is {n} power of 2? {n & (n-1) == 0}")  # False
print(f"Is 8 power of 2? {8 & 7 == 0}")          # True (1000 & 0111 = 0)

# 2. Get / clear rightmost set bit
print(f"Rightmost set bit of {n}: {n & (-n)}")    # 4 (0100)
print(f"Clear rightmost set bit: {n & (n-1)}")    # 8 (1000)

# 3. Check / set / toggle bit at position i
i = 1
print(f"Bit {i} of {n} is: {(n >> i) & 1}")       # 0
print(f"Set bit {i}: {n | (1 << i)}")              # 14 (1110)
print(f"Toggle bit {i}: {n ^ (1 << i)}")           # 14 (1110)

# 4. Count set bits (Brian Kernighan's)
count, x = 0, n
while x:
    x &= (x - 1)  # Remove rightmost set bit
    count += 1
print(f"Set bits in {n}: {count}")                 # 2
````

## 🎯 Problem 1: Number of 1 Bits (LC #191)

Count set bits in an integer.

````python
def hammingWeight(n):
    """
    Brian Kernighan's trick: n & (n-1) clears rightmost set bit.
    Time: O(number of set bits), Space: O(1)
    """
    count = 0
    while n:
        n &= (n - 1)
        count += 1
    return count

print(hammingWeight(0b1011))  # 3
print(hammingWeight(0b10000000))  # 1
````

## 🎯 Problem 2: Reverse Bits (LC #190)

Reverse all 32 bits of a given integer.

````python
def reverseBits(n):
    """
    Pull rightmost bit of n, push to result from left.
    Time: O(32) = O(1), Space: O(1)
    """
    result = 0
    for _ in range(32):
        result = (result << 1) | (n & 1)
        n >>= 1
    return result

print(reverseBits(0b00000010100101000001111010011100))  
# 964176192 = 0b00111001011110000010100101000000
````

## 🗺️ Bit Fundamentals Pattern Map

```
╔══════════════════════════════════════════════════════════════════════╗
║  BIT TRICK CHEAT SHEET                                             ║
╠══════════════════════════════════════════════════════════════════════╣
║  n & (n-1)    → clear rightmost set bit / power-of-2 check       ║
║  n & (-n)     → isolate rightmost set bit                         ║
║  n ^ n = 0    → XOR cancellation (pairs vanish)                   ║
║  n ^ 0 = n    → XOR identity                                      ║
║  (n >> i) & 1 → check i-th bit                                    ║
║  n | (1 << i) → set i-th bit                                      ║
║  n ^ (1 << i) → toggle i-th bit                                   ║
╚══════════════════════════════════════════════════════════════════════╝
```
