## XOR — The Duplicate Killer

XOR has two magical properties:
- `a ^ a = 0` (same values cancel)
- `a ^ 0 = a` (identity)

So if you XOR all elements in an array where every number appears twice except one, all pairs cancel and you're left with the unique one. **No extra space. One pass.**

---

## 🎯 Problem 1: Single Number (LC #136)

Every element appears twice except one. Find it.

````python
def singleNumber(nums):
    """
    XOR all — pairs cancel, unique survives.
    Time: O(n), Space: O(1)
    """
    result = 0
    for n in nums:
        result ^= n
    return result

print(singleNumber([4,1,2,1,2]))  # 4
````

---

## 🎯 Problem 2: Single Number III (LC #260)

Two numbers appear once, rest appear twice. Find both.

**Trick:** XOR all gives `a ^ b`. Since a ≠ b, at least one bit differs. Use that differing bit to split all numbers into two groups — each group contains exactly one unique number.

````python
def singleNumberIII(nums):
    """
    XOR all → a^b. Split by any differing bit.
    Time: O(n), Space: O(1)
    """
    xor_all = 0
    for n in nums:
        xor_all ^= n  # = a ^ b
    
    # Get rightmost set bit (where a and b differ)
    diff_bit = xor_all & (-xor_all)
    
    a = 0
    for n in nums:
        if n & diff_bit:  # Only numbers with this bit set
            a ^= n
    
    b = xor_all ^ a  # The other unique number
    return [a, b]

print(singleNumberIII([1,2,1,3,2,5]))  # [3,5] or [5,3]
````

## 🎯 Problem 3: Missing Number (LC #268)

Array of `n` numbers from `0..n`, one missing. Find it.

**XOR approach:** XOR `0..n` with all array elements. Everything cancels except the missing one.

````python
def missingNumber(nums):
    """
    XOR indices 0..n with all values.
    Time: O(n), Space: O(1)
    """
    result = len(nums)  # Start with n (the last index)
    for i, num in enumerate(nums):
        result ^= i ^ num
    return result

print(missingNumber([3,0,1]))    # 2
print(missingNumber([0,1]))      # 2
````
