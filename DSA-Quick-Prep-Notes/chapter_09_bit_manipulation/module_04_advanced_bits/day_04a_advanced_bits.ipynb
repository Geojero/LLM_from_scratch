## Beyond the Basics

Advanced bit manipulation uses bitmasks to represent **subsets**. A number with `n` bits can represent any subset of `n` items. Bit `i` is set = item `i` is in the subset. This unlocks fast subset enumeration and bitmask DP.

---

## 🔑 Subset Enumeration with Bits

For a set of `n` elements, iterate all `2^n` subsets:

````python
# Enumerate all subsets of [a, b, c]
items = ['a', 'b', 'c']
n = len(items)

for mask in range(1 << n):  # 0 to 2^n - 1
    subset = [items[i] for i in range(n) if mask & (1 << i)]
    print(f"{mask:03b} → {subset}")
````

## 🎯 Problem 1: Subsets (LC #78) — Bit Approach

Generate all subsets using bitmasks instead of backtracking.

````python
def subsets(nums):
    """
    Each number 0..2^n-1 represents a subset.
    Time: O(n * 2^n), Space: O(n * 2^n)
    """
    n = len(nums)
    result = []
    for mask in range(1 << n):
        result.append([nums[i] for i in range(n) if mask & (1 << i)])
    return result

print(subsets([1,2,3]))
````

## 🎯 Problem 2: Sum of All Subset XOR Totals (LC #1863)

For each subset, XOR all elements. Return sum of all these XOR totals.

**Bit observation:** Each bit position contributes independently. Bit `i` is set in exactly half of all subsets → contribution = `bit_i * 2^(n-1)`.

````python
def subsetXORSum(nums):
    """
    OR all nums, multiply by 2^(n-1).
    Time: O(n), Space: O(1)
    """
    # OR collects all bits that appear in any number
    or_all = 0
    for num in nums:
        or_all |= num
    
    # Each bit appears in exactly half of all subsets
    return or_all * (1 << (len(nums) - 1))

print(subsetXORSum([1,3]))      # 6:  {1}→1, {3}→3, {1,3}→2 → total 6
print(subsetXORSum([5,1,6]))    # 28
````

## 🎯 Problem 3: Bitwise AND of Numbers Range (LC #201)

AND all numbers from `left` to `right`.

**Key insight:** AND can only keep bits that are common to ALL numbers. As you go from `left` to `right`, lower bits will flip. The answer is the **common prefix** of `left` and `right` in binary.

````python
def rangeBitwiseAnd(left, right):
    """
    Find common prefix: shift both right until equal.
    Time: O(log n), Space: O(1)
    """
    shift = 0
    while left != right:
        left >>= 1
        right >>= 1
        shift += 1
    return left << shift

print(rangeBitwiseAnd(5, 7))    # 4  (101 & 110 & 111 = 100)
print(rangeBitwiseAnd(1, 2147483647))  # 0
````
