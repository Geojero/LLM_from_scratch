## The Three DFS Traversals

The names tell you **when** you process the current node:

```
         1
        / \
       2   3
      / \
     4   5

  Preorder  (root FIRST):  1, 2, 4, 5, 3    →  root → left → right
  Inorder   (root MIDDLE): 4, 2, 5, 1, 3    →  left → root → right
  Postorder (root LAST):   4, 5, 2, 3, 1    →  left → right → root
```

**When to use which:**
- **Preorder:** Copy/serialize a tree, top-down processing
- **Inorder:** BST → sorted order (this is huge!)
- **Postorder:** Delete a tree, compute bottom-up properties

---

````python
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

from collections import deque

def build_tree(values):
    if not values:
        return None
    root = TreeNode(values[0])
    queue = deque([root])
    i = 1
    while queue and i < len(values):
        node = queue.popleft()
        if i < len(values) and values[i] is not None:
            node.left = TreeNode(values[i])
            queue.append(node.left)
        i += 1
        if i < len(values) and values[i] is not None:
            node.right = TreeNode(values[i])
            queue.append(node.right)
        i += 1
    return root
````

## 🎯 Recursive Traversals — Simple but Important

These are one-liners in concept, but you must know them cold.

````python
def preorder(root):
    if not root:
        return []
    return [root.val] + preorder(root.left) + preorder(root.right)

def inorder(root):
    if not root:
        return []
    return inorder(root.left) + [root.val] + inorder(root.right)

def postorder(root):
    if not root:
        return []
    return postorder(root.left) + postorder(root.right) + [root.val]

tree = build_tree([1, 2, 3, 4, 5])
print(f"Preorder:  {preorder(tree)}")   # [1, 2, 4, 5, 3]
print(f"Inorder:   {inorder(tree)}")    # [4, 2, 5, 1, 3]
print(f"Postorder: {postorder(tree)}")  # [4, 5, 2, 3, 1]
````

## 🎯 Iterative Traversals — Interview Favorites

Interviewers love asking "can you do it without recursion?" Here's the iterative version of each.

### Iterative Preorder (LC #144)
Use a stack. Push root, then in the loop: pop, process, push RIGHT first (so left is processed first).

````python
def preorderIterative(root):
    """Stack: pop → process → push right → push left."""
    if not root:
        return []
    result = []
    stack = [root]
    
    while stack:
        node = stack.pop()
        result.append(node.val)
        if node.right:  # Push right first!
            stack.append(node.right)
        if node.left:
            stack.append(node.left)
    
    return result

tree = build_tree([1, 2, 3, 4, 5])
print(f"Iterative preorder: {preorderIterative(tree)}")  # [1, 2, 4, 5, 3]
````

### Iterative Inorder (LC #94)

This is the trickiest iterative traversal. The idea: go as far left as possible, then process, then go right.

**Visual:** Imagine sliding your finger down the left edge of the tree. When you hit bottom, process that node, then jump to the right child and slide left again.

````python
def inorderIterative(root):
    """Go left as far as possible, process, then go right."""
    result = []
    stack = []
    curr = root
    
    while curr or stack:
        # Go all the way left
        while curr:
            stack.append(curr)
            curr = curr.left
        
        # Process the leftmost unprocessed node
        curr = stack.pop()
        result.append(curr.val)
        
        # Move to right subtree
        curr = curr.right
    
    return result

tree = build_tree([1, 2, 3, 4, 5])
print(f"Iterative inorder: {inorderIterative(tree)}")  # [4, 2, 5, 1, 3]
````

## 🎯 Level Order Traversal — BFS (LC #102)

Process the tree level by level using a queue. This is the tree equivalent of BFS.

```
         3
        / \
       9  20          Level 0: [3]
         / \          Level 1: [9, 20]
        15  7         Level 2: [15, 7]
```

**The trick:** At each step, record the queue size. That tells you how many nodes are in the current level.

````python
def levelOrder(root):
    """
    BFS with level tracking.
    Time: O(n), Space: O(n)
    """
    if not root:
        return []
    
    result = []
    queue = deque([root])
    
    while queue:
        level_size = len(queue)  # Nodes in current level
        level = []
        
        for _ in range(level_size):
            node = queue.popleft()
            level.append(node.val)
            if node.left:
                queue.append(node.left)
            if node.right:
                queue.append(node.right)
        
        result.append(level)
    
    return result

tree = build_tree([3, 9, 20, None, None, 15, 7])
print(levelOrder(tree))  # [[3], [9, 20], [15, 7]]
````

## 🎯 Zigzag Level Order (LC #103)

Same as level order, but alternate left-to-right and right-to-left.

**Simple trick:** Just reverse every other level.

````python
def zigzagLevelOrder(root):
    if not root:
        return []
    
    result = []
    queue = deque([root])
    left_to_right = True
    
    while queue:
        level = []
        for _ in range(len(queue)):
            node = queue.popleft()
            level.append(node.val)
            if node.left:
                queue.append(node.left)
            if node.right:
                queue.append(node.right)
        
        if not left_to_right:
            level.reverse()
        result.append(level)
        left_to_right = not left_to_right
    
    return result

tree = build_tree([3, 9, 20, None, None, 15, 7])
print(zigzagLevelOrder(tree))  # [[3], [20, 9], [15, 7]]
````

## 🗺️ Traversal Pattern Map

```
╔══════════════════════════════════════════════════════════════════════╗
║  TREE TRAVERSAL PATTERNS                                            ║
╠══════════════════════════════════════════════════════════════════════╣
║                                                                      ║
║  DFS (Stack/Recursion):                                              ║
║  • Preorder:  process → left → right   [copy/serialize]             ║
║  • Inorder:   left → process → right   [BST sorted order]          ║
║  • Postorder: left → right → process   [delete/bottom-up]          ║
║                                                                      ║
║  BFS (Queue):                                                        ║
║  • Level order: process level by level                              ║
║  • Trick: len(queue) = current level size                           ║
║  • Zigzag: reverse every other level                                ║
║                                                                      ║
║  ITERATIVE INORDER: the tricky one                                   ║
║    curr + stack: go left → pop & process → go right                 ║
║                                                                      ║
╚══════════════════════════════════════════════════════════════════════╝
```
