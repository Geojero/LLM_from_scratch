## 🎯 Problem 1: Climbing Stairs (LeetCode #70)

```
┌─────────────────────────────────────────────────────────────────┐
│ You are climbing a staircase. It takes n steps to reach top.   │
│ Each time you can climb 1 or 2 steps.                          │
│ In how many distinct ways can you climb to the top?            │
│                                                                 │
│ Example: n = 3 → Output: 3                                     │
│ Ways: [1,1,1], [1,2], [2,1]                                    │
└─────────────────────────────────────────────────────────────────┘
```

### 🔍 UMPIRE Walkthrough

**U - UNDERSTAND:**
- At each step, can take 1 or 2 steps
- Need to count ALL distinct ways
- This is exactly like Fibonacci!

**M - MATCH:**
- DP pattern: counting ways
- Similar to Fibonacci sequence

**P - PLAN:**
```
State: dp[i] = number of ways to reach step i

Recurrence: To reach step i, I came from:
  - Step i-1 (took 1 step), OR
  - Step i-2 (took 2 steps)
  
So: dp[i] = dp[i-1] + dp[i-2]

Base cases:
  dp[1] = 1 (one way: take 1 step)
  dp[2] = 2 (two ways: 1+1 or 2)
```

**I - IMPLEMENT:**

````python
def climbStairs(n):
    """
    LeetCode 70: Climbing Stairs
    Time: O(n), Space: O(1)
    """
    if n <= 2:
        return n
    
    # Only need last 2 values (space optimization)
    prev2of = 1  # ways to reach step 1
    prev1 = 2  # ways to reach step 2
    
    for i in range(3, n + 1):
        current = prev1 + prev2
        prev2 = prev1
        prev1 = current
    
    return prev1

# Test cases
print("n=2:", climbStairs(2))  # 2
print("n=3:", climbStairs(3))  # 3
print("n=4:", climbStairs(4))  # 5
print("n=5:", climbStairs(5))  # 8
````

````python
# Alternative: Top-Down with Memoization
def climbStairs_memo(n, memo={}):
    """
    Top-down approach.
    Time: O(n), Space: O(n)
    """
    if n in memo:
        return memo[n]
    
    if n <= 2:
        return n
    
    memo[n] = climbStairs_memo(n-1, memo) + climbStairs_memo(n-2, memo)
    return memo[n]

print("Memo n=10:", climbStairs_memo(10))  # 89
````

## 🎯 Problem 2: Min Cost Climbing Stairs (LeetCode #746)

```
┌─────────────────────────────────────────────────────────────────┐
│ Given an array of costs where cost[i] is the cost of step i.   │
│ You can start from step 0 or step 1.                           │
│ Once you pay the cost, you can climb 1 or 2 steps.             │
│ Return the minimum cost to reach the top of the floor.         │
│                                                                 │
│ Example: cost = [10, 15, 20] → Output: 15                      │
│ (Start at index 1, pay 15, climb 2 steps to top)               │
└─────────────────────────────────────────────────────────────────┘
```

### 🔍 UMPIRE Walkthrough

**U - UNDERSTAND:**
- Each step has a cost
- Can start from step 0 or 1
- "Top" is past the last step
- Minimize total cost

**M - MATCH:**
- DP pattern: optimization (minimize)
- Similar to climbing stairs but with costs

**P - PLAN:**
```
State: dp[i] = minimum cost to reach step i

Recurrence: To reach step i:
  - Come from i-1: dp[i-1] + cost[i-1]
  - Come from i-2: dp[i-2] + cost[i-2]
  Take the minimum!

dp[i] = min(dp[i-1] + cost[i-1], dp[i-2] + cost[i-2])

Base cases:
  dp[0] = 0 (can start here free)
  dp[1] = 0 (can start here free)
```

**I - IMPLEMENT:**

````python
def minCostClimbingStairs(cost):
    """
    LeetCode 746: Min Cost Climbing Stairs
    Time: O(n), Space: O(1)
    """
    n = len(cost)
    
    # dp[i] = min cost to reach step i
    # We want to reach step n (top, past the last step)
    
    prev2 = 0  # cost to reach step 0
    prev1 = 0  # cost to reach step 1
    
    for i in range(2, n + 1):
        current = min(
            prev1 + cost[i-1],  # came from step i-1
            prev2 + cost[i-2]   # came from step i-2
        )
        prev2 = prev1
        prev1 = current
    
    return prev1

# Test cases
print("[10,15,20]:", minCostClimbingStairs([10, 15, 20]))  # 15
print("[1,100,1,1,1,100,1,1,100,1]:", 
      minCostClimbingStairs([1, 100, 1, 1, 1, 100, 1, 1, 100, 1]))  # 6
````

### 📊 Visual Trace

```
cost = [10, 15, 20]

Step 0: cost=10    Step 1: cost=15    Step 2: cost=20    TOP
   |                  |                  |                |
   0 ─────────────────1─────────────────2────────────────3
   
dp[0] = 0 (start free)
dp[1] = 0 (start free)
dp[2] = min(dp[1] + cost[1], dp[0] + cost[0])
      = min(0 + 15, 0 + 10) = 10
dp[3] = min(dp[2] + cost[2], dp[1] + cost[1])
      = min(10 + 20, 0 + 15) = 15 ✓
```

## 🎯 Problem 3: N-th Tribonacci Number (LeetCode #1137)

```
┌─────────────────────────────────────────────────────────────────┐
│ Tribonacci sequence:                                           │
│ T(0) = 0, T(1) = 1, T(2) = 1                                  │
│ T(n) = T(n-1) + T(n-2) + T(n-3) for n >= 3                    │
│                                                                 │
│ Return T(n).                                                   │
│                                                                 │
│ Example: n = 4 → Output: 4                                     │
│ T(3) = 0+1+1 = 2, T(4) = 1+1+2 = 4                            │
└─────────────────────────────────────────────────────────────────┘
```

### 🔍 UMPIRE Walkthrough

**P - PLAN:**
```
State: dp[i] = T(i)
Recurrence: dp[i] = dp[i-1] + dp[i-2] + dp[i-3]
Base cases: dp[0]=0, dp[1]=1, dp[2]=1

Space optimization: only keep last 3 values!
```

**I - IMPLEMENT:**

````python
def tribonacci(n):
    """
    LeetCode 1137: N-th Tribonacci Number
    Time: O(n), Space: O(1)
    """
    if n == 0:
        return 0
    if n <= 2:
        return 1
    
    # Keep track of last 3 values
    t0, t1, t2 = 0, 1, 1
    
    for i in range(3, n + 1):
        current = t0 + t1 + t2
        t0, t1, t2 = t1, t2, current
    
    return t2

# Test cases
print("T(0):", tribonacci(0))  # 0
print("T(1):", tribonacci(1))  # 1
print("T(4):", tribonacci(4))  # 4
print("T(25):", tribonacci(25))  # 1389537
````

## 🗺️ DP Fundamentals Pattern Map

```
╔═══════════════════════════════════════════════════════════════════════╗
║  DP FUNDAMENTALS PATTERNS                                            ║
╠═══════════════════════════════════════════════════════════════════════╣
║                                                                       ║
║  1. FIBONACCI-STYLE: dp[i] depends on dp[i-1], dp[i-2]             ║
║     → Climbing Stairs, Tribonacci, Decode Ways                       ║
║                                                                       ║
║  2. MIN/MAX COST: dp[i] = min/max(dp[i-1], dp[i-2]) + cost[i]     ║
║     → Min Cost Climbing Stairs, House Robber                         ║
║                                                                       ║
║  3. COUNT WAYS: dp[i] = dp[i-1] + dp[i-2] + ...                    ║
║     → Climbing Stairs, Decode Ways                                    ║
║                                                                       ║
║  SPACE OPTIMIZATION: If dp[i] only needs dp[i-1] and dp[i-2],      ║
║  keep two variables instead of full array → O(1) space              ║
║                                                                       ║
╚═══════════════════════════════════════════════════════════════════════╝
```
