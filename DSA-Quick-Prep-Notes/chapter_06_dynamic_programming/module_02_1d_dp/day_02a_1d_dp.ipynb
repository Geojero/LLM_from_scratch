## 1D DP = dp[i] depends on previous values

The simplest form of DP. You have a single dimension (usually an index or amount), and each state builds from smaller states.

**The mental model:** You're standing at step `i`. To figure out the answer at `i`, you look back at `i-1`, `i-2`, etc.

```
dp[i] = some function of dp[i-1], dp[i-2], ...

Example (Fibonacci):
dp[0] = 0, dp[1] = 1
dp[i] = dp[i-1] + dp[i-2]
```

**Three steps for any DP problem:**
1. **Define the state:** What does `dp[i]` represent?
2. **Transition:** How does `dp[i]` relate to previous states?
3. **Base case:** What are the starting values?

## 🎯 Problem 1: Climbing Stairs (LC #70)

You can climb 1 or 2 steps. How many distinct ways to reach step `n`?

**State:** `dp[i]` = number of ways to reach step `i`.

**Transition:** To reach step `i`, you came from step `i-1` (took 1 step) or step `i-2` (took 2 steps).

```
dp[i] = dp[i-1] + dp[i-2]   ← That's literally Fibonacci!
```

````python
def climbStairs(n):
    """
    Space-optimized 1D DP. Only need last 2 values.
    Time: O(n), Space: O(1)
    """
    if n <= 2:
        return n
    prev2, prev1 = 1, 2  # dp[1]=1, dp[2]=2
    for i in range(3, n + 1):
        curr = prev1 + prev2
        prev2, prev1 = prev1, curr
    return prev1

print(climbStairs(2))   # 2
print(climbStairs(3))   # 3
print(climbStairs(5))   # 8
````

**Space optimization trick:** If `dp[i]` only depends on the last 2 values, you don't need the full array. Just keep two variables. This works for most 1D DP problems.

## 🎯 Problem 2: House Robber (LC #198)

Can't rob two adjacent houses. Maximize total money.

**State:** `dp[i]` = max money robbing from houses `0...i`.

**Transition:** At house `i`, you either:
- **Rob it:** Take `nums[i]` + best from `dp[i-2]` (skip adjacent)
- **Skip it:** Take `dp[i-1]` (best without house i)

```
dp[i] = max(dp[i-1], dp[i-2] + nums[i])
```

````python
def rob(nums):
    """
    At each house: rob it + dp[i-2], or skip it (dp[i-1]).
    Time: O(n), Space: O(1)
    """
    if not nums:
        return 0
    prev2, prev1 = 0, 0  # dp[i-2], dp[i-1]
    for num in nums:
        curr = max(prev1, prev2 + num)
        prev2, prev1 = prev1, curr
    return prev1

print(rob([1, 2, 3, 1]))     # 4 (rob house 0 + house 2)
print(rob([2, 7, 9, 3, 1]))  # 12 (rob house 0 + house 2 + house 4)
````

## 🎯 Problem 3: Coin Change (LC #322)

Given coins of different denominations, find FEWEST coins to make `amount`. Return -1 if impossible.

**State:** `dp[a]` = minimum coins needed for amount `a`.

**Transition:** For each coin `c`, if we use it: `dp[a] = dp[a - c] + 1`.

Try all coins, take the minimum.

```
dp[a] = min(dp[a - c] + 1) for all valid coins c
```

````python
def coinChange(coins, amount):
    """
    Bottom-up DP. Build from amount 0 to target.
    Time: O(amount * len(coins)), Space: O(amount)
    """
    dp = [float('inf')] * (amount + 1)
    dp[0] = 0  # Base: 0 coins to make amount 0
    
    for a in range(1, amount + 1):
        for coin in coins:
            if coin <= a:
                dp[a] = min(dp[a], dp[a - coin] + 1)
    
    return dp[amount] if dp[amount] != float('inf') else -1

print(coinChange([1, 5, 10, 25], 30))  # 2 (25 + 5)
print(coinChange([2], 3))               # -1 (impossible)
````

## 🎯 Problem 4: Decode Ways (LC #91)

A message "12" can decode as "AB" (1,2) or "L" (12). Count total decode ways.

**State:** `dp[i]` = number of ways to decode `s[0:i]`.

**Transition:**
- If `s[i-1]` is '1'-'9': single digit valid → `dp[i] += dp[i-1]`
- If `s[i-2:i]` is '10'-'26': two digits valid → `dp[i] += dp[i-2]`

````python
def numDecodings(s):
    """
    1D DP with two-digit lookahead.
    Time: O(n), Space: O(1)
    """
    if not s or s[0] == '0':
        return 0
    
    prev2, prev1 = 1, 1  # dp[-1]=1 (empty), dp[0]=1 (first char valid)
    
    for i in range(1, len(s)):
        curr = 0
        # Single digit
        if s[i] != '0':
            curr += prev1
        # Two digits
        two_digit = int(s[i-1:i+1])
        if 10 <= two_digit <= 26:
            curr += prev2
        prev2, prev1 = prev1, curr
    
    return prev1

print(numDecodings("12"))   # 2  ("AB" or "L")
print(numDecodings("226"))  # 3  ("BZ", "VF", "BBF")
print(numDecodings("06"))   # 0  (leading zero invalid)
````

## 🎯 Problem 5: Jump Game (LC #55)

Can you reach the last index? `nums[i]` = max jump length from position `i`.

**Greedy insight (simpler than DP):** Track the farthest reachable position. If at any point `i > farthest`, you're stuck.

````python
def canJump(nums):
    """
    Greedy: track max reachable index.
    Time: O(n), Space: O(1)
    """
    max_reach = 0
    for i in range(len(nums)):
        if i > max_reach:
            return False  # Can't reach position i
        max_reach = max(max_reach, i + nums[i])
    return True

print(canJump([2, 3, 1, 1, 4]))  # True
print(canJump([3, 2, 1, 0, 4]))  # False (stuck at index 3)
````

## 🗺️ 1D DP Pattern Map

```
╔══════════════════════════════════════════════════════════════════════╗
║  1D DP PATTERNS                                                     ║
╠══════════════════════════════════════════════════════════════════════╣
║                                                                      ║
║  FIBONACCI-LIKE:  dp[i] = dp[i-1] + dp[i-2]                       ║
║    → Climbing Stairs, Decode Ways, House Robber                    ║
║                                                                      ║
║  MIN/MAX OVER CHOICES:  dp[i] = min/max(options)                   ║
║    → Coin Change, Jump Game II                                      ║
║                                                                      ║
║  SPACE OPTIMIZATION: If dp[i] only needs last K values,            ║
║    use K variables instead of full array.                           ║
║                                                                      ║
║  3 STEPS: 1. Define state  2. Transition  3. Base case             ║
║                                                                      ║
╚══════════════════════════════════════════════════════════════════════╝
```
