## What Is Interval DP?

DP on a **range [i, j]** where you try all possible split points `k` between `i` and `j`.

```
dp[i][j] = best of trying all splits:
    for k in range(i, j):
        dp[i][j] = optimize(dp[i][k] + dp[k+1][j] + cost)
```

**The mental model:** Think of merging adjacent groups. You're trying every way to split the range into two parts, solving each part, and combining.

**Iteration order matters:** Process by **increasing length** (small ranges first, then build up to full range).

## 🎯 Problem 1: Burst Balloons (LC #312)

Burst all balloons to maximize coins. Bursting balloon `i` gives `nums[left] * nums[i] * nums[right]`.

**Key insight:** Think in REVERSE. Instead of "which to burst first", think "which to burst LAST in range [i,j]".

If balloon `k` is the LAST burst in range `[i,j]`, then:
- Left subproblem `[i, k-1]` is already solved
- Right subproblem `[k+1, j]` is already solved
- Bursting `k` last means its neighbors are `nums[i-1]` and `nums[j+1]` (boundaries!)

```
dp[i][j] = max over k of:
    dp[i][k-1] + nums[i-1]*nums[k]*nums[j+1] + dp[k+1][j]
```

````python
def maxCoins(nums):
    """
    Interval DP: which balloon to burst LAST in each range.
    Time: O(n³), Space: O(n²)
    """
    # Add virtual balloons at boundaries
    nums = [1] + nums + [1]
    n = len(nums)
    dp = [[0] * n for _ in range(n)]
    
    # Process by increasing range length
    for length in range(1, n - 1):  # length of range (excluding boundaries)
        for i in range(1, n - length):  # left boundary
            j = i + length - 1           # right boundary
            for k in range(i, j + 1):    # which to burst last
                coins = nums[i-1] * nums[k] * nums[j+1]
                coins += dp[i][k-1] + dp[k+1][j]
                dp[i][j] = max(dp[i][j], coins)
    
    return dp[1][n-2]  # Full range (excluding virtual boundaries)

print(maxCoins([3, 1, 5, 8]))  # 167
````

## 🎯 Problem 2: Stone Game (LC #877)

Two players pick stones from either end. Determine if player 1 wins (gets more stones).

**State:** `dp[i][j]` = maximum score difference (current player - opponent) for piles[i:j+1].

**Transition:** Current player picks left (`piles[i]`) or right (`piles[j]`). After picking, the opponent plays optimally (so we subtract their advantage).

```
dp[i][j] = max(
    piles[i] - dp[i+1][j],   # Pick left, opponent plays [i+1..j]
    piles[j] - dp[i][j-1]    # Pick right, opponent plays [i..j-1]
)
```

````python
def stoneGame(piles):
    """
    Interval DP: score difference from current player's perspective.
    Time: O(n²), Space: O(n²)
    """
    n = len(piles)
    dp = [[0] * n for _ in range(n)]
    
    # Base: single pile → take it
    for i in range(n):
        dp[i][i] = piles[i]
    
    # Fill by increasing length
    for length in range(2, n + 1):
        for i in range(n - length + 1):
            j = i + length - 1
            dp[i][j] = max(
                piles[i] - dp[i+1][j],
                piles[j] - dp[i][j-1]
            )
    
    return dp[0][n-1] > 0  # Player 1 wins if positive

print(stoneGame([5, 3, 4, 5]))  # True
# (Fun fact: Player 1 ALWAYS wins with even piles — math proof exists!)
````

## 🗺️ Interval DP Pattern Map

```
╔══════════════════════════════════════════════════════════════════════╗
║  INTERVAL DP PATTERNS                                               ║
╠══════════════════════════════════════════════════════════════════════╣
║                                                                      ║
║  STRUCTURE:                                                         ║
║    dp[i][j] = answer for range [i, j]                              ║
║    Try all split points k in [i, j]                                ║
║    Combine: dp[i][k] + dp[k+1][j] + merge cost                    ║
║                                                                      ║
║  ITERATION: by increasing length (small → large ranges)            ║
║                                                                      ║
║  GAME THEORY variant:                                               ║
║    dp[i][j] = score difference (me - opponent)                     ║
║    Pick from ends: max(take_left - dp_rest, take_right - dp_rest)  ║
║                                                                      ║
║  TRIGGER: "burst", "merge adjacent", "game from ends"             ║
║                                                                      ║
╚══════════════════════════════════════════════════════════════════════╝
```
