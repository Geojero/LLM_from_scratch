## The Core Idea

Array greedy problems ask: "Can you make it from A to B?" or "What's the optimal assignment?"

The trick is always to think about what the **locally best choice** is at each position, and prove (or trust) that it leads to globally best.

---

## 🎯 Problem 1: Jump Game II (LC #45)

Minimum number of jumps to reach the last index.

**Think of it as BFS on an array.** Each "level" = one jump. At each level, track the farthest you can reach. When you hit the edge of the current level, that's another jump.

````python
def jump(nums):
    """
    BFS-style greedy: track current level end and farthest reachable.
    Time: O(n), Space: O(1)
    """
    jumps = 0
    curr_end = 0      # Edge of current jump level
    farthest = 0      # Farthest reachable from this level
    
    for i in range(len(nums) - 1):  # Don't need to jump FROM last index
        farthest = max(farthest, i + nums[i])
        
        if i == curr_end:        # Reached edge of current level
            jumps += 1
            curr_end = farthest  # Next level edge
    
    return jumps

print(jump([2,3,1,1,4]))    # 2 (index 0→1→4)
print(jump([2,3,0,1,4]))    # 2
````

## 🎯 Problem 2: Gas Station (LC #134)

Circular route with gas stations. Can you complete the circuit?

**Two insights:**
1. If total gas ≥ total cost, a solution MUST exist.
2. If running tank goes negative at station `i`, start from `i+1` — everything before is dead weight.

````python
def canCompleteCircuit(gas, cost):
    """
    Single pass: reset start whenever tank goes negative.
    Time: O(n), Space: O(1)
    """
    if sum(gas) < sum(cost):
        return -1  # Impossible
    
    tank = 0
    start = 0
    
    for i in range(len(gas)):
        tank += gas[i] - cost[i]
        if tank < 0:          # Can't reach next station
            start = i + 1     # Start fresh from next station
            tank = 0
    
    return start

print(canCompleteCircuit([1,2,3,4,5], [3,4,5,1,2]))  # 3
````

## 🎯 Problem 3: Candy (LC #135)

Each child gets at least 1 candy. Higher rating than neighbor → more candy than that neighbor. Minimize total.

**Two-pass greedy:** Left-to-right ensures right neighbor rule. Right-to-left ensures left neighbor rule. Take the max at each position.

````python
def candy(ratings):
    """
    Two-pass: left→right, then right→left.
    Time: O(n), Space: O(n)
    """
    n = len(ratings)
    candies = [1] * n
    
    # Left → Right: higher rating than left neighbor gets more
    for i in range(1, n):
        if ratings[i] > ratings[i-1]:
            candies[i] = candies[i-1] + 1
    
    # Right → Left: higher rating than right neighbor gets more
    for i in range(n-2, -1, -1):
        if ratings[i] > ratings[i+1]:
            candies[i] = max(candies[i], candies[i+1] + 1)
    
    return sum(candies)

print(candy([1,0,2]))      # 5 → [2,1,2]
print(candy([1,2,2]))      # 4 → [1,2,1]
````
