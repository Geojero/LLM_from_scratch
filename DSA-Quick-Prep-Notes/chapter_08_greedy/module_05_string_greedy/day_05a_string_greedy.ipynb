## The Pattern

String greedy problems typically involve **removing or rearranging characters** to get the smallest/best result. The weapon of choice is usually a **monotonic stack** — process characters one by one, and greedily pop from the stack when you can improve the answer.

---

## 🎯 Problem 1: Remove K Digits (LC #402)

Remove `k` digits from number string to make it smallest.

**Greedy via monotonic stack:** Scan left to right. If current digit is smaller than stack top, pop the top (that's a removal). We want the leftmost digits to be as small as possible.

````python
def removeKdigits(num, k):
    """
    Monotonic increasing stack.
    Time: O(n), Space: O(n)
    """
    stack = []
    
    for digit in num:
        # Pop larger digits from stack while we have removals left
        while k and stack and stack[-1] > digit:
            stack.pop()
            k -= 1
        stack.append(digit)
    
    # If k removals remain, trim from end (stack is non-decreasing)
    if k:
        stack = stack[:-k]
    
    # Strip leading zeros, handle empty
    return ''.join(stack).lstrip('0') or '0'

print(removeKdigits("1432219", 3))  # "1219"
print(removeKdigits("10200", 1))    # "200"
print(removeKdigits("10", 2))       # "0"
````

**Trace for `"1432219", k=3`:**
- `1` → stack=[1]
- `4` → stack=[1,4]
- `3` → 3<4, pop 4 (k=2), stack=[1,3]
- `2` → 2<3, pop 3 (k=1), stack=[1,2]
- `2` → stack=[1,2,2]
- `1` → 1<2, pop 2 (k=0), stack=[1,2,1]
- `9` → stack=[1,2,1,9] → **"1219"** ✓

## 🎯 Problem 2: Remove Duplicate Letters (LC #316)

Remove duplicate letters so each appears once. Result must be **smallest in lexicographic order** among all possible results.

**Same monotonic stack idea** but with two extra constraints:
1. Don't add a char that's already in stack (use a `seen` set)
2. Only pop if the popped char appears later (check `last_occurrence`)

````python
def removeDuplicateLetters(s):
    """
    Monotonic stack + last occurrence + seen set.
    Time: O(n), Space: O(1) — 26 chars max
    """
    last = {c: i for i, c in enumerate(s)}  # Last occurrence of each char
    stack = []
    seen = set()
    
    for i, c in enumerate(s):
        if c in seen:
            continue  # Already placed optimally
        
        # Pop if: stack top is larger AND appears later
        while stack and stack[-1] > c and last[stack[-1]] > i:
            seen.remove(stack.pop())
        
        stack.append(c)
        seen.add(c)
    
    return ''.join(stack)

print(removeDuplicateLetters("bcabc"))    # "abc"
print(removeDuplicateLetters("cbacdcbc")) # "acdb"
````
