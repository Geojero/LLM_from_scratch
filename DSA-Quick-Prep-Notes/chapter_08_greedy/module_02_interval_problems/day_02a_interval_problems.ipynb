## The Interval Toolkit

Interval problems almost always start with: **Sort by start or end time.** Then process linearly.

| Goal | Sort By | Strategy |
|------|---------|----------|
| Merge overlapping | Start time | Extend or start new |
| Max non-overlapping | End time | Greedily pick earliest end |
| Min rooms/resources | Start time | Min-heap for end times |

---

## 🎯 Problem 1: Merge Intervals (LC #56)

Merge all overlapping intervals.

````python
def merge(intervals):
    """
    Sort by start, merge if overlapping.
    Time: O(n log n), Space: O(n)
    """
    intervals.sort()
    merged = [intervals[0]]
    
    for start, end in intervals[1:]:
        if start <= merged[-1][1]:  # Overlaps with last merged
            merged[-1][1] = max(merged[-1][1], end)  # Extend
        else:
            merged.append([start, end])  # New interval
    
    return merged

print(merge([[1,3],[2,6],[8,10],[15,18]]))  # [[1,6],[8,10],[15,18]]
````

## 🎯 Problem 2: Non-overlapping Intervals (LC #435)

Minimum removals to make all intervals non-overlapping.

**Key insight:** Sort by END time. Greedily keep intervals that end earliest (leave most room for future). Count conflicts.

````python
def eraseOverlapIntervals(intervals):
    """
    Sort by end time, greedily keep earliest-ending.
    Time: O(n log n), Space: O(1)
    """
    intervals.sort(key=lambda x: x[1])  # Sort by END
    removals = 0
    prev_end = float('-inf')
    
    for start, end in intervals:
        if start >= prev_end:  # No overlap
            prev_end = end     # Keep this interval
        else:
            removals += 1      # Overlap → remove this one
    
    return removals

print(eraseOverlapIntervals([[1,2],[2,3],[3,4],[1,3]]))  # 1 (remove [1,3])
````

## 🎯 Problem 3: Meeting Rooms II (LC #253)

Minimum number of conference rooms needed.

**Approach: Min-heap of end times.** For each meeting: if earliest room is free (heap top ≤ start), reuse it. Otherwise, allocate new.

````python
import heapq

def minMeetingRooms(intervals):
    """
    Min-heap tracks earliest room freeing up.
    Time: O(n log n), Space: O(n)
    """
    intervals.sort()  # Sort by start
    heap = []  # Min-heap of end times (when each room frees up)
    
    for start, end in intervals:
        if heap and heap[0] <= start:
            heapq.heappop(heap)   # Reuse earliest free room
        heapq.heappush(heap, end)  # Assign room
    
    return len(heap)  # Number of rooms in use

print(minMeetingRooms([[0,30],[5,10],[15,20]]))  # 2
print(minMeetingRooms([[7,10],[2,4]]))            # 1
````
