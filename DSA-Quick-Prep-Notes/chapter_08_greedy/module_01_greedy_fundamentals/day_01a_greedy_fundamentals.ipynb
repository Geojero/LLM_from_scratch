## When Does Greedy Work?

Greedy makes the **locally optimal choice** at each step and never looks back. It works when:

1. **Greedy choice property:** A local optimal leads to a global optimal
2. **Optimal substructure:** Optimal solution contains optimal sub-solutions

**Greedy vs DP:**
- Greedy: Make one choice, move on. No backtracking. O(n log n) typically.
- DP: Consider ALL choices, pick best. O(n²) or worse.

**If greedy works, it's always simpler and faster than DP.**

### How to recognize greedy problems:
- "Maximum/minimum number of..." with a simple selection rule
- Interval scheduling, activity selection
- Can you sort by some criteria and then process linearly?

## 🎯 Problem 1: Assign Cookies (LC #455)

Give cookies to children. Each child has a greed factor, each cookie has a size. Maximize satisfied children.

**Greedy:** Sort both. Give the smallest sufficient cookie to the child with the smallest greed.

````python
def findContentChildren(g, s):
    """
    Sort both, greedily match smallest cookie to least greedy child.
    Time: O(n log n), Space: O(1)
    """
    g.sort()  # Children's greed
    s.sort()  # Cookie sizes
    child = cookie = 0
    while child < len(g) and cookie < len(s):
        if s[cookie] >= g[child]:
            child += 1  # Child satisfied
        cookie += 1     # Try next cookie
    return child

print(findContentChildren([1, 2, 3], [1, 1]))    # 1
print(findContentChildren([1, 2], [1, 2, 3]))    # 2
````

## 🎯 Problem 2: Best Time to Buy and Sell Stock II (LC #122)

Multiple transactions allowed. Maximize profit.

**Greedy insight:** Capture every upward movement. If price goes up tomorrow, buy today and sell tomorrow.

````python
def maxProfit(prices):
    """
    Collect every positive difference.
    Time: O(n), Space: O(1)
    """
    profit = 0
    for i in range(1, len(prices)):
        if prices[i] > prices[i-1]:
            profit += prices[i] - prices[i-1]
    return profit

print(maxProfit([7, 1, 5, 3, 6, 4]))  # 7 (buy@1 sell@5 + buy@3 sell@6)
````
