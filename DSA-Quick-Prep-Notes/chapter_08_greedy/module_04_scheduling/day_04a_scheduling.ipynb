## The Scheduling Mindset

Scheduling problems are about **filling time slots optimally**. The greedy insight is almost always: **handle the most constrained / most frequent item first.** If you have tasks A(3), B(2), C(1) with cooldown, place A first — it has the most copies to space out.

---

## 🎯 Problem 1: Task Scheduler (LC #621)

Tasks with cooldown `n` between same tasks. Minimum intervals to finish all.

**Math approach:** The most frequent task dictates the frame. Imagine placing the most frequent task with gaps:

```
A _ _ A _ _ A      ← max_freq=3, n=2
  gaps = (max_freq - 1) * (n + 1) + count_of_max_freq_tasks
```

If we have more tasks than gaps, answer is just `len(tasks)` (no idle needed).

````python
from collections import Counter

def leastInterval(tasks, n):
    """
    Math formula based on max frequency.
    Time: O(n), Space: O(1) — at most 26 letters
    """
    freq = Counter(tasks)
    max_freq = max(freq.values())
    # How many tasks share the max frequency?
    max_count = sum(1 for v in freq.values() if v == max_freq)
    
    # Frame: (max_freq - 1) chunks of size (n+1), plus the last row
    result = (max_freq - 1) * (n + 1) + max_count
    
    # If total tasks > frame, no idle slots needed
    return max(result, len(tasks))

print(leastInterval(['A','A','A','B','B','B'], 2))  # 8: A B _ A B _ A B
print(leastInterval(['A','A','A','B','B','B'], 0))  # 6: no cooldown
````

**Walkthrough for `[A,A,A,B,B,B], n=2`:**
- max_freq = 3 (A and B both appear 3 times)
- max_count = 2 (both A and B)
- Frame: (3-1) × (2+1) + 2 = 6 + 2 = 8
- Schedule: `A B _ | A B _ | A B` → 8 slots ✓

## 🎯 Problem 2: Reorganize String (LC #767)

Rearrange string so no two adjacent chars are same. Return `""` if impossible.

**Greedy with max-heap:** Always place the most frequent char that isn't the previous one. If most frequent char appears more than `(n+1)//2` times, impossible.

````python
import heapq
from collections import Counter

def reorganizeString(s):
    """
    Max-heap: always pick most frequent that isn't prev.
    Time: O(n log 26) = O(n), Space: O(n)
    """
    freq = Counter(s)
    
    # Impossible check
    if max(freq.values()) > (len(s) + 1) // 2:
        return ""
    
    # Max-heap (negate for Python's min-heap)
    heap = [(-count, char) for char, count in freq.items()]
    heapq.heapify(heap)
    
    result = []
    prev_count, prev_char = 0, ''
    
    while heap:
        count, char = heapq.heappop(heap)
        
        # Push back the previous char if it still has count
        if prev_count < 0:
            heapq.heappush(heap, (prev_count, prev_char))
        
        result.append(char)
        prev_count = count + 1  # Used one (count is negative)
        prev_char = char
    
    return ''.join(result)

print(reorganizeString("aab"))    # "aba"
print(reorganizeString("aaab"))   # "" (impossible)
````
