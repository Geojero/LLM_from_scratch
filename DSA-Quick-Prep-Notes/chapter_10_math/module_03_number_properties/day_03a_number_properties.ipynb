## Digit Manipulation

These problems are about extracting, reversing, or checking properties of digits. The key operations:
- `n % 10` → last digit
- `n // 10` → remove last digit
- Build reversed: `rev = rev * 10 + digit`

---

## 🎯 Problem 1: Palindrome Number (LC #9)

Is integer a palindrome? Solve without converting to string.

**Reverse only the second half** and compare. When reversed half ≥ remaining, we've crossed the middle.

````python
def isPalindrome(x):
    """
    Reverse second half, compare with first half.
    Time: O(log n), Space: O(1)
    """
    if x < 0 or (x % 10 == 0 and x != 0):
        return False
    
    reversed_half = 0
    while x > reversed_half:
        reversed_half = reversed_half * 10 + x % 10
        x //= 10
    
    # Even length: x == reversed_half
    # Odd length: x == reversed_half // 10 (drop middle digit)
    return x == reversed_half or x == reversed_half // 10

print(isPalindrome(121))   # True
print(isPalindrome(1221))  # True
print(isPalindrome(-121))  # False
````

## 🎯 Problem 2: Reverse Integer (LC #7)

Reverse digits. Return 0 if overflow (outside 32-bit signed range).

````python
def reverse(x):
    """
    Pop digits and push to result. Check overflow.
    Time: O(log n), Space: O(1)
    """
    INT_MAX = 2**31 - 1  # 2147483647
    INT_MIN = -(2**31)   # -2147483648
    
    result = 0
    sign = 1 if x >= 0 else -1
    x = abs(x)
    
    while x:
        digit = x % 10
        x //= 10
        result = result * 10 + digit
    
    result *= sign
    return result if INT_MIN <= result <= INT_MAX else 0

print(reverse(123))     # 321
print(reverse(-123))    # -321
print(reverse(1534236469))  # 0 (overflow)
````

## 🎯 Problem 3: Perfect Squares (LC #279)

Minimum number of perfect squares that sum to `n`.

**BFS approach:** Each level = adding one more square. First time we hit 0 = minimum count.

````python
from collections import deque

def numSquares(n):
    """
    BFS: each level tries subtracting a perfect square.
    Time: O(n * sqrt(n)), Space: O(n)
    """
    squares = [i*i for i in range(1, int(n**0.5)+1)]
    queue = deque([(n, 0)])  # (remaining, steps)
    visited = {n}
    
    while queue:
        remain, steps = queue.popleft()
        for sq in squares:
            nxt = remain - sq
            if nxt == 0:
                return steps + 1
            if nxt > 0 and nxt not in visited:
                visited.add(nxt)
                queue.append((nxt, steps + 1))
    
    return n  # Worst case: all 1s

print(numSquares(12))  # 3 (4+4+4)
print(numSquares(13))  # 2 (4+9)
````
