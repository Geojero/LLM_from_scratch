## 🎯 Problem 1: Pascal's Triangle (LC #118)

Generate first `n` rows of Pascal's Triangle.

````python
def generate(numRows):
    """
    Each element = sum of two above it.
    Time: O(n^2), Space: O(n^2)
    """
    triangle = []
    for i in range(numRows):
        row = [1] * (i + 1)
        for j in range(1, i):
            row[j] = triangle[i-1][j-1] + triangle[i-1][j]
        triangle.append(row)
    return triangle

for row in generate(5):
    print(row)
````

## 🎯 Problem 2: Factorial Trailing Zeroes (LC #172)

How many trailing zeroes in `n!`?

**Trailing zero = factor of 10 = factor of 2 × 5.** We always have more 2s than 5s, so count factors of 5.

````python
def trailingZeroes(n):
    """
    Count multiples of 5, 25, 125, ...
    Time: O(log n), Space: O(1)
    """
    count = 0
    while n >= 5:
        n //= 5
        count += n
    return count

print(trailingZeroes(10))   # 2 (10! = 3628800)
print(trailingZeroes(25))   # 6 (25 contributes extra 5)
````
