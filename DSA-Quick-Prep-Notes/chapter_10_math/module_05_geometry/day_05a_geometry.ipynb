## 🎯 Problem 1: Valid Square (LC #593)

Given 4 points, check if they form a square.

**A square has 4 equal sides and 2 equal diagonals.** Compute all 6 pairwise distances. Should get exactly 4 equal shorter distances (sides) and 2 equal longer distances (diagonals).

````python
def validSquare(p1, p2, p3, p4):
    """
    Check all 6 pairwise distances: 4 sides + 2 diagonals.
    Time: O(1), Space: O(1)
    """
    def dist_sq(a, b):
        return (a[0]-b[0])**2 + (a[1]-b[1])**2
    
    points = [p1, p2, p3, p4]
    dists = sorted(dist_sq(points[i], points[j]) 
                   for i in range(4) for j in range(i+1, 4))
    
    # 4 equal sides (non-zero) + 2 equal diagonals
    return (dists[0] > 0 and 
            dists[0] == dists[1] == dists[2] == dists[3] and 
            dists[4] == dists[5])

print(validSquare([0,0],[1,1],[1,0],[0,1]))  # True
````

## 🎯 Problem 2: Rotate Image (LC #48)

Rotate matrix 90° clockwise in-place.

**Two steps:** Transpose (swap `[i][j]` ↔ `[j][i]`), then reverse each row.

````python
def rotate(matrix):
    """
    Transpose + reverse rows = 90° clockwise rotation.
    Time: O(n^2), Space: O(1)
    """
    n = len(matrix)
    
    # Step 1: Transpose
    for i in range(n):
        for j in range(i+1, n):
            matrix[i][j], matrix[j][i] = matrix[j][i], matrix[i][j]
    
    # Step 2: Reverse each row
    for row in matrix:
        row.reverse()

m = [[1,2,3],[4,5,6],[7,8,9]]
rotate(m)
print(m)  # [[7,4,1],[8,5,2],[9,6,3]]
````

## 🎯 Problem 3: Spiral Matrix (LC #54)

Return all elements of matrix in spiral order.

**Track four boundaries:** top, bottom, left, right. Walk each edge, then shrink that boundary.

````python
def spiralOrder(matrix):
    """
    Four-boundary spiral traversal.
    Time: O(m*n), Space: O(1) extra
    """
    result = []
    top, bottom = 0, len(matrix) - 1
    left, right = 0, len(matrix[0]) - 1
    
    while top <= bottom and left <= right:
        # → Go right along top
        for col in range(left, right + 1):
            result.append(matrix[top][col])
        top += 1
        
        # ↓ Go down along right
        for row in range(top, bottom + 1):
            result.append(matrix[row][right])
        right -= 1
        
        # ← Go left along bottom
        if top <= bottom:
            for col in range(right, left - 1, -1):
                result.append(matrix[bottom][col])
            bottom -= 1
        
        # ↑ Go up along left
        if left <= right:
            for row in range(bottom, top - 1, -1):
                result.append(matrix[row][left])
            left += 1
    
    return result

print(spiralOrder([[1,2,3],[4,5,6],[7,8,9]]))  # [1,2,3,6,9,8,7,4,5]
````
