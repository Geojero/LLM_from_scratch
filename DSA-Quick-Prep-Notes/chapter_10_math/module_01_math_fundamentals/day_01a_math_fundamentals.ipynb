## Why Math in Interviews?

Math problems test if you can spot a pattern or formula instead of brute-forcing. The trick is recognizing WHICH mathematical property applies. Most interview math boils down to: GCD/LCM, modular arithmetic, or fast exponentiation.

## The Essentials

````python
from math import gcd

# GCD — Euclidean algorithm
def my_gcd(a, b):
    while b:
        a, b = b, a % b
    return a

# LCM via GCD
def lcm(a, b):
    return a * b // gcd(a, b)

print(f"GCD(12,8) = {my_gcd(12,8)}")   # 4
print(f"LCM(4,6) = {lcm(4,6)}")         # 12

# Modular arithmetic rules
# (a + b) % m = ((a%m) + (b%m)) % m
# (a * b) % m = ((a%m) * (b%m)) % m
MOD = 10**9 + 7
print(f"(10^18) % MOD = {pow(10, 18, MOD)}")  # Python's built-in fast mod pow
````

## 🎯 Problem 1: Pow(x, n) (LC #50)

Calculate `x^n` efficiently.

**Fast exponentiation:** Instead of multiplying `n` times, square repeatedly. `x^10 = (x^5)^2 = (x * x^4)^2 = ...` → O(log n).

````python
def myPow(x, n):
    """
    Binary exponentiation: square and multiply.
    Time: O(log n), Space: O(1)
    """
    if n < 0:
        x, n = 1/x, -n
    
    result = 1
    while n:
        if n & 1:          # Odd power → multiply once
            result *= x
        x *= x             # Square the base
        n >>= 1            # Halve the exponent
    return result

print(myPow(2, 10))    # 1024
print(myPow(2, -2))    # 0.25
````

## 🎯 Problem 2: Happy Number (LC #202)

Replace number with sum of squares of digits. If it reaches 1, it's happy. Otherwise it loops forever.

**This is cycle detection** — use Floyd's (fast/slow pointer) or a set.

````python
def isHappy(n):
    """
    Floyd's cycle detection on digit-square sequence.
    Time: O(log n), Space: O(1)
    """
    def next_num(num):
        total = 0
        while num:
            num, d = divmod(num, 10)
            total += d * d
        return total
    
    slow = n
    fast = next_num(n)
    while fast != 1 and slow != fast:
        slow = next_num(slow)
        fast = next_num(next_num(fast))
    return fast == 1

print(isHappy(19))  # True: 19→82→68→100→1
print(isHappy(2))   # False
````
