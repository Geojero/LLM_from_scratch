## Primes in Interviews

Two things come up: **checking** if a number is prime, and **finding all primes** up to N (Sieve of Eratosthenes). The sieve is the one that impresses interviewers.

---

## 🎯 Problem 1: Count Primes (LC #204)

Count primes less than `n`.

**Sieve of Eratosthenes:** Mark all multiples of each prime starting from 2. What survives is prime. Start marking from `p*p` (smaller multiples already handled).

````python
def countPrimes(n):
    """
    Sieve of Eratosthenes.
    Time: O(n log log n), Space: O(n)
    """
    if n < 2:
        return 0
    
    is_prime = [True] * n
    is_prime[0] = is_prime[1] = False
    
    for p in range(2, int(n**0.5) + 1):
        if is_prime[p]:
            # Mark multiples starting from p*p
            for mult in range(p*p, n, p):
                is_prime[mult] = False
    
    return sum(is_prime)

print(countPrimes(10))   # 4 (primes: 2,3,5,7)
print(countPrimes(30))   # 10
````

## 🎯 Problem 2: Ugly Number II (LC #264)

Find the nth ugly number (prime factors only 2, 3, 5).

**Three-pointer approach:** Maintain pointers for ×2, ×3, ×5. Always pick the smallest next candidate.

````python
def nthUglyNumber(n):
    """
    Three pointers: multiply existing ugly numbers by 2, 3, 5.
    Time: O(n), Space: O(n)
    """
    ugly = [1]
    i2 = i3 = i5 = 0  # Pointers for each factor
    
    for _ in range(n - 1):
        next2 = ugly[i2] * 2
        next3 = ugly[i3] * 3
        next5 = ugly[i5] * 5
        
        nxt = min(next2, next3, next5)
        ugly.append(nxt)
        
        # Advance ALL pointers that match (handles duplicates)
        if nxt == next2: i2 += 1
        if nxt == next3: i3 += 1
        if nxt == next5: i5 += 1
    
    return ugly[-1]

print(nthUglyNumber(10))  # 12 → [1,2,3,4,5,6,8,9,10,12]
````
