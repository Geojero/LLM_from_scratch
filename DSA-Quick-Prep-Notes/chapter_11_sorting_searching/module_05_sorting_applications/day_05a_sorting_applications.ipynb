## When Sorting IS the Solution

Many problems become trivial after sorting. The trick is knowing **what to sort by** and **what to process after**.

## 🎯 Problem 1: Merge Sorted Array (LC #88)

Merge `nums2` into `nums1` (which has extra space at end). In-place.

**Merge from the back** — fill largest elements first so we never overwrite unprocessed elements.

````python
def merge(nums1, m, nums2, n):
    """
    Three pointers merging from back to front.
    Time: O(m+n), Space: O(1)
    """
    i, j, k = m - 1, n - 1, m + n - 1
    
    while j >= 0:
        if i >= 0 and nums1[i] > nums2[j]:
            nums1[k] = nums1[i]
            i -= 1
        else:
            nums1[k] = nums2[j]
            j -= 1
        k -= 1

nums1 = [1,2,3,0,0,0]
merge(nums1, 3, [2,5,6], 3)
print(nums1)  # [1,2,2,3,5,6]
````

## 🎯 Problem 2: Top K Frequent Elements (LC #347)

**Bucket sort by frequency:** Index = frequency, value = list of elements with that frequency. Walk buckets from high to low.

````python
from collections import Counter

def topKFrequent(nums, k):
    """
    Bucket sort: frequency as index.
    Time: O(n), Space: O(n)
    """
    freq = Counter(nums)
    buckets = [[] for _ in range(len(nums) + 1)]
    
    for num, count in freq.items():
        buckets[count].append(num)
    
    result = []
    for i in range(len(buckets) - 1, -1, -1):
        for num in buckets[i]:
            result.append(num)
            if len(result) == k:
                return result
    return result

print(topKFrequent([1,1,1,2,2,3], 2))  # [1,2]
````
