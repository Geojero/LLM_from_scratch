## Why Know These?

You won't use bubble sort in production, but interviewers ask about them to test your understanding of **stability**, **in-place**, and **when to use what**. Also, insertion sort is actually used inside Tim Sort (Python's sort) for small subarrays.

---

## Quick Comparison

```
╔══════════════════════════════════════════════════════════════╗
║  Sort        Time     Space  Stable  Best Case             ║
╠══════════════════════════════════════════════════════════════╣
║  Bubble      O(n²)    O(1)   ✅     O(n) if nearly sorted ║
║  Selection   O(n²)    O(1)   ❌     Always O(n²)          ║
║  Insertion   O(n²)    O(1)   ✅     O(n) if nearly sorted ║
╚══════════════════════════════════════════════════════════════╝
```

---

````python
# Insertion Sort — the most useful of the three
def insertion_sort(arr):
    """Stable, in-place, O(n) for nearly sorted."""
    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1
        while j >= 0 and arr[j] > key:
            arr[j+1] = arr[j]
            j -= 1
        arr[j+1] = key
    return arr

print(insertion_sort([5,2,4,6,1,3]))  # [1,2,3,4,5,6]
````

## 🎯 Problem 1: Sort Colors (LC #75) — Dutch National Flag

Sort array of 0s, 1s, 2s in one pass.

**Three pointers:** `lo` (next 0 position), `hi` (next 2 position), `mid` (current). Swap 0s left, 2s right, skip 1s.

````python
def sortColors(nums):
    """
    Dutch National Flag: lo, mid, hi pointers.
    Time: O(n), Space: O(1)
    """
    lo, mid, hi = 0, 0, len(nums) - 1
    
    while mid <= hi:
        if nums[mid] == 0:
            nums[lo], nums[mid] = nums[mid], nums[lo]
            lo += 1
            mid += 1
        elif nums[mid] == 1:
            mid += 1
        else:  # nums[mid] == 2
            nums[mid], nums[hi] = nums[hi], nums[mid]
            hi -= 1
            # Don't advance mid — new value at mid needs checking

arr = [2,0,2,1,1,0]
sortColors(arr)
print(arr)  # [0,0,1,1,2,2]
````

## 🗺️ Sorting Fundamentals Pattern Map

```
╔═══════════════════════════════════════════════════════════════════════╗
║  O(n²) SORTING ALGORITHMS                                           ║
╠═══════════════════════════════════════════════════════════════════════╣
║                                                                       ║
║  BUBBLE SORT: Swap adjacent pairs, bubble largest to end             ║
║  → Simple but slow. Only use when n is tiny.                        ║
║                                                                       ║
║  SELECTION SORT: Find min, swap to front, repeat                     ║
║  → Always O(n²). Minimizes swaps (good if writes are expensive).    ║
║                                                                       ║
║  INSERTION SORT: Insert each element into correct position           ║
║  → Best for nearly-sorted data. O(n) best case. Stable.            ║
║                                                                       ║
║  DUTCH NATIONAL FLAG (3-way partition):                              ║
║  → Sort Colors (LC #75): low/mid/high pointers, single pass O(n)   ║
║                                                                       ║
║  INTERVIEW RULE: Know these exist, but optimize to O(n log n).      ║
║                                                                       ║
╚═══════════════════════════════════════════════════════════════════════╝
```
