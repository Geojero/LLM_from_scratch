## Breaking the O(n log n) Barrier

Comparison-based sorts can't beat O(n log n). But if you know the data range, you can sort in O(n + k) or O(n) by using the values directly as indices.

---

````python
# Counting Sort — O(n + k) where k = range of values
def counting_sort(arr, max_val):
    count = [0] * (max_val + 1)
    for x in arr:
        count[x] += 1
    
    result = []
    for val, cnt in enumerate(count):
        result.extend([val] * cnt)
    return result

print(counting_sort([4,2,2,8,3,3,1], 8))  # [1,2,2,3,3,4,8]
````

````python
# Bucket Sort — O(n) average for uniform distribution
def bucket_sort(arr):
    if not arr:
        return arr
    
    n = len(arr)
    min_val, max_val = min(arr), max(arr)
    bucket_range = (max_val - min_val) / n + 1
    
    buckets = [[] for _ in range(n)]
    for x in arr:
        idx = int((x - min_val) / bucket_range)
        buckets[idx].append(x)
    
    result = []
    for bucket in buckets:
        result.extend(sorted(bucket))  # Small bucket → insertion sort
    return result

print(bucket_sort([0.42, 0.32, 0.23, 0.52, 0.25, 0.47, 0.51]))
````

## 🎯 Problem: Largest Number (LC #179)

Arrange numbers to form the largest number. E.g., [3,30,34] → "34330".

**Custom comparator:** Compare `a+b` vs `b+a` as strings.

````python
from functools import cmp_to_key

def largestNumber(nums):
    """
    Custom sort: compare concatenations.
    Time: O(n log n * k) where k = avg digit count
    """
    strs = [str(n) for n in nums]
    strs.sort(key=cmp_to_key(lambda a, b: -1 if a+b > b+a else 1))
    
    result = ''.join(strs)
    return '0' if result[0] == '0' else result

print(largestNumber([3,30,34,5,9]))  # "9534330"
````

## 🗺️ Non-Comparison Sorts Pattern Map

```
╔═══════════════════════════════════════════════════════════════════════╗
║  NON-COMPARISON SORTING TECHNIQUES                                   ║
╠═══════════════════════════════════════════════════════════════════════╣
║                                                                       ║
║  COUNTING SORT: Count occurrences → prefix sum → place elements     ║
║  → O(n + k) where k = value range. Only for integers.              ║
║  → Best when k is small relative to n.                              ║
║                                                                       ║
║  BUCKET SORT: Distribute into buckets → sort each → concatenate     ║
║  → O(n) average when data is uniformly distributed                  ║
║  → Great for: floating point, frequency-based grouping              ║
║                                                                       ║
║  RADIX SORT: Sort digit by digit (LSD or MSD) using stable sort    ║
║  → O(d * (n + k)) where d = digits, k = base                       ║
║                                                                       ║
║  CUSTOM COMPARATOR: sorted(key=...) or functools.cmp_to_key        ║
║  → Largest Number (LC #179): compare a+b vs b+a as strings         ║
║                                                                       ║
╚═══════════════════════════════════════════════════════════════════════╝
```
