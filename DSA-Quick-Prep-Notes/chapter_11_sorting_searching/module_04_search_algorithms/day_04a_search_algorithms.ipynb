## Binary Search Isn't Just "Find Target"

The real power of binary search is **finding boundaries** — leftmost/rightmost position where a condition changes. This generalizes to any monotonic predicate.

---

````python
# Binary search template — find leftmost True
def binary_search_left(lo, hi, condition):
    """Find smallest x in [lo, hi] where condition(x) is True."""
    while lo < hi:
        mid = (lo + hi) // 2
        if condition(mid):
            hi = mid       # Answer might be here or left
        else:
            lo = mid + 1   # Answer is to the right
    return lo
````

## 🎯 Problem 1: Find First and Last Position (LC #34)

Find first and last occurrence of target in sorted array.

````python
def searchRange(nums, target):
    """
    Two binary searches: leftmost and rightmost.
    Time: O(log n), Space: O(1)
    """
    def find_left():
        lo, hi = 0, len(nums) - 1
        while lo <= hi:
            mid = (lo + hi) // 2
            if nums[mid] < target:
                lo = mid + 1
            else:
                hi = mid - 1
        return lo
    
    def find_right():
        lo, hi = 0, len(nums) - 1
        while lo <= hi:
            mid = (lo + hi) // 2
            if nums[mid] <= target:
                lo = mid + 1
            else:
                hi = mid - 1
        return hi
    
    left, right = find_left(), find_right()
    return [left, right] if left <= right else [-1, -1]

print(searchRange([5,7,7,8,8,10], 8))  # [3,4]
print(searchRange([5,7,7,8,8,10], 6))  # [-1,-1]
````

## 🎯 Problem 2: Search in Rotated Sorted Array (LC #33)

Array was sorted then rotated. Find target in O(log n).

**Key:** One half is always sorted. Check if target falls in the sorted half.

````python
def search(nums, target):
    """
    Determine which half is sorted, check if target is there.
    Time: O(log n), Space: O(1)
    """
    lo, hi = 0, len(nums) - 1
    
    while lo <= hi:
        mid = (lo + hi) // 2
        if nums[mid] == target:
            return mid
        
        if nums[lo] <= nums[mid]:  # Left half sorted
            if nums[lo] <= target < nums[mid]:
                hi = mid - 1
            else:
                lo = mid + 1
        else:  # Right half sorted
            if nums[mid] < target <= nums[hi]:
                lo = mid + 1
            else:
                hi = mid - 1
    
    return -1

print(search([4,5,6,7,0,1,2], 0))  # 4
````

## 🗺️ Search Algorithms Pattern Map

```
╔═══════════════════════════════════════════════════════════════════════╗
║  SEARCH ALGORITHM PATTERNS                                           ║
╠═══════════════════════════════════════════════════════════════════════╣
║                                                                       ║
║  STANDARD BINARY SEARCH: while left <= right, find exact target     ║
║  → Template: mid = (left+right)//2, return mid or -1                ║
║                                                                       ║
║  BOUNDARY SEARCH: while left < right, find first/last position      ║
║  → bisect_left: first position where target could insert            ║
║  → bisect_right: position after last occurrence                     ║
║                                                                       ║
║  ROTATED ARRAY: One half is always sorted                            ║
║  → Check which half is sorted, decide if target is in that half     ║
║                                                                       ║
║  SEARCH ON ANSWER: Binary search the result space                    ║
║  → Write can_achieve(mid), binary search on feasibility boundary    ║
║                                                                       ║
╚═══════════════════════════════════════════════════════════════════════╝
```
