# 📅 Day 02: Efficient Sorts — Merge Sort & Quick Sort

---

## The Big Two

| | Merge Sort | Quick Sort |
|---|---|---|
| Strategy | Divide, sort, merge | Partition around pivot |
| Time | O(n log n) always | O(n log n) avg, O(n²) worst |
| Space | O(n) | O(log n) stack |
| Stable | ✅ | ❌ |
| When to use | Need stability, linked lists | Arrays, in-place needed |

---

````python
# Merge Sort — guaranteed O(n log n)
def merge_sort(arr):
    if len(arr) <= 1:
        return arr
    
    mid = len(arr) // 2
    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])
    
    # Merge two sorted halves
    result = []
    i = j = 0
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:  # <= for stability
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1
    result.extend(left[i:])
    result.extend(right[j:])
    return result

print(merge_sort([38,27,43,3,9,82,10]))  # sorted
````

````python
import random

# Quick Sort — in-place, randomized pivot
def quick_sort(arr, lo=0, hi=None):
    if hi is None:
        hi = len(arr) - 1
    if lo >= hi:
        return
    
    # Randomize pivot to avoid O(n²) on sorted input
    pivot_idx = random.randint(lo, hi)
    arr[pivot_idx], arr[hi] = arr[hi], arr[pivot_idx]
    pivot = arr[hi]
    
    # Lomuto partition
    i = lo
    for j in range(lo, hi):
        if arr[j] < pivot:
            arr[i], arr[j] = arr[j], arr[i]
            i += 1
    arr[i], arr[hi] = arr[hi], arr[i]
    
    quick_sort(arr, lo, i - 1)
    quick_sort(arr, i + 1, hi)

a = [3,6,8,10,1,2,1]
quick_sort(a)
print(a)  # [1,1,2,3,6,8,10]
````

## 🎯 Problem: Kth Largest Element (LC #215)

Find kth largest without fully sorting.

**Quick Select** — partition like quicksort but only recurse into the half containing the answer. Average O(n).

````python
def findKthLargest(nums, k):
    """
    Quick Select: partition + recurse only one side.
    Time: O(n) avg, O(n²) worst. Space: O(1)
    """
    k = len(nums) - k  # Convert to kth smallest (0-indexed)
    
    def quick_select(lo, hi):
        pivot_idx = random.randint(lo, hi)
        nums[pivot_idx], nums[hi] = nums[hi], nums[pivot_idx]
        pivot = nums[hi]
        
        i = lo
        for j in range(lo, hi):
            if nums[j] < pivot:
                nums[i], nums[j] = nums[j], nums[i]
                i += 1
        nums[i], nums[hi] = nums[hi], nums[i]
        
        if i == k: return nums[i]
        elif i < k: return quick_select(i + 1, hi)
        else: return quick_select(lo, i - 1)
    
    return quick_select(0, len(nums) - 1)

print(findKthLargest([3,2,1,5,6,4], 2))  # 5
````

## 🗺️ Efficient Sorts Pattern Map

```
╔═══════════════════════════════════════════════════════════════════════╗
║  O(n log n) SORTING & QUICK SELECT                                   ║
╠═══════════════════════════════════════════════════════════════════════╣
║                                                                       ║
║  MERGE SORT: Split → sort halves → merge                             ║
║  → O(n log n) guaranteed, stable, O(n) extra space                  ║
║  → Best for: linked lists, external sort, stable sort needed        ║
║                                                                       ║
║  QUICK SORT: Pick pivot → partition → recurse                        ║
║  → O(n log n) average, O(n²) worst, in-place, unstable             ║
║  → Best for: arrays (cache-friendly), when space matters            ║
║  → Always use RANDOM PIVOT to avoid worst case                      ║
║                                                                       ║
║  QUICK SELECT: Partition once → recurse on ONE side                  ║
║  → O(n) average for kth element — don't fully sort!                 ║
║  → When to use: "find kth largest/smallest"                         ║
║                                                                       ║
╚═══════════════════════════════════════════════════════════════════════╝
```
