```
╔═══════════════════════════════════════════════════════════════════╗
║  FILTERING TOOLKIT                                               ║
╠═══════════════════════════════════════════════════════════════════╣
║                                                                   ║
║  Boolean:   df[df['col'] > val]                                 ║
║  Multiple:  df[(cond1) & (cond2)]       ← must use &, |, ~     ║
║  Query:     df.query('col > val and col2 == "x"')             ║
║  isin:      df[df['col'].isin([v1, v2])]                       ║
║  between:   df[df['col'].between(lo, hi)]                      ║
║  str:       df[df['col'].str.contains('pattern')]              ║
║  dt:        df[df['date'].dt.year == 2024]                     ║
║                                                                   ║
║  ⚠️  Use &, |, ~ instead of and, or, not                       ║
║  ⚠️  Always wrap conditions in parentheses                      ║
║                                                                   ║
╚═══════════════════════════════════════════════════════════════════╝
```

````python
import pandas as pd
import numpy as np

# Sample dataset
df = pd.DataFrame({
    'name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve', 'Frank', 'Grace'],
    'age': [25, 30, 35, 28, 32, 45, 22],
    'salary': [50000, 60000, 70000, 55000, 65000, 80000, 42000],
    'dept': ['Engineering', 'Sales', 'Engineering', 'HR', 'Sales', 'Engineering', 'HR'],
    'city': ['New York', 'San Francisco', 'New York', 'Chicago', 'San Francisco', 'New York', 'Chicago'],
    'start_date': pd.to_datetime(['2020-01-15', '2019-06-01', '2018-03-20', 
                                   '2021-09-10', '2020-11-05', '2017-02-28', '2022-07-01'])
})
print(df)
print(f"\ndtypes:\n{df.dtypes}")
````

````python
# ============================================================
# 1. Boolean Indexing — The Foundation
# ============================================================

# Single condition
high_salary = df[df['salary'] > 60000]
print(f"Salary > 60k:\n{high_salary[['name', 'salary']]}\n")

# Multiple conditions: AND (&) — wrap each in ()
result = df[(df['salary'] > 50000) & (df['dept'] == 'Engineering')]
print(f"Eng + salary > 50k:\n{result[['name', 'salary', 'dept']]}\n")

# OR (|)
result = df[(df['dept'] == 'HR') | (df['age'] > 35)]
print(f"HR or age > 35:\n{result[['name', 'age', 'dept']]}\n")

# NOT (~)
result = df[~(df['dept'] == 'Sales')]
print(f"Not Sales:\n{result[['name', 'dept']]}")
````

````python
# ============================================================
# 2. .query() — Readable String-Based Filtering
# ============================================================

# Same queries, more readable
print(df.query('salary > 60000')[['name', 'salary']])
print()
print(df.query('salary > 50000 and dept == "Engineering"')[['name', 'salary']])
print()

# Use variables with @
min_sal = 55000
print(df.query('salary >= @min_sal')[['name', 'salary']])
print()

# Chained conditions
target_depts = ['Engineering', 'HR']
print(df.query('dept in @target_depts and age < 30')[['name', 'age', 'dept']])

print("\n💡 .query() is cleaner for complex conditions, especially in pipelines")
````

````python
# ============================================================
# 3. .isin() and .between()
# ============================================================

# isin — check membership in a list
target_cities = ['New York', 'Chicago']
result = df[df['city'].isin(target_cities)]
print(f"In NY or Chicago:\n{result[['name', 'city']]}\n")

# NOT isin
result = df[~df['city'].isin(target_cities)]
print(f"Not NY/Chicago:\n{result[['name', 'city']]}\n")

# between — range check (inclusive)
result = df[df['age'].between(25, 32)]
print(f"Age 25-32:\n{result[['name', 'age']]}\n")

# between with dates
result = df[df['start_date'].between('2020-01-01', '2021-12-31')]
print(f"Started 2020-2021:\n{result[['name', 'start_date']]}")
````

````python
# ============================================================
# 4. .str Accessor — String Operations on Columns
# ============================================================

# contains (partial match)
result = df[df['city'].str.contains('New')]
print(f"City contains 'New':\n{result[['name', 'city']]}\n")

# startswith / endswith
result = df[df['name'].str.startswith('A')]
print(f"Name starts with A:\n{result[['name']]}\n")

# Case-insensitive search
result = df[df['dept'].str.lower().str.contains('eng')]
print(f"Dept contains 'eng':\n{result[['name', 'dept']]}\n")

# String methods
print(f"Name lengths: {df['name'].str.len().tolist()}")
print(f"Upper names:  {df['name'].str.upper().tolist()}")
print(f"First 3 chars: {df['name'].str[:3].tolist()}")

# Split
df_temp = pd.DataFrame({'full_name': ['Alice Smith', 'Bob Jones']})
df_temp[['first', 'last']] = df_temp['full_name'].str.split(' ', expand=True)
print(f"\nSplit:\n{df_temp}")
````

````python
# ============================================================
# 5. .dt Accessor — Datetime Operations
# ============================================================

# Extract components
print(f"Year:   {df['start_date'].dt.year.tolist()}")
print(f"Month:  {df['start_date'].dt.month.tolist()}")
print(f"Day:    {df['start_date'].dt.day.tolist()}")
print(f"DayName:{df['start_date'].dt.day_name().tolist()}")
print(f"Quarter:{df['start_date'].dt.quarter.tolist()}")

# Filter by datetime components
result = df[df['start_date'].dt.year == 2020]
print(f"\nStarted in 2020:\n{result[['name', 'start_date']]}\n")

# Days since start
df['tenure_days'] = (pd.Timestamp.now() - df['start_date']).dt.days
print(f"Tenure:\n{df[['name', 'tenure_days']].sort_values('tenure_days', ascending=False)}")

# Clean up
df = df.drop(columns=['tenure_days'])
````

````python
# ============================================================
# 6. 🔥 10 Filtering Challenges — Try Before Looking!
# ============================================================

# Challenge 1: Engineers in New York
c1 = df[(df['dept'] == 'Engineering') & (df['city'] == 'New York')]
print(f"1. Eng + NY: {c1['name'].tolist()}")

# Challenge 2: Salary between 50k-65k, not in HR
c2 = df[(df['salary'].between(50000, 65000)) & (df['dept'] != 'HR')]
print(f"2. 50-65k, not HR: {c2['name'].tolist()}")

# Challenge 3: Name contains 'a' (case-insensitive)
c3 = df[df['name'].str.lower().str.contains('a')]
print(f"3. Name has 'a': {c3['name'].tolist()}")

# Challenge 4: Top 3 earners
c4 = df.nlargest(3, 'salary')
print(f"4. Top 3 earners: {c4['name'].tolist()}")

# Challenge 5: Started on a weekday (Mon-Fri)
c5 = df[df['start_date'].dt.dayofweek < 5]
print(f"5. Started weekday: {c5['name'].tolist()}")

# Challenge 6: Above average salary
c6 = df[df['salary'] > df['salary'].mean()]
print(f"6. Above avg salary: {c6['name'].tolist()}")

# Challenge 7: Youngest person in each department
c7 = df.loc[df.groupby('dept')['age'].idxmin()]
print(f"7. Youngest per dept: {c7[['name', 'dept', 'age']].values.tolist()}")

# Challenge 8: People in cities with more than 1 employee
city_counts = df['city'].value_counts()
big_cities = city_counts[city_counts > 1].index
c8 = df[df['city'].isin(big_cities)]
print(f"8. Multi-employee cities: {c8['name'].tolist()}")

# Challenge 9: Started in Q1 (Jan-Mar) of any year
c9 = df[df['start_date'].dt.quarter == 1]
print(f"9. Q1 starters: {c9['name'].tolist()}")

# Challenge 10: Salary rank within department
df['dept_rank'] = df.groupby('dept')['salary'].rank(ascending=False).astype(int)
print(f"10. Dept salary rank:\n{df[['name', 'dept', 'salary', 'dept_rank']]}")
df = df.drop(columns=['dept_rank'])
````

## 🗺️ Filtering Pattern Map

```
╔═══════════════════════════════════════════════════════════════════╗
║  FILTERING — Pick the Right Tool                                 ║
╠═══════════════════════════════════════════════════════════════════╣
║                                                                   ║
║  Simple compare    → df[df['col'] > val]                       ║
║  Multi-condition   → df[(c1) & (c2)]  or  df.query('...')      ║
║  Membership list   → df[df['col'].isin(list)]                  ║
║  Range             → df[df['col'].between(lo, hi)]             ║
║  String pattern    → df[df['col'].str.contains('pat')]         ║
║  Date component    → df[df['col'].dt.year == 2024]             ║
║  Top-N             → df.nlargest(n, 'col')                     ║
║  Above average     → df[df['col'] > df['col'].mean()]         ║
║  Per-group min/max → df.loc[df.groupby('g')['c'].idxmin()]    ║
║                                                                   ║
╚═══════════════════════════════════════════════════════════════════╝
```

## 📝 Interview Questions

| # | Question | Key Points |
|---|----------|-----------|
| 1 | & vs and in Pandas filtering? | Use & (element-wise). 'and' is Python's scalar boolean, doesn't work with Series |
| 2 | .query() vs boolean indexing? | query() is cleaner for complex conditions, supports @ for variables. Boolean indexing is more flexible |
| 3 | How to filter rows with string patterns? | .str.contains(), .str.startswith(), .str.match() for regex |
| 4 | Find above-average salary per dept? | df[df['salary'] > df.groupby('dept')['salary'].transform('mean')] |
| 5 | .isin() vs .between()? | isin = membership in list, between = range check (inclusive both ends) |
