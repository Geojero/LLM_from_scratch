```
╔═══════════════════════════════════════════════════════════════════╗
║  PANDAS — TWO DATA STRUCTURES                                    ║
╠═══════════════════════════════════════════════════════════════════╣
║                                                                   ║
║  Series   = 1D labeled array       (like a column)              ║
║  DataFrame = 2D labeled table       (dict of Series)            ║
║                                                                   ║
║  .loc[row_label, col_label]  → label-based (INCLUSIVE end)      ║
║  .iloc[row_idx, col_idx]     → integer-based (EXCLUSIVE end)    ║
║                                                                   ║
║  Memory trick:                                                    ║
║    loc  = Labels, On, Columns                                   ║
║    iloc = Integer LOCation                                       ║
║                                                                   ║
╚═══════════════════════════════════════════════════════════════════╝
```

````python
import pandas as pd
import numpy as np

# ============================================================
# 1. Series Creation
# ============================================================

# From list
s = pd.Series([10, 20, 30], index=['a', 'b', 'c'], name='scores')
print(f"Series:\n{s}\n")

# From dict
s2 = pd.Series({'math': 90, 'english': 85, 'science': 92})
print(f"From dict:\n{s2}\n")

# Access
print(f"s['a'] = {s['a']}")
print(f"s.values = {s.values}")
print(f"s.index = {s.index.tolist()}")
````

````python
# ============================================================
# 2. DataFrame Creation (4 ways)
# ============================================================

# Way 1: From dict of lists
df = pd.DataFrame({
    'name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve'],
    'age': [25, 30, 35, 28, 32],
    'salary': [50000, 60000, 70000, 55000, 65000],
    'dept': ['Eng', 'Sales', 'Eng', 'HR', 'Sales']
})
print(f"From dict:\n{df}\n")

# Way 2: From list of dicts
df2 = pd.DataFrame([{'x': 1, 'y': 2}, {'x': 3, 'y': 4}])
print(f"From list of dicts:\n{df2}\n")

# Way 3: From NumPy array
df3 = pd.DataFrame(np.random.randn(3, 2), columns=['A', 'B'])
print(f"From NumPy:\n{df3}\n")

# Way 4: From CSV (most common in practice)
# df = pd.read_csv('data.csv')
# df = pd.read_csv('data.csv', index_col=0, parse_dates=['date'])
````

````python
# ============================================================
# 3. Inspection — First Things You Do With Any DataFrame
# ============================================================

print(f"Shape: {df.shape}")           # (5, 4)
print(f"Columns: {df.columns.tolist()}")
print(f"dtypes:\n{df.dtypes}\n")

# info() — types, non-null counts, memory
print("=== df.info() ===")
df.info()

# describe() — stats for numeric columns
print(f"\n=== df.describe() ===")
print(df.describe())

# describe for ALL columns (include='all')
print(f"\n=== describe(include='all') ===")
print(df.describe(include='all'))
````

````python
# ============================================================
# 4. Quick Peek Methods
# ============================================================

print(df.head(2))       # First 2 rows
print()
print(df.tail(2))       # Last 2 rows
print()
print(df.sample(2))     # Random 2 rows

# Value counts — frequency of each unique value
print(f"\nDept counts:\n{df['dept'].value_counts()}")

# Unique values
print(f"\nUnique depts: {df['dept'].unique()}")
print(f"Number unique: {df['dept'].nunique()}")
````

````python
# ============================================================
# 5. .loc vs .iloc — THE MOST IMPORTANT DISTINCTION
# ============================================================

df_indexed = df.set_index('name')
print(f"Indexed DataFrame:\n{df_indexed}\n")

# .loc — Label-based (inclusive on BOTH ends)
print(f"loc['Alice']:       \n{df_indexed.loc['Alice']}\n")
print(f"loc['Alice':'Charlie']:  \n{df_indexed.loc['Alice':'Charlie']}\n")  # inclusive!
print(f"loc['Alice', 'salary']: {df_indexed.loc['Alice', 'salary']}\n")

# .iloc — Integer position-based (exclusive end like Python)
print(f"iloc[0]:            \n{df_indexed.iloc[0]}\n")
print(f"iloc[0:2]:          \n{df_indexed.iloc[0:2]}\n")  # exclusive!
print(f"iloc[0, 1]:         {df_indexed.iloc[0, 1]}\n")   # row 0, col 1

# Multiple rows and columns
print(f"loc with list:\n{df_indexed.loc[['Alice', 'Eve'], ['age', 'salary']]}")
````

````python
# ============================================================
# 6. Column Operations
# ============================================================

# Select single column → Series
ages = df['age']
print(f"Type: {type(ages).__name__}")

# Select multiple columns → DataFrame
subset = df[['name', 'salary']]
print(f"Type: {type(subset).__name__}\n")

# Add column
df['bonus'] = df['salary'] * 0.1

# Rename columns
df_renamed = df.rename(columns={'name': 'employee', 'dept': 'department'})
print(f"Renamed columns: {df_renamed.columns.tolist()}")

# Drop column
df_dropped = df.drop(columns=['bonus'])
print(f"After drop: {df_dropped.columns.tolist()}")

# Change dtypes
df['age'] = df['age'].astype(float)
print(f"\nAge dtype: {df['age'].dtype}")
````

````python
# ============================================================
# 7. Index Operations
# ============================================================

# set_index — make a column the index
df_idx = df.set_index('name')
print(f"After set_index:\n{df_idx.head(2)}\n")

# reset_index — move index back to column
df_reset = df_idx.reset_index()
print(f"After reset_index:\n{df_reset.head(2)}\n")

# Sort by index
df_sorted = df_idx.sort_index(ascending=False)
print(f"Sorted by index:\n{df_sorted.head(3)}\n")

# Sort by values
df_sorted2 = df.sort_values('salary', ascending=False)
print(f"Sorted by salary:\n{df_sorted2[['name', 'salary']]}")
````

````python
# ============================================================
# 8. Quick Aggregations
# ============================================================

print(f"Mean salary:   {df['salary'].mean():.0f}")
print(f"Median salary: {df['salary'].median():.0f}")
print(f"Sum salary:    {df['salary'].sum()}")
print(f"Min/Max age:   {df['age'].min()} / {df['age'].max()}")
print(f"Std salary:    {df['salary'].std():.0f}")

# idxmin / idxmax — index of min/max value
print(f"\nHighest salary index: {df['salary'].idxmax()}")
print(f"Highest salary name:  {df.loc[df['salary'].idxmax(), 'name']}")

# Multiple aggs at once
print(f"\nMultiple aggs:\n{df[['age', 'salary']].agg(['mean', 'std', 'min', 'max'])}")
````

