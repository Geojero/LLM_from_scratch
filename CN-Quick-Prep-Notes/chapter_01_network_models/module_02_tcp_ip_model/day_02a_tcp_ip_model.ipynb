```
╔══════════════════════════════════════════════════════════════════════╗
║            OSI (7 Layers)   vs   TCP/IP (4 Layers)                  ║
╠══════════════════════════════════════════════════════════════════════╣
║                                                                      ║
║   7. Application  ─┐                                                ║
║   6. Presentation  ├──→  4. Application    (HTTP, DNS, SMTP, FTP)   ║
║   5. Session       ─┘                                                ║
║                                                                      ║
║   4. Transport     ────→  3. Transport      (TCP, UDP)              ║
║                                                                      ║
║   3. Network       ────→  2. Internet        (IP, ICMP, ARP)        ║
║                                                                      ║
║   2. Data Link     ─┐                                                ║
║   1. Physical      ─┴──→  1. Network Access  (Ethernet, WiFi)       ║
║                                                                      ║
╠══════════════════════════════════════════════════════════════════════╣
║  OSI = theoretical reference model (ISO standard)                   ║
║  TCP/IP = practical implementation (what the internet uses!)        ║
╚══════════════════════════════════════════════════════════════════════╝
```

````python
# ============================================
# 1. TCP/IP Layer Simulator
# ============================================

class TCPIPStack:
    """Simulate data flow through 4-layer TCP/IP model."""
    
    LAYERS = [
        (4, "Application",    "HTTP Header"),
        (3, "Transport",      "TCP Header (src_port, dst_port, seq, ack)"),
        (2, "Internet",       "IP Header (src_ip, dst_ip, TTL)"),
        (1, "Network Access", "Ethernet Header (src_mac, dst_mac) + Trailer (CRC)"),
    ]
    
    def send(self, message, src_ip, dst_ip, src_port, dst_port):
        print(f"  ═══ SENDING: '{message}' ═══\n")
        
        # Layer 4: Application
        app_data = f"GET / HTTP/1.1\r\nHost: example.com\r\n\r\n{message}"
        print(f"  L4 Application:  HTTP request created")
        print(f"     {app_data[:50]}...\n")
        
        # Layer 3: Transport
        segment = {
            "src_port": src_port,
            "dst_port": dst_port,
            "seq": 1001,
            "ack": 0,
            "flags": "PSH,ACK",
            "payload": app_data,
        }
        print(f"  L3 Transport:    TCP segment")
        print(f"     src_port={src_port} dst_port={dst_port} seq=1001 flags=PSH,ACK\n")
        
        # Layer 2: Internet
        packet = {
            "src_ip": src_ip,
            "dst_ip": dst_ip,
            "ttl": 64,
            "protocol": "TCP",
            "payload": segment,
        }
        print(f"  L2 Internet:     IP packet")
        print(f"     {src_ip} → {dst_ip} TTL=64 proto=TCP\n")
        
        # Layer 1: Network Access
        frame = {
            "src_mac": "AA:BB:CC:DD:EE:01",
            "dst_mac": "AA:BB:CC:DD:EE:02",
            "type": "0x0800",  # IPv4
            "payload": packet,
            "crc": "0xDEADBEEF",
        }
        print(f"  L1 Network Access: Ethernet frame")
        print(f"     src_mac={frame['src_mac']} dst_mac={frame['dst_mac']}")
        print(f"     type=0x0800 (IPv4) crc={frame['crc']}\n")
        
        print("  📡 Transmitted as electrical/optical signals on wire!")
        return frame

stack = TCPIPStack()
stack.send("Hello!", "192.168.1.10", "93.184.216.34", 54321, 80)
````

````python
# ============================================
# 2. OSI vs TCP/IP — Detailed Comparison
# ============================================

print("""
  ┌─────────────────┬───────────────────────┬───────────────────────┐
  │ Feature          │ OSI Model             │ TCP/IP Model          │
  ├─────────────────┼───────────────────────┼───────────────────────┤
  │ Layers           │ 7                     │ 4                     │
  │ Developed by     │ ISO                   │ DARPA (DoD)           │
  │ Approach         │ Theory first          │ Implementation first  │
  │ Protocol-dep.    │ Protocol-independent  │ Protocol-dependent    │
  │ Adoption         │ Reference model       │ Actual internet       │
  │ Session/Pres.    │ Separate layers       │ Merged into App       │
  │ Network Access   │ 2 layers (DL + Phy)   │ 1 layer               │
  │ Real-world use   │ Teaching/certification│ Production networks   │
  └─────────────────┴───────────────────────┴───────────────────────┘

  Why TCP/IP won:
  1. Built FIRST, standardized AFTER (practical > theoretical)
  2. Free and open (OSI was ISO bureaucracy)
  3. BSD Unix shipped TCP/IP stack (instant adoption)
  4. Simpler (4 layers vs 7)
  5. DARPA funded it → US military → internet
""")
````

````python
# ============================================
# 3. Protocol Suite Map
# ============================================

protocol_map = {
    "Application": {
        "Web":     ["HTTP (80)", "HTTPS (443)"],
        "Email":   ["SMTP (25)", "POP3 (110)", "IMAP (143)"],
        "File":    ["FTP (20/21)", "SFTP (22)", "TFTP (69)"],
        "DNS":     ["DNS (53)"],
        "Remote":  ["SSH (22)", "Telnet (23)", "RDP (3389)"],
        "Other":   ["DHCP (67/68)", "SNMP (161)", "NTP (123)"],
    },
    "Transport": {
        "Reliable":   ["TCP — ordered, reliable, connection-oriented"],
        "Fast":       ["UDP — unordered, unreliable, connectionless"],
    },
    "Internet": {
        "Addressing": ["IPv4 (32-bit)", "IPv6 (128-bit)"],
        "Routing":    ["OSPF", "BGP", "RIP"],
        "Support":    ["ICMP (ping)", "ARP (IP→MAC)", "RARP (MAC→IP)"],
        "Multicast":  ["IGMP"],
    },
    "Network Access": {
        "Wired":      ["Ethernet (802.3)"],
        "Wireless":   ["WiFi (802.11)", "Bluetooth"],
        "WAN":        ["PPP", "HDLC", "Frame Relay"],
    },
}

print("  ═══ TCP/IP Protocol Suite ═══\n")
for layer, categories in protocol_map.items():
    print(f"  [{layer}]")
    for category, protocols in categories.items():
        print(f"    {category:12}: {', '.join(protocols)}")
    print()
````

## 📝 Interview Questions

| # | Question | Key Point |
|---|----------|------------|
| 1 | OSI vs TCP/IP? | OSI=7 layers (theory), TCP/IP=4 layers (practice). TCP/IP merges L5-7 into Application |
| 2 | Why TCP/IP won over OSI? | Built first, free, shipped with Unix, simpler, funded by DARPA |
| 3 | Map HTTP to both models | OSI: L7 Application. TCP/IP: L4 Application. Same protocol, different model |
| 4 | Port 80 vs 443? | 80=HTTP (unencrypted), 443=HTTPS (TLS encrypted) |
| 5 | What is an ephemeral port? | Dynamic port (49152-65535) assigned by OS for client-side connections |
