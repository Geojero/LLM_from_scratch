### 🎭 The Analogy

Sending a letter through the postal system:
- **Application**: You write the letter (HTTP, FTP)
- **Presentation**: You translate it to English (encryption, compression)
- **Session**: You start and end the conversation (session management)
- **Transport**: You choose registered mail vs regular (TCP vs UDP)
- **Network**: The post office routes it city-to-city (IP, routers)
- **Data Link**: The mail truck delivers street-to-street (MAC, switches)
- **Physical**: The actual road/wire carrying the truck (cables, signals)

```
╠══════════════════════════════════════════════════════════════════════╣

║  ──────┼───────────────┼──────────┼────────────────────┼────────── ║

  Mnemonic (top→bottom): All People Seem To Need Data Processing
  Mnemonic (bottom→top): Please Do Not Throw Sausage Pizza Away
```

````python
# ============================================
# 1. OSI Layer Simulator
# ============================================

class OSILayer:
    def __init__(self, number, name, pdu, protocols, devices):
        self.number = number
        self.name = name
        self.pdu = pdu
        self.protocols = protocols
        self.devices = devices
    
    def encapsulate(self, data):
        header = f"[L{self.number}-HDR]"
        return f"{header}{data}"
    
    def decapsulate(self, data):
        header = f"[L{self.number}-HDR]"
        return data.replace(header, "", 1)

# Build the 7 layers
layers = [
    OSILayer(7, "Application",   "Data",    ["HTTP","FTP","SMTP","DNS"],    []),
    OSILayer(6, "Presentation",  "Data",    ["SSL/TLS","JPEG","ASCII"],     []),
    OSILayer(5, "Session",       "Data",    ["NetBIOS","RPC"],              []),
    OSILayer(4, "Transport",     "Segment", ["TCP","UDP"],                  []),
    OSILayer(3, "Network",       "Packet",  ["IP","ICMP","ARP"],            ["Router"]),
    OSILayer(2, "Data Link",     "Frame",   ["Ethernet","WiFi","PPP"],      ["Switch","Bridge"]),
    OSILayer(1, "Physical",      "Bits",    ["USB","Bluetooth","Fiber"],    ["Hub","Cable"]),
]

# SENDING: Top → Bottom (encapsulation)
print("  ═══ SENDING: Encapsulation (Layer 7 → Layer 1) ═══\n")
data = '"Hello, World!"'
print(f"    Original data: {data}\n")

for layer in layers:
    data = layer.encapsulate(data)
    print(f"    Layer {layer.number} ({layer.name:12}): {data}")
    print(f"      PDU: {layer.pdu:8} | Protocols: {', '.join(layer.protocols)}")

print(f"\n  📡 Transmitted on wire as: {data}")
````

````python
# RECEIVING: Bottom → Top (decapsulation)
received = data  # from sender
print("  ═══ RECEIVING: Decapsulation (Layer 1 → Layer 7) ═══\n")
print(f"    Received on wire: {received}\n")

for layer in reversed(layers):
    received = layer.decapsulate(received)
    print(f"    Layer {layer.number} ({layer.name:12}): strips header → {received}")

print(f"\n  ✅ Application receives: {received}")
````

````python
# ============================================
# 2. Layer Deep Dive — What Each Layer Does
# ============================================

layer_details = {
    7: {
        "name": "Application",
        "does": "User-facing interface. HTTP for web, SMTP for email, FTP for files.",
        "key": "Closest to the user. Does NOT include the browser itself.",
        "interview": "HTTP, DNS, FTP operate here. Browser is NOT a layer-7 entity.",
    },
    6: {
        "name": "Presentation",
        "does": "Data translation, encryption/decryption, compression.",
        "key": "Converts between formats: JPEG, ASCII, encryption (SSL/TLS).",
        "interview": "SSL/TLS encryption happens here (or debatably at L4/L5).",
    },
    5: {
        "name": "Session",
        "does": "Establishes, manages, terminates sessions between apps.",
        "key": "Session checkpointing, synchronization, dialog control.",
        "interview": "In practice, TCP handles most session duties. L5 is thin.",
    },
    4: {
        "name": "Transport",
        "does": "End-to-end delivery. Segmentation, flow control, error recovery.",
        "key": "TCP (reliable, ordered) vs UDP (fast, unreliable). Port numbers.",
        "interview": "TCP 3-way handshake, port numbers, segment = L4 PDU.",
    },
    3: {
        "name": "Network",
        "does": "Logical addressing (IP), routing between networks.",
        "key": "Router operates here. IP addresses. Best-path selection.",
        "interview": "IP, ICMP (ping), ARP (IP→MAC), OSPF, BGP.",
    },
    2: {
        "name": "Data Link",
        "does": "Physical addressing (MAC), frame creation, error detection.",
        "key": "Switch operates here. MAC addresses. CRC for error check.",
        "interview": "LLC + MAC sublayers. Ethernet frame structure.",
    },
    1: {
        "name": "Physical",
        "does": "Bit transmission over physical medium (electrical/optical/radio).",
        "key": "Hub operates here. Cables, connectors, voltage, bit rate.",
        "interview": "No intelligence. Just sends raw bits. Hub = L1 device.",
    },
}

for num in range(7, 0, -1):
    d = layer_details[num]
    print(f"  Layer {num}: {d['name']}")
    print(f"    Does: {d['does']}")
    print(f"    Key:  {d['key']}")
    print(f"    💼:   {d['interview']}")
    print()
````

````python
# ============================================
# 3. "Which Layer?" Interview Drill
# ============================================

quiz = [
    ("HTTP request",          7, "Application — web protocol"),
    ("Encryption (TLS)",      6, "Presentation — data format/security"),
    ("TCP 3-way handshake",   4, "Transport — connection setup"),
    ("IP routing",            3, "Network — logical addressing"),
    ("MAC address lookup",    2, "Data Link — physical addressing"),
    ("Ethernet frame CRC",    2, "Data Link — error detection"),
    ("Fiber optic signal",    1, "Physical — bit transmission"),
    ("Port number 443",       4, "Transport — port = L4 concept"),
    ("Router forwarding",     3, "Network — routers work at L3"),
    ("Switch forwarding",     2, "Data Link — switches work at L2"),
    ("DNS query",             7, "Application — DNS is an app protocol"),
    ("ARP resolution",        3, "Network (or L2.5) — IP to MAC"),
    ("Ping (ICMP)",           3, "Network — ICMP operates at L3"),
    ("JPEG compression",      6, "Presentation — data format"),
    ("Hub broadcasting",      1, "Physical — hub = L1 device"),
]

print("  ═══ INTERVIEW DRILL: Which Layer? ═══\n")
for concept, layer, explanation in quiz:
    print(f"    {concept:25} → Layer {layer} — {explanation}")
````

## 📝 Interview Questions

| # | Question | Key Point |
|---|----------|------------|
| 1 | Name all 7 OSI layers | Physical, Data Link, Network, Transport, Session, Presentation, Application |
| 2 | What layer does a router operate at? | Layer 3 (Network) — uses IP addresses for routing |
| 3 | Difference between switch and hub? | Hub (L1): broadcasts to all. Switch (L2): uses MAC table, forwards to specific port |
| 4 | What is the PDU at each layer? | L4=Segment, L3=Packet, L2=Frame, L1=Bits |
| 5 | Where does encryption happen? | Presentation (L6) or Transport (L4) — TLS straddles both |
